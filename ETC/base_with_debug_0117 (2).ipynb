```python
"""
Main Engine (Threshold-free Stability/Transition Engine)
- Dataset: MHEALTH (.log)
- Multi-sensor/modalities with missing groups supported
- Detector: per-group steady score s_t in [0,1] (threshold-free)
- Sensor fusion: self-consistency quality -> softmax weights
- Model: transition probability sequence + latent + reconstruction
- Training: recon + soft transition supervision + pair consistency contrast + inertial invariance (optional)
- (Optional) Count K supervision is included as a hook, but NOT used at test-time.

✅ Patch 핵심 (0.5 몰림 제거):
- "윈도우마다" c->robust_zscore->s 를 계산하지 않고,
- "block(trial) 전체"에서 c_full->robust_zscore->s_full을 한 번 만든 뒤,
- window는 s_full[st:ed]만 slice해서 사용.

✅ 추가 패치 (detector_use_acc=False KeyError 해결):
- "모델 입력 groups" (cfg["groups"]) 과
- "detector/fusion groups" (det_groups: gyro-only 옵션) 를 분리
- block_cache는 det_groups로만 만들고,
- window에서 detector는 det_groups만, 모델 입력 x는 cfg["groups"] 전체를 사용
"""

import os, glob, random
import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from scipy.signal import savgol_filter
import matplotlib.pyplot as plt
from collections import defaultdict

# =========================================================
# 0) CONFIG
# =========================================================
CONFIG = {
    # data
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "target_activities": [6, 7, 12],   # waist bends / arm elevation / jump
    "fs": 50,

    # windowing
    "window_size": 100,
    "stride": 50,

    # groups (missing-friendly)
    "groups": ["chest_acc", "ankle_acc", "arm_acc", "ankle_gyro", "arm_gyro"],

    # training
    "batch_size": 64,
    "epochs": 20,
    "lr": 1e-3,
    "seed": 42,

    # model
    "latent_dim": 64,
    "hidden_dim": 128,

    # losses
    "lambda_recon": 1.0,
    "lambda_soft": 0.5,        # soft supervision: p_hat(t) ~= 1 - s_det(t)
    "lambda_pair": 0.3,        # consistent/inconsistent pair contrast
    "lambda_inertial": 0.2,    # inertial invariance

    # detector smoothing
    "det_savgol": True,
    "det_savgol_win": 11,
    "det_savgol_poly": 2,

    # detector robust normalization
    "mad_eps": 1e-6,
    "sigmoid_tau": 1.5,
    "sigmoid_ref_q": 80,     # 기준 quantile (80~95 추천)
    "mad_floor": 1e-3,       # gyro처럼 MAD 너무 작을 때 폭주 방지

    # fusion quality (self-consistency)
    "quality_quantile": 0.2,
    "quality_pairs": 128,
    "quality_tau": 1.0,

    # pair loss sampling
    "pair_delta": 5,
    "pair_margin": 0.2,
    'pair_q': 0.2,

    'detector_use_acc': True,   # set False only for Act12 1-run check
    'pair_delta': 5,
    'pair_thr_hi': 0.70,  # transition if y >= thr_hi
    'pair_thr_lo': 0.30,  # steady if y <= thr_lo
}

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


set_seed(CONFIG["seed"])


# =========================================================
# 1) MHEALTH LOADING
# =========================================================
def load_mhealth_df(data_dir: str, target_activities):
    log_files = glob.glob(os.path.join(data_dir, "*.log"))
    if not log_files:
        raise FileNotFoundError(f"No .log files found in {data_dir}")

    all_rows = []
    for fpath in log_files:
        filename = os.path.basename(fpath)
        try:
            sub_id = int("".join(filter(str.isdigit, filename)))
        except:
            sub_id = 0

        try:
            df = pd.read_csv(fpath, sep=r"\s+", header=None, engine="python")
        except:
            df = pd.read_csv(fpath, sep="\t", header=None)

        # MHEALTH indices (0-based): 0-2 chest acc, 5-7 ankle acc, 8-10 ankle gyro,
        # 14-16 arm acc, 17-19 arm gyro, 23 label
        if df.shape[1] <= 23:
            raise ValueError("Unexpected MHEALTH column count.")

        df = df.copy()
        df["label"] = df.iloc[:, 23].astype(int)
        df["subject_id"] = sub_id

        df = df[df["label"].isin(target_activities)].copy()
        if len(df) > 0:
            all_rows.append(df)

    if not all_rows:
        raise ValueError("No data found for specified activities.")

    df_all = pd.concat(all_rows, ignore_index=True)
    print(f"[Data] Total samples: {len(df_all)} | Subjects: {sorted(df_all['subject_id'].unique())} | Acts: {sorted(df_all['label'].unique())}")
    return df_all


def get_group_array_from_block(block_np: np.ndarray, group: str):
    if group == "chest_acc":
        return block_np[:, 0:3]
    if group == "ankle_acc":
        return block_np[:, 5:8]
    if group == "arm_acc":
        return block_np[:, 14:17]
    if group == "ankle_gyro":
        return block_np[:, 8:11]
    if group == "arm_gyro":
        return block_np[:, 17:20]
    return None


def create_blocks_by_subject_activity(df_all: pd.DataFrame):
    blocks = []
    base_cols = list(range(24))

    for sub in sorted(df_all["subject_id"].unique()):
        sub_df = df_all[df_all["subject_id"] == sub]
        for act in sorted(sub_df["label"].unique()):
            act_df = sub_df[sub_df["label"] == act]
            raw = act_df[base_cols].to_numpy(dtype=np.float32)
            if len(raw) >= CONFIG["window_size"]:
                blocks.append({"subject": int(sub), "act": int(act), "raw": raw})
    return blocks


# =========================================================
# 2) Detector primitives
# =========================================================
def robust_zscore(x: np.ndarray, eps=1e-6, mad_floor=0.0):
    med = np.median(x)
    mad = np.median(np.abs(x - med)) + eps
    if mad_floor > 0:
        mad = max(mad, mad_floor)
    return (x - med) / mad


def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))


def acc_change_score(acc_3: np.ndarray, var_win: int = 10):
    a = acc_3
    T = len(a)
    u = a / (np.linalg.norm(a, axis=1, keepdims=True) + 1e-8)

    cos_prev = np.sum(u[1:] * u[:-1], axis=1)
    cos_prev = np.clip(cos_prev, -1.0, 1.0)
    dir_change = np.zeros(T, dtype=np.float32)
    dir_change[1:] = 1.0 - cos_prev

    half = var_win // 2
    dir_var = np.zeros(T, dtype=np.float32)
    for i in range(T):
        s = max(0, i - half)
        e = min(T, i + half + 1)
        local = u[s:e]
        dir_var[i] = float(np.mean(np.var(local, axis=0)))

    c = dir_change + dir_var
    return c.astype(np.float32)


def gyro_change_score(gyro_3: np.ndarray, var_win: int = 10):
    w = gyro_3
    T = len(w)
    mag = np.linalg.norm(w, axis=1)

    dmag = np.zeros(T, dtype=np.float32)
    dmag[1:] = np.abs(mag[1:] - mag[:-1])

    half = var_win // 2
    lvar = np.zeros(T, dtype=np.float32)
    for i in range(T):
        s = max(0, i - half)
        e = min(T, i + half + 1)
        lvar[i] = float(np.var(mag[s:e]))

    c = dmag + lvar
    return c.astype(np.float32)


def compute_group_change(group_key: str, Xg: np.ndarray, cfg):
    if Xg is None:
        return None
    if "acc" in group_key:
        return acc_change_score(Xg, var_win=10)
    if "gyro" in group_key:
        return gyro_change_score(Xg, var_win=10)
    # fallback
    return np.linalg.norm(np.diff(Xg, axis=0, prepend=Xg[:1]), axis=1).astype(np.float32)


def steady_score_from_change(c: np.ndarray, tau: float, eps: float,
                             do_savgol: bool, win: int, poly: int,
                             ref_q: float = 80.0, mad_floor: float = 0.0):
    cz = robust_zscore(c, eps=eps, mad_floor=mad_floor)

    # ✅ 핵심: quantile 기준점 shift (median=0 → y median=0.5 문제 제거)
    b = np.percentile(cz, ref_q)
    y = sigmoid((cz - b) / tau)     # transition prob
    s = 1.0 - y                     # steady score

    if do_savgol and len(s) >= win and (win % 2 == 1):
        s = savgol_filter(s, window_length=win, polyorder=poly).astype(np.float32)
        s = np.clip(s, 0.0, 1.0)
    return s.astype(np.float32)


def compute_group_steady_score(group_key: str, Xg: np.ndarray, cfg):
    if Xg is None:
        return None
    c = compute_group_change(group_key, Xg, cfg)
    s = steady_score_from_change(
        c,
        tau=cfg["sigmoid_tau"],
        eps=cfg["mad_eps"],
        do_savgol=cfg["det_savgol"],
        win=cfg["det_savgol_win"],
        poly=cfg["det_savgol_poly"],
        ref_q=cfg["sigmoid_ref_q"],
        mad_floor=cfg["mad_floor"],
    )
    return s


# =========================================================
# 3) Self-consistency quality for fusion
# =========================================================
def cosine_sim(a: np.ndarray, b: np.ndarray, eps=1e-8):
    na = np.linalg.norm(a) + eps
    nb = np.linalg.norm(b) + eps
    return float(np.dot(a, b) / (na * nb))


def estimate_self_consistency_quality(group_key: str, Xg: np.ndarray, sg: np.ndarray, cfg):
    if Xg is None or sg is None:
        return None

    T = len(sg)
    q = cfg["quality_quantile"]
    k_pairs = cfg["quality_pairs"]

    order = np.argsort(sg)
    n = max(5, int(q * T))
    low_idx = order[:n]
    high_idx = order[-n:]
    if len(low_idx) < 5 or len(high_idx) < 5:
        return 0.0

    if "acc" in group_key:
        V = Xg / (np.linalg.norm(Xg, axis=1, keepdims=True) + 1e-8)
    else:
        V = Xg

    def mean_pair_sim(idxs):
        sims = []
        for _ in range(k_pairs):
            i, j = np.random.choice(idxs, size=2, replace=True)
            sims.append(cosine_sim(V[i], V[j]))
        return float(np.mean(sims))

    S_high = mean_pair_sim(high_idx)
    S_low = mean_pair_sim(low_idx)
    return float(S_high - S_low)


def fuse_group_scores(groups_dict, cfg):
    avail = [(g, d) for g, d in groups_dict.items() if d["s"] is not None]
    if not avail:
        raise ValueError("No available groups for fusion.")

    qs = np.array([d["q"] for _, d in avail], dtype=np.float32)
    tau = cfg["quality_tau"]
    ws = np.exp(qs / max(tau, 1e-6))
    ws = ws / (ws.sum() + 1e-8)

    T = len(avail[0][1]["s"])
    s_fused = np.zeros(T, dtype=np.float32)
    weights, qualities = {}, {}
    for (g, d), w in zip(avail, ws):
        s_fused += float(w) * d["s"]
        weights[g] = float(w)
        qualities[g] = float(d["q"])

    s_fused = np.clip(s_fused, 0.0, 1.0)
    return s_fused, weights, qualities


# =========================================================
# 4) Dataset (PATCHED): block-level s_full -> window slice
# =========================================================
class MainEngineDataset(Dataset):
    def __init__(self, blocks, cfg):
        self.cfg = cfg
        self.groups = cfg["groups"]
        self.win = cfg["window_size"]
        self.stride = cfg["stride"]

        self.samples = []  # {"x": (win,C), "s": (win,), "y": (win,), "meta": {...}}

        # ✅ detector/fusion groups vs model input groups 분리
        if cfg.get("detector_use_acc", True):
            det_groups = list(self.groups)
        else:
            det_groups = [g for g in self.groups if "gyro" in g]
        self.det_groups = det_groups  # 디버그/확인용

        for b in blocks:
            raw = b["raw"]  # (T,24)
            subject = b["subject"]
            act = b["act"]
            T = len(raw)

            # full group signals (모델 입력용: 항상 cfg["groups"] 기준)
            full_groups = {}
            for g in self.groups:
                Xg = get_group_array_from_block(raw, g)
                full_groups[g] = None if Xg is None else Xg.astype(np.float32)

            # ---------------------------------------------------------
            # ✅ 핵심: block(trial) 단위로 s_full을 한 번만 계산 (det_groups만)
            # ---------------------------------------------------------
            block_cache = {}  # det g -> {"X_full":..., "s_full":...}
            for g in self.det_groups:
                Xg_full = full_groups.get(g, None)
                if Xg_full is None:
                    block_cache[g] = {"X_full": None, "s_full": None}
                    continue

                c_full = compute_group_change(g, Xg_full, cfg)
                s_full = steady_score_from_change(
                    c_full,
                    tau=cfg["sigmoid_tau"],
                    eps=cfg["mad_eps"],
                    do_savgol=cfg["det_savgol"],
                    win=cfg["det_savgol_win"],
                    poly=cfg["det_savgol_poly"],
                    ref_q=cfg["sigmoid_ref_q"],
                    mad_floor=cfg["mad_floor"],
                )
                block_cache[g] = {"X_full": Xg_full, "s_full": s_full}

            # windows
            for st in range(0, T - self.win + 1, self.stride):
                ed = st + self.win

                # (A) 모델 입력: cfg["groups"] 전체 (acc 포함 가능)
                x_parts = []
                for g in self.groups:
                    Xg_full_in = full_groups.get(g, None)
                    if Xg_full_in is not None:
                        x_parts.append(Xg_full_in[st:ed])  # (win,3)

                if len(x_parts) == 0:
                    continue

                # (B) detector/fusion: det_groups만
                groups_dict = {}
                for g in self.det_groups:
                    Xg_full = block_cache[g]["X_full"]
                    s_full = block_cache[g]["s_full"]

                    if Xg_full is None or s_full is None:
                        continue

                    Xw = Xg_full[st:ed]     # (win,3)
                    sw = s_full[st:ed]      # (win,) ✅ slice only
                    qw = estimate_self_consistency_quality(g, Xw, sw, cfg)

                    groups_dict[g] = {"X": Xw, "s": sw, "q": (0.0 if qw is None else float(qw))}

                if len(groups_dict) == 0:
                    continue

                s_fused, w_dict, q_dict = fuse_group_scores(groups_dict, cfg)
                y_pseudo = (1.0 - s_fused).astype(np.float32)

                # input: concat available groups
                Xcat = np.concatenate(x_parts, axis=1).astype(np.float32)  # (win,C)
                mu = Xcat.mean(axis=0, keepdims=True)
                sd = Xcat.std(axis=0, keepdims=True) + 1e-6
                Xcat = (Xcat - mu) / sd

                self.samples.append({
                    "x": Xcat,                 # (win,C)
                    "s": s_fused.astype(np.float32),
                    "y": y_pseudo,
                    "meta": {"subject": subject, "act": act, "weights": w_dict, "qualities": q_dict}
                })

        print(f"[Dataset] samples/windows = {len(self.samples)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        d = self.samples[idx]
        x = torch.tensor(d["x"], dtype=torch.float32).transpose(0, 1)  # (C,T)
        s = torch.tensor(d["s"], dtype=torch.float32)                  # (T,)
        y = torch.tensor(d["y"], dtype=torch.float32)                  # (T,)
        return x, s, y


# =========================================================
# 5) Model
# =========================================================
class MainEngineNet(nn.Module):
    def __init__(self, input_ch: int, hidden_dim: int, latent_dim: int, win: int):
        super().__init__()
        self.win = win
        self.input_ch = input_ch

        self.backbone = nn.Sequential(
            nn.Conv1d(input_ch, hidden_dim, kernel_size=7, padding=3),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Conv1d(hidden_dim, hidden_dim, kernel_size=5, padding=2),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
        )

        self.to_latent = nn.Sequential(
            nn.Conv1d(hidden_dim, latent_dim, kernel_size=1),
            nn.ReLU(),
        )

        self.trans_head = nn.Sequential(
            nn.Conv1d(latent_dim, 32, kernel_size=1),
            nn.ReLU(),
            nn.Conv1d(32, 1, kernel_size=1),
        )

        self.grav_proj = nn.Linear(latent_dim, 3)

        self.decoder = nn.Sequential(
            nn.Conv1d(latent_dim, hidden_dim, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv1d(hidden_dim, input_ch, kernel_size=3, padding=1),
        )

    def forward(self, x):
        h = self.backbone(x)
        z = self.to_latent(h)
        logits = self.trans_head(z)
        p = torch.sigmoid(logits)
        x_recon = self.decoder(z)
        return z, p, x_recon


# =========================================================
# 6) Losses
# =========================================================
def weighted_soft_bce(p_hat, y_soft, eps=1e-6):
    p = p_hat.squeeze(1)
    y = y_soft
    return F.binary_cross_entropy(p.clamp(eps, 1 - eps), y.clamp(eps, 1 - eps))


def pair_consistency_loss(z, s, cfg):
    B, D, T = z.shape
    delta = cfg["pair_delta"]
    margin = cfg["pair_margin"]
    if T <= delta:
        return torch.tensor(0.0, device=z.device)

    z1 = z[:, :, :-delta]
    z2 = z[:, :, delta:]
    s1 = s[:, :-delta]
    s2 = s[:, delta:]

    z1n = F.normalize(z1, dim=1)
    z2n = F.normalize(z2, dim=1)
    sim = (z1n * z2n).sum(dim=1)

    w_pos = (s1 * s2).detach()
    w_neg = ((1 - s1) * (1 - s2)).detach()

    pos_loss = (w_pos * (1.0 - sim)).mean()
    neg_loss = (w_neg * F.relu(sim - margin)).mean()
    return pos_loss + neg_loss


def inertial_invariance_loss(z, x, s, cfg):
    B, D, T = z.shape
    C = x.shape[1]

    nvec = C // 3
    if nvec == 0:
        return torch.tensor(0.0, device=z.device)

    x_reshaped = x[:, :nvec * 3, :].reshape(B, nvec, 3, T)
    g = x_reshaped.mean(dim=3).mean(dim=1)  # (B,3)
    g = F.normalize(g, dim=1)

    w = s / (s.sum(dim=1, keepdim=True) + 1e-6)
    z_bar = (z * w.unsqueeze(1)).sum(dim=2)  # (B,D)

    z3 = cfg["_grav_proj"](z_bar)
    z3 = F.normalize(z3, dim=1)
    corr = torch.abs(F.cosine_similarity(z3, g, dim=1)).mean()
    return corr


# =========================================================
# 7) Train / Eval
# =========================================================
def train_one_fold(model, loader, cfg):
    opt = torch.optim.AdamW(model.parameters(), lr=cfg["lr"], weight_decay=1e-4)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=cfg["epochs"])

    model.train()
    for ep in range(cfg["epochs"]):
        L = []
        for x, s, y in loader:
            x = x.to(DEVICE)
            s = s.to(DEVICE)
            y = y.to(DEVICE)

            z, p, x_recon = model(x)

            loss_recon = F.mse_loss(x_recon, x)
            loss_soft = weighted_soft_bce(p, y)
            loss_pair = pair_consistency_loss(z, s, cfg)

            cfg["_grav_proj"] = model.grav_proj
            loss_inert = inertial_invariance_loss(z, x, s, cfg)

            loss = (cfg["lambda_recon"] * loss_recon +
                    cfg["lambda_soft"] * loss_soft +
                    cfg["lambda_pair"] * loss_pair +
                    cfg["lambda_inertial"] * loss_inert)

            opt.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step()

            L.append([loss.item(), loss_recon.item(), loss_soft.item(), loss_pair.item(), loss_inert.item()])

        sched.step()
        if (ep + 1) == cfg["epochs"]:
            arr = np.array(L)
            print(f"  > Epoch {ep+1:02d} | Loss {arr[:,0].mean():.4f} | Recon {arr[:,1].mean():.4f} | Soft {arr[:,2].mean():.4f} | Pair {arr[:,3].mean():.4f} | Inert {arr[:,4].mean():.4f}")

    return model


@torch.no_grad()
def eval_one_fold(model, loader):
    model.eval()
    all_p, all_y = [], []
    for x, s, y in loader:
        x = x.to(DEVICE)
        _, p, _ = model(x)
        all_p.append(p.squeeze(1).cpu().numpy())
        all_y.append(y.cpu().numpy())

    P = np.concatenate(all_p, axis=0)
    Y = np.concatenate(all_y, axis=0)
    mae = float(np.mean(np.abs(P - Y)))
    mse = float(np.mean((P - Y) ** 2))
    return {"mae_soft": mae, "mse_soft": mse}


# =========================================================
# 7.5) Debug utilities
# =========================================================
def build_full_groups_from_raw(raw_24: np.ndarray, cfg):
    full_groups = {}
    for g in cfg["groups"]:
        Xg = get_group_array_from_block(raw_24, g)
        full_groups[g] = None if Xg is None else Xg.astype(np.float32)
    return full_groups


def detector_fused_on_block(raw_24: np.ndarray, cfg):
    """
    block-level detector (이미 block-level이라 OK)
    Returns: s_fused(T,), y_pseudo(T,), weights, qualities, s_each
    """
    full_groups = build_full_groups_from_raw(raw_24, cfg)
    groups_dict = {}
    s_each = {}

    if cfg.get("detector_use_acc", True):
        groups_iter = list(cfg["groups"])
    else:
        groups_iter = [g for g in cfg["groups"] if "acc" not in g]

    for g in groups_iter:
        Xg = full_groups.get(g, None)
        if Xg is None:
            sg, qg = None, None
        else:
            # block-level sg(T,)
            sg = compute_group_steady_score(g, Xg, cfg)
            qg = estimate_self_consistency_quality(g, Xg, sg, cfg)

        groups_dict[g] = {"X": Xg, "s": sg, "q": (0.0 if qg is None else float(qg))}
        if sg is not None:
            s_each[g] = sg

    s_fused, w_dict, q_dict = fuse_group_scores(groups_dict, cfg)
    y_pseudo = (1.0 - s_fused).astype(np.float32)
    return s_fused, y_pseudo, w_dict, q_dict, s_each


@torch.no_grad()
def infer_p_hat_on_block(model, raw_24: np.ndarray, cfg):
    model.eval()
    win = cfg["window_size"]
    stride = cfg["stride"]
    T = len(raw_24)

    full_groups = build_full_groups_from_raw(raw_24, cfg)

    p_sum = np.zeros(T, dtype=np.float32)
    p_cnt = np.zeros(T, dtype=np.float32)

    avail_groups = [g for g in cfg["groups"] if full_groups.get(g, None) is not None]
    if len(avail_groups) == 0:
        return None

    for st in range(0, T - win + 1, stride):
        ed = st + win
        x_parts = [full_groups[g][st:ed] for g in avail_groups]
        Xcat = np.concatenate(x_parts, axis=1).astype(np.float32)

        # per-window standardize (Dataset와 동일)
        mu = Xcat.mean(axis=0, keepdims=True)
        sd = Xcat.std(axis=0, keepdims=True) + 1e-6
        Xcat = (Xcat - mu) / sd

        x = torch.tensor(Xcat, dtype=torch.float32).transpose(0, 1).unsqueeze(0).to(DEVICE)
        _, p, _ = model(x)
        pw = p.squeeze(0).squeeze(0).detach().cpu().numpy().astype(np.float32)

        p_sum[st:ed] += pw
        p_cnt[st:ed] += 1.0

    return p_sum / (p_cnt + 1e-8)


def summarize_dataset_meta(ds: MainEngineDataset, cfg):
    if len(ds.samples) == 0:
        return None

    groups = cfg["groups"]

    Y = np.concatenate([d["y"] for d in ds.samples], axis=0)
    y_stats = {
        "mean": float(np.mean(Y)),
        "std": float(np.std(Y)),
        "p50": float(np.percentile(Y, 50)),
        "p90": float(np.percentile(Y, 90)),
        "p99": float(np.percentile(Y, 99)),
        "max": float(np.max(Y)),
    }

    W = {g: [] for g in groups}
    Q = {g: [] for g in groups}

    for d in ds.samples:
        w = d["meta"]["weights"]
        q = d["meta"]["qualities"]
        for g in groups:
            if g in w:
                W[g].append(w[g])
            if g in q:
                Q[g].append(q[g])

    def stat_list(vals):
        if len(vals) == 0:
            return None
        arr = np.array(vals, dtype=np.float32)
        return (float(arr.mean()), float(arr.std()))

    w_stats = {g: stat_list(W[g]) for g in groups}
    q_stats = {g: stat_list(Q[g]) for g in groups}
    return {"y_stats": y_stats, "w_stats": w_stats, "q_stats": q_stats}


def plot_trial_60s(raw_24: np.ndarray, subject: int, act: int, cfg,
                   y_pseudo: np.ndarray, p_full: np.ndarray,
                   weights: dict, qualities: dict, sec: int = 60):
    fs = cfg["fs"]
    T_show = min(int(sec * fs), len(raw_24))
    t = np.arange(T_show) / fs

    def mag(g):
        Xg = get_group_array_from_block(raw_24[:T_show], g)
        if Xg is None:
            return None
        return np.linalg.norm(Xg, axis=1)

    chest_acc = mag("chest_acc")
    ankle_acc = mag("ankle_acc")
    arm_acc = mag("arm_acc")
    ankle_gyro = mag("ankle_gyro")
    arm_gyro = mag("arm_gyro")

    y = y_pseudo[:T_show]
    p = p_full[:T_show] if p_full is not None else None

    plt.figure(figsize=(14, 9))

    plt.subplot(3, 1, 1)
    if chest_acc is not None: plt.plot(t, chest_acc, label="chest_acc |mag|", alpha=0.9)
    if ankle_acc is not None: plt.plot(t, ankle_acc, label="ankle_acc |mag|", alpha=0.9)
    if arm_acc is not None: plt.plot(t, arm_acc, label="arm_acc |mag|", alpha=0.9)
    if ankle_gyro is not None: plt.plot(t, ankle_gyro, label="ankle_gyro |mag|", alpha=0.7)
    if arm_gyro is not None: plt.plot(t, arm_gyro, label="arm_gyro |mag|", alpha=0.7)
    plt.title(f"Trial debug | subject={subject} act={act} | first {sec}s")
    plt.legend(ncol=3)
    plt.grid(alpha=0.25)

    plt.subplot(3, 1, 2)
    plt.plot(t, y, label="pseudo transition (1 - s_fused)")
    plt.ylim(0, 1)
    plt.legend()
    plt.grid(alpha=0.25)

    plt.subplot(3, 1, 3)
    plt.plot(t, y, label="pseudo", alpha=0.6)
    if p is not None:
        plt.plot(t, p, label="model p_hat (overlap-avg)", alpha=0.9)
    plt.ylim(0, 1)
    plt.legend()
    plt.grid(alpha=0.25)

    plt.tight_layout()
    plt.show()

    print("[Trial weights]")
    for k, v in sorted(weights.items(), key=lambda x: -x[1]):
        print(f"  {k}: {v:.3f}")
    print("[Trial qualities]")
    for k, v in sorted(qualities.items(), key=lambda x: -x[1]):
        print(f"  {k}: {v:.3f}")


def debug_detector_distributions_on_block(raw_24: np.ndarray, cfg, sec: int = 60):
    """
    block(trial) 1개에 대해
    group별: c_full -> robust_z -> s_full(steady) -> y_full(transition=1-s)
    fused까지 포함해서 분포를 숫자로 찍어줌.
    """
    fs = cfg["fs"]
    T_show = min(int(sec * fs), len(raw_24))
    raw = raw_24[:T_show]

    print("\n" + "-" * 70)
    print(f"[DBG] Detector dist on block (first {sec}s, T={T_show})")
    print("-" * 70)

    if cfg.get("detector_use_acc", True):
        groups_iter = list(cfg["groups"])
    else:
        groups_iter = [g for g in cfg["groups"] if "acc" not in g]

    groups_dict = {}
    for g in groups_iter:
        Xg = get_group_array_from_block(raw, g)
        if Xg is None:
            print(f"  - {g}: MISSING")
            groups_dict[g] = {"X": None, "s": None, "q": 0.0}
            continue

        c = compute_group_change(g, Xg, cfg)                 # (T,)
        cz = robust_zscore(c, eps=cfg["mad_eps"], mad_floor=cfg["mad_floor"])  # (T,)
        s = steady_score_from_change(                        # (T,)
            c,
            tau=cfg["sigmoid_tau"],
            eps=cfg["mad_eps"],
            do_savgol=cfg["det_savgol"],
            win=cfg["det_savgol_win"],
            poly=cfg["det_savgol_poly"],
            ref_q=cfg["sigmoid_ref_q"],
            mad_floor=cfg["mad_floor"],
        )
        y = 1.0 - s

        def qtile(a):
            return (np.percentile(a, 10), np.percentile(a, 50), np.percentile(a, 90), np.max(a))

        c10, c50, c90, cmax = qtile(c)
        z10, z50, z90, zmax = qtile(cz)
        y10, y50, y90, ymax = qtile(y)

        qg = estimate_self_consistency_quality(g, Xg, s, cfg)
        groups_dict[g] = {"X": Xg, "s": s, "q": float(qg) if qg is not None else 0.0}

        print(f"  - {g}: "
              f"c[p10/p50/p90/max]={c10:.3f}/{c50:.3f}/{c90:.3f}/{cmax:.3f} | "
              f"cz[p10/p50/p90/max]={z10:.3f}/{z50:.3f}/{z90:.3f}/{zmax:.3f} | "
              f"y=1-s[p10/p50/p90/max]={y10:.3f}/{y50:.3f}/{y90:.3f}/{ymax:.3f} | "
              f"quality={groups_dict[g]['q']:.3f}")

    # fused
    s_fused, w_dict, q_dict = fuse_group_scores(groups_dict, cfg)
    y_fused = 1.0 - s_fused
    y10, y50, y90, ymax = (np.percentile(y_fused, 10), np.percentile(y_fused, 50),
                           np.percentile(y_fused, 90), np.max(y_fused))

    print("\n  [FUSED] y=1-s_fused "
          f"[p10/p50/p90/max]={y10:.3f}/{y50:.3f}/{y90:.3f}/{ymax:.3f}")
    print("  [FUSED] weights:", {k: round(v, 3) for k, v in sorted(w_dict.items(), key=lambda x: -x[1])})
    print("  [FUSED] qualities:", {k: round(v, 3) for k, v in sorted(q_dict.items(), key=lambda x: -x[1])})
    print("-" * 70 + "\n")


def corrcoef_safe(a, b):
    a = np.asarray(a).reshape(-1)
    b = np.asarray(b).reshape(-1)
    if len(a) < 3 or np.std(a) < 1e-8 or np.std(b) < 1e-8:
        return 0.0
    return float(np.corrcoef(a, b)[0, 1])


def compute_dirchange_only(acc_3: np.ndarray):
    # acc_change_score 내부에서 "방향 변화"에 해당하는 핵심만 뽑아 간단히 사용
    a = acc_3
    u = a / (np.linalg.norm(a, axis=1, keepdims=True) + 1e-8)
    cos_prev = np.sum(u[1:] * u[:-1], axis=1)
    cos_prev = np.clip(cos_prev, -1.0, 1.0)
    dir_change = np.zeros(len(a), dtype=np.float32)
    dir_change[1:] = 1.0 - cos_prev
    return dir_change  # 클수록 방향 변화 큼


def overlay_debug_on_block(raw_24: np.ndarray, cfg, title="DBG"):
    """
    y(=1-s_fused)가 '전이'를 의미하는지 빠르게 검증:
    - corr(y, |gyro|), corr(y, dir_change), corr(y, |acc|)
    - 오버레이 plot: y vs |gyro| vs dir_change (첫 60초)
    """
    fs = cfg["fs"]
    T = len(raw_24)

    # detector (block-level)
    s_fused, y, w_dict, q_dict, s_each = detector_fused_on_block(raw_24, cfg)

    # signals for reference
    ankle_acc = get_group_array_from_block(raw_24, "ankle_acc")
    ankle_gyro = get_group_array_from_block(raw_24, "ankle_gyro")

    # pick "most reliable" acc/gyro for correlation (간단히 ankle 기준)
    acc_mag = np.linalg.norm(ankle_acc, axis=1) if ankle_acc is not None else None
    gyro_mag = np.linalg.norm(ankle_gyro, axis=1) if ankle_gyro is not None else None
    dirchg = compute_dirchange_only(ankle_acc) if ankle_acc is not None else None

    print("\n" + "-" * 70)
    print(f"[{title}] y meaning check (block-level)")
    if gyro_mag is not None:
        print(f"  corr(y, |ankle_gyro|)      = {corrcoef_safe(y, gyro_mag):+.3f}")
    if dirchg is not None:
        print(f"  corr(y, ankle_acc dirchg)  = {corrcoef_safe(y, dirchg):+.3f}")
    if acc_mag is not None:
        print(f"  corr(y, |ankle_acc|)       = {corrcoef_safe(y, acc_mag):+.3f}")
    print("  (If corr with |acc| dominates and gyro/dirchg are low -> y is 'motion intensity' not 'transition')")
    print("-" * 70)

    # plot (first 60s)
    T_show = min(int(60 * fs), T)
    t = np.arange(T_show) / fs

    plt.figure(figsize=(14, 7))
    ax1 = plt.subplot(2, 1, 1)
    ax1.plot(t, y[:T_show], label="y=transition (1-s_fused)")
    ax1.set_ylim(0, 1)
    ax1.grid(alpha=0.25)
    ax1.legend()

    ax2 = plt.subplot(2, 1, 2)
    if gyro_mag is not None:
        ax2.plot(t, gyro_mag[:T_show], label="|ankle_gyro|")
    if dirchg is not None:
        ax2.plot(t, dirchg[:T_show], label="ankle_acc dir_change")
    if acc_mag is not None:
        ax2.plot(t, acc_mag[:T_show], label="|ankle_acc|", alpha=0.6)
    ax2.grid(alpha=0.25)
    ax2.legend(ncol=3)

    plt.tight_layout()
    plt.show()

    return y, w_dict, q_dict


@torch.no_grad()
def pair_separation_debug(model, raw_24: np.ndarray, cfg, title="DBG"):
    """
    (1) 한 block에서 window들을 굴려서 z를 얻고
    (2) delta 간 cosine sim 분포를 s에 따라 분리해 확인
        - steady-steady (high s * high s)
        - trans-trans   (low s * low s)
    """
    model.eval()
    fs = cfg["fs"]
    win = cfg["window_size"]
    stride = cfg["stride"]
    delta = cfg["pair_delta"]

    # detector block-level
    s_fused, y, _, _, _ = detector_fused_on_block(raw_24, cfg)

    # build full groups for model inference
    full_groups = build_full_groups_from_raw(raw_24, cfg)
    avail_groups = [g for g in cfg["groups"] if full_groups.get(g, None) is not None]
    if len(avail_groups) == 0:
        print(f"[{title}] No available groups for model inference.")
        return

    # accumulate per-time z by overlap-avg (간단/가볍게)
    T = len(raw_24)
    z_sum = None
    z_cnt = np.zeros(T, dtype=np.float32)

    for st in range(0, T - win + 1, stride):
        ed = st + win
        x_parts = [full_groups[g][st:ed] for g in avail_groups]
        Xcat = np.concatenate(x_parts, axis=1).astype(np.float32)
        mu = Xcat.mean(axis=0, keepdims=True)
        sd = Xcat.std(axis=0, keepdims=True) + 1e-6
        Xcat = (Xcat - mu) / sd
        x = torch.tensor(Xcat, dtype=torch.float32).transpose(0, 1).unsqueeze(0).to(DEVICE)

        z, _, _ = model(x)  # z: (1,D,win)
        z = z.squeeze(0).detach().cpu().numpy()  # (D,win)

        if z_sum is None:
            z_sum = np.zeros((z.shape[0], T), dtype=np.float32)

        z_sum[:, st:ed] += z
        z_cnt[st:ed] += 1.0

    z_full = z_sum / (z_cnt[None, :] + 1e-8)  # (D,T)

    # cosine sim between z_t and z_{t+delta}
    if T <= delta + 1:
        print(f"[{title}] Block too short for delta={delta}.")
        return

    z1 = z_full[:, :-delta]
    z2 = z_full[:, delta:]
    z1n = z1 / (np.linalg.norm(z1, axis=0, keepdims=True) + 1e-8)
    z2n = z2 / (np.linalg.norm(z2, axis=0, keepdims=True) + 1e-8)
    sim = np.sum(z1n * z2n, axis=0)  # (T-delta,)

    s1 = s_fused[:-delta]
    s2 = s_fused[delta:]

    y1 = (1.0 - s1)
    y2 = (1.0 - s2)

    q = float(cfg.get("pair_q", 0.2))
    thr_hi = float(np.quantile(y, 1.0 - q))
    thr_lo = float(np.quantile(y, q))

    mask_steady_1 = (y1 <= thr_lo)
    mask_steady_2 = (y2 <= thr_lo)
    mask_trans_1  = (y1 >= thr_hi)
    mask_trans_2  = (y2 >= thr_hi)

    mask_ss = mask_steady_1 & mask_steady_2   # steady -> steady
    mask_tt = mask_trans_1  & mask_trans_2    # trans  -> trans
    mask_st = mask_steady_1 & mask_trans_2    # steady -> trans
    mask_ts = mask_trans_1  & mask_steady_2   # trans  -> steady

    def stats(x):
        if x.size == 0:
            return None
        return dict(mean=float(x.mean()), p10=float(np.quantile(x, 0.1)),
                    p50=float(np.quantile(x, 0.5)), p90=float(np.quantile(x, 0.9)))

    sim_ss = sim[mask_ss]
    sim_tt = sim[mask_tt]
    sim_st = sim[mask_st]
    sim_ts = sim[mask_ts]

    print("\n" + "-" * 70)
    print(f"[{title}] Pair separation (cos sim of z_t vs z_t+delta, delta={delta})")
    print(f"  thresholds (y): steady<= {thr_lo:.2f}, trans>= {thr_hi:.2f}")
    print(f"  SS (steady->steady): {mask_ss.sum():4d} | stats: {stats(sim_ss)}")
    print(f"  TT (trans ->trans) : {mask_tt.sum():4d} | stats: {stats(sim_tt)}")
    print(f"  ST (steady->trans) : {mask_st.sum():4d} | stats: {stats(sim_st)}")
    print(f"  TS (trans ->steady): {mask_ts.sum():4d} | stats: {stats(sim_ts)}")
    print("  Expectation: ST/TS similarity should drop the most (clear separation).")
    print("-" * 70)

    # quick overlay plot for first 60s (optional but very informative)
    T_show = min(int(60 * fs), len(sim))
    t = np.arange(T_show) / fs

    plt.figure(figsize=(14, 6))
    plt.subplot(2, 1, 1)
    plt.plot(t, s_fused[:T_show], label="s_fused (steady)")
    plt.ylim(0, 1)
    plt.grid(alpha=0.25)
    plt.legend()

    plt.subplot(2, 1, 2)
    plt.plot(t, sim[:T_show], label="cos sim(z_t, z_{t+delta})")
    plt.grid(alpha=0.25)
    plt.legend()

    plt.tight_layout()
    plt.show()


def _topk_peaks_simple(y, k=5, min_gap=50):
    """아주 단순한 peak picker: y 큰 순서로 뽑되, min_gap 이내 중복 제거"""
    idx_sorted = np.argsort(y)[::-1]
    picked = []
    for i in idx_sorted:
        if all(abs(i - j) >= min_gap for j in picked):
            picked.append(int(i))
        if len(picked) >= k:
            break
    return sorted(picked)


def debug_peak_zoom_panels(raw_24, y, cfg, sec=60, k=5, zoom_sec=1.0, sensor_for_overlay="ankle"):
    """
    y의 topK peak 근처(±zoom_sec)만 확대해서,
    y vs (|acc|, |gyro|, dirchg) 가 같이 튀는지 확인.
    """
    fs = cfg["fs"]
    T = min(int(sec * fs), raw_24.shape[0])
    y = y[:T]
    raw = raw_24[:T]

    ankle_acc = get_group_array_from_block(raw, "ankle_acc")
    ankle_gyro = get_group_array_from_block(raw, "ankle_gyro")

    if ankle_acc is None or ankle_gyro is None:
        print("[DBG] ankle_acc/ankle_gyro missing -> skip peak zoom")
        return

    acc_mag = np.linalg.norm(ankle_acc, axis=1)
    gyro_mag = np.linalg.norm(ankle_gyro, axis=1)
    dirchg = compute_dirchange_only(ankle_acc)

    peaks = _topk_peaks_simple(y, k=k, min_gap=int(0.5*fs))

    zN = int(zoom_sec * fs)
    for pi in peaks:
        a = max(0, pi - zN)
        b = min(T, pi + zN)

        tt = np.arange(a, b) / fs
        fig, ax = plt.subplots(4, 1, figsize=(12, 8), sharex=True)
        ax[0].plot(tt, y[a:b]);      ax[0].set_ylabel("y (transition)")
        ax[1].plot(tt, dirchg[a:b]); ax[1].set_ylabel("dirchg")
        ax[2].plot(tt, gyro_mag[a:b]); ax[2].set_ylabel("|gyro|")
        ax[3].plot(tt, acc_mag[a:b]);  ax[3].set_ylabel("|acc|")
        ax[3].set_xlabel("time (s)")
        fig.suptitle(f"Peak zoom @ t={pi/fs:.2f}s (±{zoom_sec:.1f}s)")
        plt.tight_layout()
        plt.show()


def debug_conditional_means(raw_24, y, cfg, sec=60, q=0.1):
    """
    y 상/하위 q 비율 구간에서 feature 평균 비교.
    y가 transition이면: dirchg/gyro가 상위에서 확 올라가야 정상.
    """
    fs = cfg["fs"]
    T = min(int(sec * fs), raw_24.shape[0])
    y = y[:T]
    raw = raw_24[:T]

    ankle_acc = get_group_array_from_block(raw, "ankle_acc")
    ankle_gyro = get_group_array_from_block(raw, "ankle_gyro")

    if ankle_acc is None or ankle_gyro is None:
        print("[DBG] ankle_acc/ankle_gyro missing -> skip conditional means")
        return None

    acc_mag = np.linalg.norm(ankle_acc, axis=1)
    gyro_mag = np.linalg.norm(ankle_gyro, axis=1)
    dirchg = compute_dirchange_only(ankle_acc)

    lo = np.quantile(y, q)
    hi = np.quantile(y, 1 - q)
    mask_hi = (y >= hi)
    mask_lo = (y <= lo)

    def m(x, mask):
        return float(np.mean(x[mask])) if np.any(mask) else float("nan")

    out = {
        "y_hi_mean": float(np.mean(y[mask_hi])),
        "y_lo_mean": float(np.mean(y[mask_lo])),
        "|gyro|_hi": m(gyro_mag, mask_hi),
        "|gyro|_lo": m(gyro_mag, mask_lo),
        "dirchg_hi": m(dirchg, mask_hi),
        "dirchg_lo": m(dirchg, mask_lo),
        "|acc|_hi":  m(acc_mag, mask_hi),
        "|acc|_lo":  m(acc_mag, mask_lo),
    }
    print("\n[DBG] Conditional means (top vs bottom)")
    for k, v in out.items():
        print(f"  {k:>10s}: {v:.4f}")
    print("  Expect: dirchg/|gyro| gap >> |acc| gap (if y=transition)\n")
    return out


# =========================================================
# 8) Main (LOSO)
# =========================================================
def main():
    print("=" * 70)
    print("Main Engine (Threshold-free) - MHEALTH LOSO (per-activity)")
    print("=" * 70)

    df_all = load_mhealth_df(CONFIG["data_dir"], CONFIG["target_activities"])
    blocks_all = create_blocks_by_subject_activity(df_all)
    acts = sorted({b["act"] for b in blocks_all})

    overall_results = []

    for act in acts:
        print("\n" + "=" * 70)
        print(f"[Activity {act}] Single-activity LOSO")
        print("=" * 70)

        blocks = [b for b in blocks_all if b["act"] == act]
        subjects = sorted({b["subject"] for b in blocks})
        print(f"[Blocks] act={act} total={len(blocks)} | folds={len(subjects)}")

        act_results = []

        for test_sub in subjects:
            print(f"\n--- Fold: act {act} | test subject {test_sub} ---")

            train_blocks = [b for b in blocks if b["subject"] != test_sub]
            test_blocks = [b for b in blocks if b["subject"] == test_sub]

            train_ds = MainEngineDataset(train_blocks, CONFIG)
            test_ds = MainEngineDataset(test_blocks, CONFIG)

            if len(train_ds) == 0 or len(test_ds) == 0:
                print("  [Skip] empty dataset in this fold.")
                continue

            train_loader = DataLoader(train_ds, batch_size=CONFIG["batch_size"], shuffle=True, drop_last=True)
            test_loader = DataLoader(test_ds, batch_size=CONFIG["batch_size"], shuffle=False)

            x0, s0, y0 = train_ds[0]
            input_ch = x0.shape[0]

            model = MainEngineNet(
                input_ch=input_ch,
                hidden_dim=CONFIG["hidden_dim"],
                latent_dim=CONFIG["latent_dim"],
                win=CONFIG["window_size"]
            ).to(DEVICE)

            model = train_one_fold(model, train_loader, CONFIG)
            met = eval_one_fold(model, test_loader)
            print(f"  > Eval (sanity): MAE={met['mae_soft']:.4f} | MSE={met['mse_soft']:.4f}")

            act_results.append(met)
            overall_results.append(met)

            # Debug once per act
            if test_sub == subjects[0]:
                summ_tr = summarize_dataset_meta(train_ds, CONFIG)
                summ_te = summarize_dataset_meta(test_ds, CONFIG)

                if summ_tr is not None:
                    ys = summ_tr["y_stats"]
                    print(f"[DBG][Act {act}] Train y_pseudo stats: mean={ys['mean']:.4f} std={ys['std']:.4f} "
                          f"p50={ys['p50']:.4f} p90={ys['p90']:.4f} p99={ys['p99']:.4f} max={ys['max']:.4f}")
                    print(f"[DBG][Act {act}] Train weight avg±std:")
                    for g in CONFIG["groups"]:
                        st = summ_tr["w_stats"].get(g, None)
                        if st is not None:
                            print(f"  {g}: {st[0]:.3f} ± {st[1]:.3f}")
                    print(f"[DBG][Act {act}] Train quality avg±std:")
                    for g in CONFIG["groups"]:
                        st = summ_tr["q_stats"].get(g, None)
                        if st is not None:
                            print(f"  {g}: {st[0]:.3f} ± {st[1]:.3f}")

                if summ_te is not None:
                    ys = summ_te["y_stats"]
                    print(f"[DBG][Act {act}] Test  y_pseudo stats: mean={ys['mean']:.4f} std={ys['std']:.4f} "
                          f"p50={ys['p50']:.4f} p90={ys['p90']:.4f} p99={ys['p99']:.4f} max={ys['max']:.4f}")

                b0 = test_blocks[0]
                raw0 = b0["raw"]
                s_fused, y_blk, w_blk, q_blk, _ = detector_fused_on_block(raw0, CONFIG)
                p_blk = infer_p_hat_on_block(model, raw0, CONFIG)

                debug_detector_distributions_on_block(raw0, CONFIG, sec=60)
                plot_trial_60s(
                    raw_24=raw0,
                    subject=b0["subject"],
                    act=b0["act"],
                    cfg=CONFIG,
                    y_pseudo=y_blk,
                    p_full=p_blk,
                    weights=w_blk,
                    qualities=q_blk,
                    sec=60
                )
                # (1) y 의미(부호) 확인
                overlay_debug_on_block(raw0, CONFIG, title=f"Act{act}-Sub{test_sub}")

                # (2) pair 분리 확인
                pair_separation_debug(model, raw0, CONFIG, title=f"Act{act}-Sub{test_sub}")

                # (1) peak zoom plot
                debug_peak_zoom_panels(raw0, y_blk, CONFIG, sec=60, k=5, zoom_sec=1.0)

                # (2) conditional means
                debug_conditional_means(raw0, y_blk, CONFIG, sec=60, q=0.1)

        if act_results:
            mae_a = np.mean([r["mae_soft"] for r in act_results])
            mse_a = np.mean([r["mse_soft"] for r in act_results])
            print(f"\n[Activity {act}] Avg MAE={mae_a:.4f} | Avg MSE={mse_a:.4f}")

    if overall_results:
        mae = np.mean([r["mae_soft"] for r in overall_results])
        mse = np.mean([r["mse_soft"] for r in overall_results])
        print("\n" + "=" * 70)
        print("FINAL (sanity) LOSO RESULTS (overall)")
        print("=" * 70)
        print(f"Avg MAE(p_hat vs pseudo y): {mae:.4f}")
        print(f"Avg MSE(p_hat vs pseudo y): {mse:.4f}")


if __name__ == "__main__":
    main()
```
