```python
import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from scipy.signal import savgol_filter
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 0. Configuration
# ==========================================
CONFIG = {
    # mHealth root (contains mHealth_subject1.log ... mHealth_subject10.log)
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "activity_code": 8,          # Knees bending
    "subjects": list(range(1, 11)),
    "sensor": "ankle",           # ankle only (acc+gyro)

    # windowing
    "window_size": 100,          # 2 sec @ 50Hz
    "stride": 50,                # 1 sec stride
    "fs": 50,

    # training
    "batch_size": 64,
    "lr": 1e-3,
    "epochs": 20,
    "feature_dim": 64,
    "lambda_recon": 1.0,
    "lambda_orient": 0.5,
    "lambda_inertial": 0.3,
    "lambda_contrast": 0.2,
    "seed": 42,

    # ===== Detector upgrade params (구간용) =====
    # gravity LPF: SavGol window length in samples (odd). 추천: ~0.8~1.2 sec
    "g_smooth_sec": 1.0,

    # g_hat angle-change thresholding (hysteresis)
    # theta(t) = acos( ghat(t) dot ghat(t-1) )
    "theta_on": 0.08,    # enter transition if theta exceeds this (rad/sample scale after smoothing)
    "theta_off": 0.05,   # exit transition if below this

    # local variance of g_hat
    "var_on": 0.0020,
    "var_off": 0.0012,

    # gyro gate (보조): 전이로 들어갈 때 gyro도 어느 정도 있어야 한다 (peak counter 방지용이 아니라, "움직임 수반"만 확인)
    "gyro_gate_percentile": 70,  # per-window(또는 per-seq) 기준 상위 p% 이상이면 gate 통과

    # segment cleanup
    "min_trans_dur_sec": 0.20,   # 전이 구간 최소 길이 (초)
    "merge_gap_sec": 0.20,       # 전이 구간 사이 짧은 gap은 합치기 (초)

    # window label threshold (window-level binary for training/eval)
    "steady_thr": 0.3,
    "trans_thr": 0.5,
}

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


def set_seed(seed: int):
    torch.manual_seed(seed)
    np.random.seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


set_seed(CONFIG["seed"])


# ==========================================
# 1. (UPGRADED) Orientation-Based Transition Detection (SEGMENT-MASK)
# ==========================================
class OrientationInvarianceDetector:
    """
    업그레이드 요약:
      - raw acc로 '즉시' orientation을 만들지 않음 (동작 주기에 매우 취약)
      - acc를 저주파로 smooth하여 중력 방향 g_hat(t) 기반으로 "안정 상태 변화"를 전이로 정의
      - hysteresis(th_on/th_off)로 구간 안정화
      - min-duration, gap-merge로 blob(구간) 정리
      - gyro는 score에 섞지 않고 "gate"로만 사용 (에너지/peak counter shortcut 완화)
    """

    def __init__(self, config):
        self.fs = config["fs"]

        self.g_smooth_sec = float(config["g_smooth_sec"])
        self.theta_on = float(config["theta_on"])
        self.theta_off = float(config["theta_off"])
        self.var_on = float(config["var_on"])
        self.var_off = float(config["var_off"])
        self.gyro_gate_percentile = float(config["gyro_gate_percentile"])

        self.min_trans_len = int(round(config["min_trans_dur_sec"] * self.fs))
        self.merge_gap_len = int(round(config["merge_gap_sec"] * self.fs))

    @staticmethod
    def _odd(k: int) -> int:
        return k if (k % 2 == 1) else (k + 1)

    def _smooth_acc_to_gravity(self, acc: np.ndarray) -> np.ndarray:
        """
        Savitzky-Golay smoothing on each axis -> approximates low-pass gravity component.
        acc: (T,3)
        """
        T = acc.shape[0]
        win = self._odd(int(round(self.g_smooth_sec * self.fs)))
        # window length must be >= 5 and <= T
        win = max(5, min(win, T if T % 2 == 1 else T - 1))
        if win < 5:
            # too short, return raw
            return acc.copy()

        g = np.zeros_like(acc, dtype=np.float32)
        for i in range(3):
            g[:, i] = savgol_filter(acc[:, i], window_length=win, polyorder=2, mode="interp")
        return g

    @staticmethod
    def _normalize_rows(x: np.ndarray, eps: float = 1e-8) -> np.ndarray:
        n = np.linalg.norm(x, axis=1, keepdims=True) + eps
        return x / n

    def _ghat(self, acc: np.ndarray) -> np.ndarray:
        g = self._smooth_acc_to_gravity(acc)
        return self._normalize_rows(g)

    @staticmethod
    def _angle_between_unit(u: np.ndarray, v: np.ndarray, eps: float = 1e-8) -> float:
        dot = float(np.clip(np.sum(u * v), -1.0, 1.0))
        return float(np.arccos(dot + eps) if dot < 1.0 else 0.0)

    def _theta_series(self, ghat: np.ndarray) -> np.ndarray:
        """
        theta[t] = angle between ghat[t] and ghat[t-1]
        """
        T = ghat.shape[0]
        theta = np.zeros(T, dtype=np.float32)
        for t in range(1, T):
            theta[t] = self._angle_between_unit(ghat[t], ghat[t - 1])
        return theta

    def _local_var(self, ghat: np.ndarray, sec: float = 1.0) -> np.ndarray:
        """
        local variance of ghat direction vectors in a sliding window (longer time-scale).
        """
        T = ghat.shape[0]
        w = max(3, int(round(sec * self.fs)))
        half = w // 2
        v = np.zeros(T, dtype=np.float32)
        for t in range(T):
            s = max(0, t - half)
            e = min(T, t + half + 1)
            seg = ghat[s:e]
            v[t] = float(np.mean(np.var(seg, axis=0)))
        return v

    @staticmethod
    def _hysteresis_mask(score_on: np.ndarray, score_off: np.ndarray) -> np.ndarray:
        """
        hysteresis mask:
          - enter when score_on is True
          - stay until score_off is False
        Here, we handle it by scanning with two conditions.
        """
        T = score_on.shape[0]
        m = np.zeros(T, dtype=np.float32)
        state = 0
        for t in range(T):
            if state == 0:
                if score_on[t]:
                    state = 1
            else:
                if not score_off[t]:
                    state = 0
            m[t] = float(state)
        return m

    def _remove_short_segments(self, mask: np.ndarray) -> np.ndarray:
        """
        remove segments shorter than min_trans_len
        """
        if self.min_trans_len <= 1:
            return mask
        T = len(mask)
        out = mask.copy()
        t = 0
        while t < T:
            if out[t] == 1:
                s = t
                while t < T and out[t] == 1:
                    t += 1
                e = t
                if (e - s) < self.min_trans_len:
                    out[s:e] = 0
            else:
                t += 1
        return out

    def _merge_small_gaps(self, mask: np.ndarray) -> np.ndarray:
        """
        merge gaps shorter than merge_gap_len between two transition segments
        """
        if self.merge_gap_len <= 0:
            return mask
        T = len(mask)
        out = mask.copy()
        t = 0
        while t < T:
            if out[t] == 1:
                # end current seg
                while t < T and out[t] == 1:
                    t += 1
                gap_s = t
                while t < T and out[t] == 0:
                    t += 1
                gap_e = t
                # if next is a transition segment and gap is short -> fill gap
                if gap_e < T and (gap_e - gap_s) > 0 and (gap_e - gap_s) <= self.merge_gap_len:
                    if out[gap_e] == 1:
                        out[gap_s:gap_e] = 1
            else:
                t += 1
        return out

    def detect_transitions(self, acc: np.ndarray, gyro: np.ndarray):
        """
        Returns:
          transition_mask: (T,) float {0,1}
          orientation_score: (T,) float continuous (for visualization/debug)
        """
        ghat = self._ghat(acc)                          # (T,3)
        theta = self._theta_series(ghat)                # (T,)
        v = self._local_var(ghat, sec=1.0)              # (T,)

        # gyro gate (per sequence)
        gyro_mag = np.linalg.norm(gyro, axis=1)         # (T,)
        gate_thr = np.percentile(gyro_mag, self.gyro_gate_percentile)
        gyro_gate = gyro_mag > gate_thr

        # two-channel condition (theta + var) with hysteresis
        on = (theta > self.theta_on) | (v > self.var_on)
        off = (theta > self.theta_off) | (v > self.var_off)

        # enter transition only if gyro gate OR (theta is very large)
        very_large_theta = theta > (self.theta_on * 1.8)
        on = (on & gyro_gate) | very_large_theta

        mask = self._hysteresis_mask(on, off)

        # cleanup: merge gaps then remove too-short segments (order matters)
        mask = self._merge_small_gaps(mask)
        mask = self._remove_short_segments(mask)

        # continuous score for debugging/visualization (not used for hard masking)
        # NOTE: score is not a simple energy sum; it's based on smoothed gravity-direction change
        score = (theta / (self.theta_on + 1e-8) + v / (self.var_on + 1e-8)) / 2.0
        score = savgol_filter(score.astype(np.float32), window_length=min(11, len(score) if len(score)%2==1 else len(score)-1), polyorder=2, mode="interp")

        return mask.astype(np.float32), score.astype(np.float32)


# ==========================================
# 2. Dataset (window-level transition ratio label)
# ==========================================
class OrientationAwareDataset(Dataset):
    def __init__(self, windows, detector, config):
        """
        windows: (N, T, C=6)
        label: transition_ratio in [0,1] from segment mask
        """
        self.windows = windows
        self.detector = detector
        self.config = config
        self.transition_labels = []
        self.orientation_scores = []

        print("Computing orientation-based transition labels (UPGRADED mask)...")
        for w in windows:
            acc = w[:, :3]
            gyro = w[:, 3:6]
            mask, score = detector.detect_transitions(acc, gyro)
            self.transition_labels.append(float(mask.mean()))
            self.orientation_scores.append(score)

        self.transition_labels = np.asarray(self.transition_labels, dtype=np.float32)
        print(f"Dataset created: {len(self.windows)} windows")
        thr = float(config["steady_thr"])
        print(f"Transition windows (label>{thr}): {np.sum(self.transition_labels > thr)}/{len(self.windows)} ({100*np.mean(self.transition_labels > thr):.1f}%)")

    def __len__(self):
        return len(self.windows)

    def __getitem__(self, idx):
        seq = self.windows[idx]  # (T,6)
        y = self.transition_labels[idx]

        x = torch.tensor(seq, dtype=torch.float32).transpose(0, 1)  # (6,T)
        y = torch.tensor(y, dtype=torch.float32)
        return x, y


# ==========================================
# 3. Model (same as your prototype)
# ==========================================
class OrientationInvariantEncoder(nn.Module):
    def __init__(self, input_ch=6, hidden_dim=128, latent_dim=64, win_size=100):
        super().__init__()

        self.feature_extractor = nn.Sequential(
            nn.Conv1d(input_ch, hidden_dim, kernel_size=7, padding=3),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Conv1d(hidden_dim, hidden_dim, kernel_size=5, padding=2),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
        )

        self.enc_steady = nn.Sequential(
            nn.Conv1d(hidden_dim, hidden_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1),
            nn.Flatten(),
            nn.Linear(hidden_dim, latent_dim),
            nn.ReLU()
        )

        self.enc_transition = nn.Sequential(
            nn.Conv1d(hidden_dim, hidden_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.AdaptiveMaxPool1d(1),
            nn.Flatten(),
            nn.Linear(hidden_dim, latent_dim),
            nn.ReLU()
        )

        self.orientation_predictor = nn.Sequential(
            nn.Linear(latent_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim * 2, hidden_dim * win_size // 4),
            nn.ReLU(),
            nn.Unflatten(1, (hidden_dim, win_size // 4)),
            nn.ConvTranspose1d(hidden_dim, hidden_dim, kernel_size=4, stride=2, padding=1),
            nn.ReLU(),
            nn.ConvTranspose1d(hidden_dim, input_ch, kernel_size=4, stride=2, padding=1),
        )

        self.win_size = win_size

    def forward(self, x):
        features = self.feature_extractor(x)
        z_steady = self.enc_steady(features)
        z_transition = self.enc_transition(features)
        orient_pred = self.orientation_predictor(z_transition)

        z_combined = torch.cat([z_steady, z_transition], dim=1)
        x_recon = self.decoder(z_combined)

        if x_recon.size(2) != self.win_size:
            x_recon = F.interpolate(x_recon, size=self.win_size, mode='linear', align_corners=False)
        return z_steady, z_transition, orient_pred, x_recon


# ==========================================
# 4. Losses (same as your prototype)
# ==========================================
def orientation_continuity_loss(z_steady, transition_labels, steady_thr=0.3,
                                var_weight=0.25, var_gamma=0.5):
    device = z_steady.device
    steady_idx = torch.where(transition_labels < steady_thr)[0]
    if steady_idx.numel() < 2:
        return torch.tensor(0.0, device=device)

    z = z_steady[steady_idx]  # (Ns, D)

    dist = torch.cdist(z, z, p=2)
    n = dist.size(0)
    off_diag = dist[~torch.eye(n, dtype=torch.bool, device=device)]
    loss_invar = off_diag.mean()

    std = z.std(dim=0) + 1e-6
    loss_var = F.relu(var_gamma - std).mean()

    return loss_invar + var_weight * loss_var


def inertial_invariance_loss(z_steady, x, eps=1e-6):
    acc = x[:, :3, :]
    g = acc.mean(dim=2)
    g = F.normalize(g, dim=1, eps=eps)

    z = z_steady
    z = z - z.mean(dim=0, keepdim=True)
    z = z / (z.std(dim=0, keepdim=True) + eps)

    g = g - g.mean(dim=0, keepdim=True)
    g = g / (g.std(dim=0, keepdim=True) + eps)

    B = z.size(0)
    C = (z.T @ g) / (B + eps)
    return (C ** 2).mean()


def steady_transition_contrast_loss(z_steady, z_transition, transition_labels,
                                   steady_thr=0.3, trans_thr=0.5,
                                   margin=0.5, ortho_weight=0.1):
    device = z_steady.device
    trans_idx = torch.where(transition_labels > trans_thr)[0]
    steady_idx = torch.where(transition_labels < steady_thr)[0]
    if trans_idx.numel() < 1 or steady_idx.numel() < 1:
        return torch.tensor(0.0, device=device)

    zt = z_transition
    trans_norm = torch.norm(zt[trans_idx], dim=1).mean()
    steady_norm = torch.norm(zt[steady_idx], dim=1).mean()
    loss_sep = F.relu((steady_norm + margin) - trans_norm)

    zs = F.normalize(z_steady, dim=1)
    zt_n = F.normalize(z_transition, dim=1)
    cos = torch.sum(zs * zt_n, dim=1).abs().mean()
    loss_ortho = cos

    return loss_sep + ortho_weight * loss_ortho


# ==========================================
# 5. Training
# ==========================================
def train(model, loader, config):
    optimizer = torch.optim.AdamW(model.parameters(), lr=config["lr"], weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=config["epochs"])

    model.train()
    print("\n[Training Orientation-Invariant Transition Detector]")

    for epoch in range(config["epochs"]):
        total_loss = total_recon = total_orient = total_inertial = total_contrast = 0.0

        for x, transition_labels in loader:
            x = x.to(DEVICE)
            transition_labels = transition_labels.to(DEVICE)

            z_steady, z_transition, orient_pred, x_recon = model(x)

            loss_recon = F.mse_loss(x_recon, x)

            # (중요) window-level ratio(0~1)를 그대로 soft target으로 사용
            pred = orient_pred.view(-1)
            target = transition_labels.view(-1).clamp(0.0, 1.0)
            loss_orient_pred = F.binary_cross_entropy(pred, target)

            loss_orient_cont = orientation_continuity_loss(
                z_steady, transition_labels, steady_thr=config["steady_thr"]
            )
            loss_inertial = inertial_invariance_loss(z_steady, x)
            loss_contrast = steady_transition_contrast_loss(
                z_steady, z_transition, transition_labels,
                steady_thr=config["steady_thr"], trans_thr=config["trans_thr"]
            )

            loss = (config["lambda_recon"] * loss_recon +
                    config["lambda_orient"] * (loss_orient_pred + loss_orient_cont) +
                    config["lambda_inertial"] * loss_inertial +
                    config["lambda_contrast"] * loss_contrast)

            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            total_loss += float(loss.item())
            total_recon += float(loss_recon.item())
            total_orient += float((loss_orient_pred + loss_orient_cont).item())
            total_inertial += float(loss_inertial.item())
            total_contrast += float(loss_contrast.item())

        scheduler.step()

        if (epoch + 1) % 5 == 0:
            n = len(loader)
            print(f"Epoch {epoch+1}/{config['epochs']} | "
                  f"Loss: {total_loss/n:.4f} | "
                  f"Recon: {total_recon/n:.4f} | "
                  f"Orient: {total_orient/n:.4f} | "
                  f"Inertial: {total_inertial/n:.4f} | "
                  f"Contrast: {total_contrast/n:.4f}")

    return model


# ==========================================
# 6. Evaluation (window-level binary metrics + visualization)
# ==========================================
def evaluate_transition_detection(model, test_loader, config, save_prefix=""):
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

    model.eval()
    all_z_steady, all_z_transition = [], []
    all_pred, all_true = [], []

    with torch.no_grad():
        for x, transition_labels in test_loader:
            x = x.to(DEVICE)
            z_steady, z_transition, orient_pred, _ = model(x)

            all_z_steady.append(z_steady.cpu().numpy())
            all_z_transition.append(z_transition.cpu().numpy())
            all_pred.append(orient_pred.cpu().numpy().ravel())
            all_true.append(transition_labels.cpu().numpy().ravel())

    z_steady = np.vstack(all_z_steady)
    z_transition = np.vstack(all_z_transition)
    y_pred_score = np.concatenate(all_pred).ravel()
    y_true_ratio = np.concatenate(all_true).ravel()

    # binary for reporting (same style as your prototype)
    y_true = (y_true_ratio > config["trans_thr"]).astype(int)
    y_pred = (y_pred_score > 0.5).astype(int)

    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, zero_division=0)
    rec = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    print(f"\n--- Window-level Transition Detection Performance ---")
    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-Score:  {f1:.4f}")

    visualize_results(z_steady, z_transition, y_true, y_pred_score,
                      save_path=f"{save_prefix}orientation_invariance_results.png" if save_prefix else "orientation_invariance_results.png")

    return {"accuracy": acc, "precision": prec, "recall": rec, "f1": f1}


def visualize_results(z_steady, z_transition, y_true, y_pred_score, save_path="orientation_invariance_results.png"):
    from sklearn.decomposition import PCA
    from sklearn.metrics import confusion_matrix

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    if len(z_steady) > 2:
        pca = PCA(n_components=2)
        z2 = pca.fit_transform(z_steady)
        axes[0, 0].scatter(z2[y_true == 0, 0], z2[y_true == 0, 1], alpha=0.5, s=10, c="blue", label="Steady")
        axes[0, 0].scatter(z2[y_true == 1, 0], z2[y_true == 1, 1], alpha=0.5, s=10, c="red", label="Transition")
        axes[0, 0].set_title("Steady Features (PCA)")
        axes[0, 0].legend()

    if len(z_transition) > 2:
        pca2 = PCA(n_components=2)
        zt2 = pca2.fit_transform(z_transition)
        axes[0, 1].scatter(zt2[y_true == 0, 0], zt2[y_true == 0, 1], alpha=0.5, s=10, c="blue", label="Steady")
        axes[0, 1].scatter(zt2[y_true == 1, 0], zt2[y_true == 1, 1], alpha=0.5, s=10, c="red", label="Transition")
        axes[0, 1].set_title("Transition Features (PCA)")
        axes[0, 1].legend()

    # score distribution
    data_to_plot, labels_to_plot, colors_to_plot = [], [], []
    steady_preds = y_pred_score[y_true == 0]
    trans_preds = y_pred_score[y_true == 1]
    if len(steady_preds) > 0:
        data_to_plot.append(steady_preds); labels_to_plot.append("True Steady"); colors_to_plot.append("blue")
    if len(trans_preds) > 0:
        data_to_plot.append(trans_preds); labels_to_plot.append("True Transition"); colors_to_plot.append("red")

    if data_to_plot:
        axes[1, 0].hist(data_to_plot, bins=50, alpha=0.7, label=labels_to_plot, color=colors_to_plot, stacked=False)
        axes[1, 0].set_title("Prediction Score Distribution")
        axes[1, 0].set_xlabel("Predicted Transition Probability")
        axes[1, 0].legend()
    else:
        axes[1, 0].text(0.5, 0.5, "No data", ha="center")

    y_pred_binary = (y_pred_score > 0.5).astype(int)
    cm = confusion_matrix(y_true, y_pred_binary)
    sns.heatmap(cm, annot=True, fmt="d", ax=axes[1, 1], cmap="Blues")
    axes[1, 1].set_title("Confusion Matrix")
    axes[1, 1].set_xlabel("Predicted")
    axes[1, 1].set_ylabel("True")

    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    print(f"\nVisualization saved to '{save_path}'")
    plt.show()


# ==========================================
# 7. mHealth .log Loading + Activity=8 segment windowing (ANKLE only)
# ==========================================
def _read_mhealth_subject_log(data_dir: str, subject_id: int) -> np.ndarray:
    """
    Reads mHealth_subject{subject}.log (space-separated, no header)
    Returns numpy array shape (N, D)
    """
    path = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    if not os.path.exists(path):
        raise FileNotFoundError(f"Not found: {path}")

    # whitespace separated
    df = pd.read_csv(path, delim_whitespace=True, header=None)
    return df.values.astype(np.float32)


def _extract_activity_segments(arr: np.ndarray, activity_code: int):
    """
    arr: (N, D) last column is label/activity code
    return list of segments indices [ (s,e) ... ] where label==activity_code contiguous
    """
    labels = arr[:, -1].astype(int)
    idx = np.where(labels == activity_code)[0]
    if len(idx) == 0:
        return []

    segments = []
    s = idx[0]
    prev = idx[0]
    for k in idx[1:]:
        if k == prev + 1:
            prev = k
        else:
            segments.append((s, prev + 1))
            s = k
            prev = k
    segments.append((s, prev + 1))
    return segments


def _get_ankle_6ch(arr: np.ndarray) -> np.ndarray:
    """
    mHealth column layout (common):
      chest_acc(3)=0..2
      ecg(2)=3..4
      ankle_acc(3)=5..7
      ankle_gyro(3)=8..10
      ankle_mag(3)=11..13
      arm_acc(3)=14..16
      arm_gyro(3)=17..19
      arm_mag(3)=20..22
      label=23
    We'll take ankle_acc (5..7) + ankle_gyro (8..10) => 6ch
    """
    if arr.shape[1] < 24:
        raise ValueError(f"Unexpected mHealth format (cols={arr.shape[1]}). Expected >=24.")
    ankle_acc = arr[:, 5:8]
    ankle_gyro = arr[:, 8:11]
    return np.concatenate([ankle_acc, ankle_gyro], axis=1)


def _standardize_per_subject(X: np.ndarray) -> np.ndarray:
    mu = X.mean(axis=0, keepdims=True)
    sd = X.std(axis=0, keepdims=True) + 1e-6
    return (X - mu) / sd


def build_windows_for_subject(data_dir: str, subject_id: int, activity_code: int, config) -> np.ndarray:
    """
    Return windows (Nw, T, 6) for given subject & activity_code
    """
    arr = _read_mhealth_subject_log(data_dir, subject_id)
    segments = _extract_activity_segments(arr, activity_code)
    if len(segments) == 0:
        return np.zeros((0, config["window_size"], 6), dtype=np.float32)

    X6 = _get_ankle_6ch(arr)  # (N,6)

    windows = []
    W = int(config["window_size"])
    S = int(config["stride"])

    for (s, e) in segments:
        seg = X6[s:e]
        if len(seg) < W:
            continue

        seg = _standardize_per_subject(seg)

        for t in range(0, len(seg) - W + 1, S):
            windows.append(seg[t:t + W])

    if len(windows) == 0:
        return np.zeros((0, W, 6), dtype=np.float32)
    return np.stack(windows, axis=0).astype(np.float32)


# ==========================================
# 8. LOSO runner (Activity=8 Knees bending)
# ==========================================
def run_loso_knees_bending(config):
    act = config["activity_code"]
    data_dir = config["data_dir"]
    subjects = config["subjects"]

    print("=" * 80)
    print(f"LOSO | Activity={act} (Knees bending={act}) | Sensor={config['sensor']}")
    print(f"Subjects: {subjects}")
    print("=" * 80)

    fold_mae = []   # (optional) not used here; kept for parity with your other logs
    fold_f1 = []
    per_fold = []

    for test_subj in subjects:
        train_subjs = [s for s in subjects if s != test_subj]
        print("\n" + "#" * 80)
        print(f"[Fold] TEST subject{test_subj} | TRAIN subjects {train_subjs}")
        print("#" * 80)

        # Build train/test windows
        train_windows_list = []
        for s in train_subjs:
            w = build_windows_for_subject(data_dir, s, act, config)
            if len(w) > 0:
                train_windows_list.append(w)
        test_windows = build_windows_for_subject(data_dir, test_subj, act, config)

        if len(train_windows_list) == 0 or len(test_windows) == 0:
            print(f"[Skip] subject{test_subj} (no windows)")
            continue

        train_windows = np.concatenate(train_windows_list, axis=0)
        print(f"[Data] Train windows: {len(train_windows)} | Test windows: {len(test_windows)}")

        # Detector (UPGRADED)
        detector = OrientationInvarianceDetector(config)

        # Dataset/Loader
        train_dataset = OrientationAwareDataset(train_windows, detector, config)
        test_dataset = OrientationAwareDataset(test_windows, detector, config)

        train_loader = DataLoader(train_dataset, batch_size=config["batch_size"], shuffle=True, num_workers=0)
        test_loader = DataLoader(test_dataset, batch_size=config["batch_size"], shuffle=False, num_workers=0)

        # Model
        model = OrientationInvariantEncoder(
            input_ch=6,
            hidden_dim=128,
            latent_dim=config["feature_dim"],
            win_size=config["window_size"]
        ).to(DEVICE)

        # Train
        model = train(model, train_loader, config)

        # Eval
        results = evaluate_transition_detection(
            model, test_loader, config,
            save_prefix=f"subject{test_subj}_"
        )

        per_fold.append((test_subj, len(test_dataset), results["f1"], results["accuracy"]))
        fold_f1.append(results["f1"])

        print(f"[Fold Result] subject{test_subj} | F1={results['f1']:.4f} | Acc={results['accuracy']:.4f}")

    if len(per_fold) == 0:
        print("\nNo valid folds.")
        return

    f1_mean = float(np.mean(fold_f1))
    f1_std = float(np.std(fold_f1))
    print("\n" + "=" * 80)
    print("[LOSO Summary] (Window-level transition detection, upgraded physics mask)")
    print(f"F1   : mean={f1_mean:.4f} | std={f1_std:.4f}")
    print("=" * 80)

    print("\nPer-subject results:")
    for (s, n, f1, acc) in per_fold:
        print(f"  subject{s:2d} | n={n:5d} | F1={f1:.4f} | Acc={acc:.4f}")


# ==========================================
# 9. Run
# ==========================================
if __name__ == "__main__":
    run_loso_knees_bending(CONFIG)
```
