```python
import os
import math
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass
from typing import Dict, List, Tuple
import matplotlib.pyplot as plt

# -----------------------------
# 0) CONFIG
# -----------------------------
CONFIG = {
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "fs": 50,

    # activities
    "activities": [6, 7, 8, 12],
    "subjects": list(range(1, 11)),

    # windowing (for representation learning)
    "win": 100,
    "stride": 50,

    # pseudo-eventness
    "beta": 2.0,            # sigmoid sharpness (fixed)
    "eps": 1e-8,
    "steady_var_win": 25,   # ~0.5 sec @ 50Hz
    "prepost_win": 25,      # ~0.5 sec

    # training
    "epochs": 50,
    "batch_size": 128,
    "lr": 1e-3,
    "device": "cuda" if torch.cuda.is_available() else "cpu",

    # losses
    "lambda_recon": 1.0,
    "lambda_sep": 0.5,
    "lambda_smooth": 0.2,
    "lambda_count": 1.0,
    "lambda_cons": 0.2,     # consistency between count head and summed eventness

    # logging / viz
    "log_every": 10,
    "viz_every_fold": True, # show PCA for each fold (True면 많이 뜸)
}

COUNT_TABLE = {
    6: {
        "subject1": 21, "subject2": 19, "subject3": 21, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 21, "subject9": 21, "subject10": 20,
    },
    7: {
        "subject1": 20, "subject2": 20, "subject3": 20, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 19, "subject9": 19, "subject10": 20,
    },
    8: {
        "subject1": 20, "subject2": 21, "subject3": 21, "subject4": 19, "subject5": 20,
        "subject6": 20, "subject7": 21, "subject8": 21, "subject9": 21, "subject10": 21,
    },
    12: {
        "subject1": 40, "subject2": 44, "subject3": 42, "subject4": 42, "subject5": 40,
        "subject6": 42, "subject7": 38, "subject8": 40, "subject9": 40, "subject10": 40,
    },
}

# -----------------------------
# 1) Utilities
# -----------------------------
def set_seed(seed=42):
    import random
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def robust_z(x: np.ndarray, eps=1e-8) -> np.ndarray:
    """Robust z-score: (x - median) / (MAD*1.4826)."""
    med = np.median(x)
    mad = np.median(np.abs(x - med)) + eps
    return (x - med) / (1.4826 * mad)

def sigmoid_np(x):
    x = np.clip(x, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-x))

def moving_var(x: np.ndarray, win: int) -> np.ndarray:
    """Simple moving variance with reflect padding."""
    if win <= 1:
        return np.zeros_like(x)
    pad = win // 2
    xp = np.pad(x, (pad, pad), mode="reflect")
    out = np.zeros_like(x, dtype=np.float32)
    for i in range(len(x)):
        seg = xp[i:i+win]
        out[i] = np.var(seg)
    return out

def cosine_angle(u: np.ndarray, v: np.ndarray, eps=1e-8) -> float:
    uu = u / (np.linalg.norm(u) + eps)
    vv = v / (np.linalg.norm(v) + eps)
    dot = np.clip(np.dot(uu, vv), -1.0, 1.0)
    return float(np.arccos(dot))

def gravity_turning(acc_xyz: np.ndarray, eps=1e-8) -> np.ndarray:
    """
    acc_xyz: (T,3)
    g_t = acc / ||acc||, d_t = arccos(<g_t, g_{t-1}>)
    """
    a = acc_xyz
    g = a / (np.linalg.norm(a, axis=1, keepdims=True) + eps)
    d = np.zeros((len(g),), dtype=np.float32)
    for t in range(1, len(g)):
        dot = np.clip(np.sum(g[t]*g[t-1]), -1.0, 1.0)
        d[t] = float(np.arccos(dot))
    return d

def gyro_turning(gyro_xyz: np.ndarray) -> np.ndarray:
    """Rotation intensity: ||gyro||."""
    return np.linalg.norm(gyro_xyz, axis=1).astype(np.float32)

# -----------------------------
# 2) MHEALTH loader (UCI column meanings)
# -----------------------------
# Column mapping (1-indexed in UCI description):
#  1-3: chest acc (x,y,z)
#  4-5: ECG (2 leads) [unused]
#  6-8: ankle acc
#  9-11: ankle gyro
# 12-14: ankle mag [unused]
# 15-17: arm acc
# 18-20: arm gyro
# 21-23: arm mag [unused]
# 24: label

def read_mhealth_subject(data_dir: str, subject_id: int) -> np.ndarray:
    path = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    if not os.path.exists(path):
        raise FileNotFoundError(f"Not found: {path}")
    arr = np.loadtxt(path)  # (T,24)
    return arr

def extract_activity_segments(labels: np.ndarray, act: int) -> List[Tuple[int,int]]:
    """
    returns list of (start,end) inclusive-exclusive for contiguous regions where label==act
    """
    segs = []
    in_seg = False
    s = 0
    for i, y in enumerate(labels):
        if (y == act) and not in_seg:
            in_seg = True
            s = i
        elif (y != act) and in_seg:
            in_seg = False
            segs.append((s, i))
    if in_seg:
        segs.append((s, len(labels)))
    return segs

def get_sensor_groups(arr: np.ndarray) -> Dict[str, np.ndarray]:
    """
    returns:
      chest: (T,3)  -> acc
      ankle: (T,6)  -> acc+gyro
      arm:   (T,6)  -> acc+gyro
    """
    # columns are 0-indexed now:
    chest_acc = arr[:, 0:3]
    ankle_acc = arr[:, 5:8]
    ankle_gyr = arr[:, 8:11]
    arm_acc   = arr[:, 14:17]
    arm_gyr   = arr[:, 17:20]

    groups = {
        "chest": chest_acc.astype(np.float32),
        "ankle": np.concatenate([ankle_acc, ankle_gyr], axis=1).astype(np.float32),
        "arm":   np.concatenate([arm_acc, arm_gyr], axis=1).astype(np.float32),
    }
    return groups

def pick_main_trial_for_subject_activity(arr: np.ndarray, act: int) -> Tuple[np.ndarray, Dict[str,np.ndarray]]:
    """
    Many subjects have exactly one segment for rep-based acts; to be safe:
    choose the longest contiguous segment for this activity.
    """
    labels = arr[:, -1].astype(int)
    segs = extract_activity_segments(labels, act)
    if len(segs) == 0:
        return None, None
    # choose longest
    lens = [(e-s) for (s,e) in segs]
    idx = int(np.argmax(lens))
    s, e = segs[idx]
    trial_arr = arr[s:e].copy()
    groups = get_sensor_groups(trial_arr)
    return trial_arr, groups

# -----------------------------
# 3) Pseudo-eventness (multi-physics, threshold-free)
# -----------------------------
def make_pseudo_eventness(groups: Dict[str,np.ndarray], cfg: Dict) -> np.ndarray:
    """
    Uses >=2 physics definitions:
      (1) gravity turning (from acc)
      (2) gyro turning intensity (from gyro where available)
    plus (optional) pre/post steadiness gating to suppress "fully dynamic" regions.
    Output: p_t in [0,1], shape (T,)
    """
    beta = cfg["beta"]
    eps = cfg["eps"]
    T = next(iter(groups.values())).shape[0]

    # Per-group scores
    # chest: only acc -> gravity turning
    chest_acc = groups["chest"]
    d_g_chest = gravity_turning(chest_acc, eps=eps)
    p_g_chest = sigmoid_np(beta * robust_z(d_g_chest, eps=eps))

    # ankle: acc+gyro (0:3 acc, 3:6 gyro)
    ankle = groups["ankle"]
    ankle_acc = ankle[:, 0:3]
    ankle_gyr = ankle[:, 3:6]
    d_g_ankle = gravity_turning(ankle_acc, eps=eps)
    d_w_ankle = gyro_turning(ankle_gyr)
    p_g_ankle = sigmoid_np(beta * robust_z(d_g_ankle, eps=eps))
    p_w_ankle = sigmoid_np(beta * robust_z(d_w_ankle, eps=eps))

    # arm: acc+gyro
    arm = groups["arm"]
    arm_acc = arm[:, 0:3]
    arm_gyr = arm[:, 3:6]
    d_g_arm = gravity_turning(arm_acc, eps=eps)
    d_w_arm = gyro_turning(arm_gyr)
    p_g_arm = sigmoid_np(beta * robust_z(d_g_arm, eps=eps))
    p_w_arm = sigmoid_np(beta * robust_z(d_w_arm, eps=eps))

    # --- Multi-definition agreement ---
    # For ankle/arm: combine gravity-turn + gyro-turn as AND-like product.
    p_ankle = p_g_ankle * p_w_ankle
    p_arm   = p_g_arm   * p_w_arm
    # chest has only gravity component
    p_chest = p_g_chest

    # vote across sensor groups (no manual sensor choice)
    p_vote = (p_chest + p_ankle + p_arm) / 3.0

    # --- Pre/Post steadiness gating (optional, still threshold-free) ---
    # steady implies low variance of gravity-turning in a neighborhood
    var_win = cfg["steady_var_win"]
    prepost = cfg["prepost_win"]

    # use a pooled gravity turning signal (average)
    d_g_avg = (d_g_chest + d_g_ankle + d_g_arm) / 3.0
    var_t = moving_var(d_g_avg, win=var_win)
    # steadiness score high when var is low -> invert z
    S = sigmoid_np(-beta * robust_z(var_t, eps=eps))

    # pre/post steadiness
    # S_pre(t) = mean S[t-prepost:t], S_post(t)=mean S[t:t+prepost]
    pad = prepost
    Sp = np.pad(S, (pad, pad), mode="reflect")
    S_pre = np.zeros((T,), dtype=np.float32)
    S_post = np.zeros((T,), dtype=np.float32)
    for t in range(T):
        S_pre[t]  = float(np.mean(Sp[t:t+prepost]))
        S_post[t] = float(np.mean(Sp[t+prepost:t+2*prepost]))

    # final eventness: agreement * (pre steady) * (post steady)
    p = p_vote * S_pre * S_post
    p = np.clip(p, 0.0, 1.0).astype(np.float32)
    return p

# -----------------------------
# 4) Dataset for windows (time-wise learning + trial-level count)
# -----------------------------
@dataclass
class Trial:
    subject: int
    act: int
    X: np.ndarray      # (T, C_total=15) concatenated chest(3)+ankle(6)+arm(6)
    p: np.ndarray      # (T,)
    count: float       # GT count

def concat_groups(groups: Dict[str,np.ndarray]) -> np.ndarray:
    # fixed order
    return np.concatenate([groups["chest"], groups["ankle"], groups["arm"]], axis=1).astype(np.float32)

class WindowDataset(torch.utils.data.Dataset):
    def __init__(self, trials: List[Trial], win: int, stride: int):
        self.samples = []
        self.trials = trials
        self.win = win
        self.stride = stride

        for tidx, tr in enumerate(trials):
            T = tr.X.shape[0]
            for s in range(0, max(1, T - win + 1), stride):
                e = s + win
                if e > T:
                    break
                self.samples.append((tidx, s, e))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        tidx, s, e = self.samples[idx]
        tr = self.trials[tidx]
        x = tr.X[s:e]          # (win,C)
        p = tr.p[s:e]          # (win,)
        c = tr.count           # scalar
        return torch.from_numpy(x), torch.from_numpy(p), torch.tensor([c], dtype=torch.float32)

# -----------------------------
# 5) Model (sensor gating + encoder/decoder + event head + count head)
# -----------------------------
class SensorGate(nn.Module):
    """
    Learnable soft gate per sensor group:
      chest(3), ankle(6), arm(6)
    """
    def __init__(self):
        super().__init__()
        self.logits = nn.Parameter(torch.zeros(3))  # chest, ankle, arm

    def forward(self, x):
        # x: (B,T,15) -> apply group-wise weights
        w = torch.softmax(self.logits, dim=0)  # (3,)
        chest = x[..., 0:3]   * w[0]
        ankle = x[..., 3:9]   * w[1]
        arm   = x[..., 9:15]  * w[2]
        xg = torch.cat([chest, ankle, arm], dim=-1)
        return xg, w

class ConvEncoder(nn.Module):
    def __init__(self, in_ch=15, hid=64, zdim=32):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_ch, hid, 5, padding=2),
            nn.BatchNorm1d(hid),
            nn.ReLU(),
            nn.Conv1d(hid, hid, 5, padding=2),
            nn.BatchNorm1d(hid),
            nn.ReLU(),
            nn.Conv1d(hid, zdim, 3, padding=1),
        )

    def forward(self, x):
        # x: (B,T,C) -> (B,T,z)
        x = x.transpose(1,2)          # (B,C,T)
        z = self.net(x).transpose(1,2)
        return z

class ConvDecoder(nn.Module):
    def __init__(self, zdim=32, hid=64, out_ch=15):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(zdim, hid, 3, padding=1),
            nn.BatchNorm1d(hid),
            nn.ReLU(),
            nn.Conv1d(hid, hid, 5, padding=2),
            nn.BatchNorm1d(hid),
            nn.ReLU(),
            nn.Conv1d(hid, out_ch, 5, padding=2),
        )

    def forward(self, z):
        # z: (B,T,z) -> (B,T,C)
        z = z.transpose(1,2)          # (B,z,T)
        xhat = self.net(z).transpose(1,2)
        return xhat

class EventCountNet(nn.Module):
    def __init__(self, in_ch=15, hid=64, zdim=32):
        super().__init__()
        self.gate = SensorGate()
        self.enc = ConvEncoder(in_ch=in_ch, hid=hid, zdim=zdim)
        self.dec = ConvDecoder(zdim=zdim, hid=hid, out_ch=in_ch)

        # time-wise eventness head
        self.event_head = nn.Sequential(
            nn.Linear(zdim, zdim),
            nn.ReLU(),
            nn.Linear(zdim, 1)
        )

        # trial-wise count head (simple mean pooling + MLP)
        self.count_head = nn.Sequential(
            nn.Linear(zdim, zdim),
            nn.ReLU(),
            nn.Linear(zdim, 1)
        )

    def forward(self, x):
        # x: (B,T,15)
        xg, w = self.gate(x)
        z = self.enc(xg)                          # (B,T,z)
        xhat = self.dec(z)                        # (B,T,15)
        logits_t = self.event_head(z).squeeze(-1) # (B,T)
        p_hat = torch.sigmoid(logits_t)

        z_pool = z.mean(dim=1)                    # (B,z)
        c_hat = self.count_head(z_pool)           # (B,1)
        return xhat, z, p_hat, c_hat, w

# -----------------------------
# 6) Losses + metrics
# -----------------------------
def supervised_contrast_loss(z: torch.Tensor, p: torch.Tensor, tau=0.2):
    """
    Soft supervised contrast for steady vs transition.
    z: (B,T,Z), p: (B,T) in [0,1]
    We'll sample anchors from high-p and low-p within each batch.
    Minimal, no fancy mining.
    """
    B, T, Z = z.shape
    z2 = F.normalize(z.reshape(B*T, Z), dim=1)  # (N,Z)
    p2 = p.reshape(B*T)                         # (N,)

    # pick low/high sets by percentile within batch (NOT fixed threshold)
    # (still "threshold-free" in the sense of no absolute numeric threshold; it adapts per batch)
    with torch.no_grad():
        q_low = torch.quantile(p2, 0.2)
        q_high = torch.quantile(p2, 0.8)
        low_idx = torch.where(p2 <= q_low)[0]
        high_idx = torch.where(p2 >= q_high)[0]

    if len(low_idx) < 8 or len(high_idx) < 8:
        return torch.tensor(0.0, device=z.device)

    # sample up to K points
    K = min(256, len(low_idx), len(high_idx))
    low_idx = low_idx[torch.randperm(len(low_idx), device=z.device)[:K]]
    high_idx = high_idx[torch.randperm(len(high_idx), device=z.device)[:K]]

    z_low = z2[low_idx]    # (K,Z)
    z_high = z2[high_idx]  # (K,Z)

    # InfoNCE: low as positives with low (same set), negatives are high
    # (simple set-level separation)
    sim_ll = (z_low @ z_low.t()) / tau
    sim_lh = (z_low @ z_high.t()) / tau

    # remove self-sim
    mask = torch.eye(K, device=z.device).bool()
    sim_ll = sim_ll.masked_fill(mask, -1e9)

    # numerator: sum exp(sim with other lows), denominator: sum exp(sim with lows) + sum exp(sim with highs)
    num = torch.logsumexp(sim_ll, dim=1)
    den = torch.logsumexp(torch.cat([sim_ll, sim_lh], dim=1), dim=1)
    loss = -(num - den).mean()
    return loss

def smoothness_loss(z: torch.Tensor, p: torch.Tensor):
    """
    steady smoothness: (1-p_t) * ||z_t - z_{t-1}||
    """
    dz = (z[:, 1:, :] - z[:, :-1, :]).norm(dim=-1)  # (B,T-1)
    w = (1.0 - p[:, 1:])                              # (B,T-1)
    return (w * dz).mean()

def bce_loss(p_hat: torch.Tensor, p: torch.Tensor, eps=1e-8):
    return F.binary_cross_entropy(p_hat.clamp(eps, 1-eps), p)

def corrcoef_np(a: np.ndarray, b: np.ndarray):
    if a.std() < 1e-8 or b.std() < 1e-8:
        return 0.0
    return float(np.corrcoef(a, b)[0,1])

def auc_approx(p_pred: np.ndarray, p_true: np.ndarray) -> float:
    """
    Approx AUC by binarizing 'true' at top/bottom quantiles (ranking sanity),
    so it's not a hard absolute threshold.
    """
    q_low = np.quantile(p_true, 0.2)
    q_high = np.quantile(p_true, 0.8)
    y = np.zeros_like(p_true, dtype=np.int32) - 1
    y[p_true <= q_low] = 0
    y[p_true >= q_high] = 1
    m = (y != -1)
    if m.sum() < 50:
        return 0.0
    y = y[m]
    s = p_pred[m]
    # compute AUC via ranking
    pos = s[y==1]
    neg = s[y==0]
    if len(pos)==0 or len(neg)==0:
        return 0.0
    # probability a random pos > random neg
    cnt = 0
    tot = len(pos)*len(neg)
    for pv in pos:
        cnt += np.sum(pv > neg)
    return float(cnt / (tot + 1e-8))

# -----------------------------
# 7) Build trials for an activity
# -----------------------------
def build_trials_for_activity(act: int) -> List[Trial]:
    trials = []
    for s in CONFIG["subjects"]:
        arr = read_mhealth_subject(CONFIG["data_dir"], s)
        trial_arr, groups = pick_main_trial_for_subject_activity(arr, act)
        if trial_arr is None:
            print(f"[WARN] subject{s} has no activity {act}")
            continue

        X = concat_groups(groups)  # (T,15)
        p = make_pseudo_eventness(groups, CONFIG)  # (T,)

        key = f"subject{s}"
        if act not in COUNT_TABLE or key not in COUNT_TABLE[act]:
            raise ValueError(f"Missing GT count for act={act}, {key}")
        c = float(COUNT_TABLE[act][key])

        trials.append(Trial(subject=s, act=act, X=X, p=p, count=c))
    return trials

# -----------------------------
# 8) Train/Eval one LOSO fold
# -----------------------------
def run_fold(act: int, test_subject: int, trials: List[Trial]):
    device = CONFIG["device"]
    win = CONFIG["win"]
    stride = CONFIG["stride"]

    train_trials = [t for t in trials if t.subject != test_subject]
    test_trials  = [t for t in trials if t.subject == test_subject]
    if len(test_trials) != 1:
        print(f"[WARN] act={act} test_subject={test_subject} has {len(test_trials)} trials")
        return None

    train_ds = WindowDataset(train_trials, win=win, stride=stride)
    test_ds  = WindowDataset(test_trials,  win=win, stride=stride)

    train_loader = torch.utils.data.DataLoader(train_ds, batch_size=CONFIG["batch_size"], shuffle=True, drop_last=True)
    test_loader  = torch.utils.data.DataLoader(test_ds,  batch_size=CONFIG["batch_size"], shuffle=False, drop_last=False)

    model = EventCountNet(in_ch=15, hid=64, zdim=32).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=CONFIG["lr"])

    # ---------------- train ----------------
    for ep in range(1, CONFIG["epochs"]+1):
        model.train()
        losses = {"total":0,"recon":0,"sep":0,"smooth":0,"event":0,"count":0,"cons":0}
        n = 0
        for x, p, c in train_loader:
            x = x.to(device)         # (B,T,15)
            p = p.to(device)         # (B,T)
            c = c.to(device)         # (B,1)

            xhat, z, p_hat, c_hat, w = model(x)

            # losses
            L_recon = F.mse_loss(xhat, x)
            L_event = bce_loss(p_hat, p)
            L_sep   = supervised_contrast_loss(z, p)
            L_smooth= smoothness_loss(z, p)
            L_count = F.l1_loss(c_hat, c)

            # consistency between count head and summed eventness (time-wise)
            # NOTE: not used as a "count algorithm", only as structural coupling.
            c_from_p = p_hat.mean(dim=1, keepdim=True) * (win / CONFIG["stride"])  # scale to be roughly count-like
            L_cons = F.l1_loss(c_hat, c_from_p.detach())  # detach to avoid trivial solutions

            L = (CONFIG["lambda_recon"]*L_recon +
                 CONFIG["lambda_sep"]*L_sep +
                 CONFIG["lambda_smooth"]*L_smooth +
                 0.5*L_event +
                 CONFIG["lambda_count"]*L_count +
                 CONFIG["lambda_cons"]*L_cons)

            opt.zero_grad()
            L.backward()
            opt.step()

            losses["total"] += float(L.item())
            losses["recon"] += float(L_recon.item())
            losses["sep"]   += float(L_sep.item())
            losses["smooth"]+= float(L_smooth.item())
            losses["event"] += float(L_event.item())
            losses["count"] += float(L_count.item())
            losses["cons"]  += float(L_cons.item())
            n += 1

        for k in losses:
            losses[k] /= max(1,n)

        if ep % CONFIG["log_every"] == 0 or ep == 1:
            # sensor gate
            with torch.no_grad():
                w = torch.softmax(model.gate.logits, dim=0).detach().cpu().numpy()
            print(f"[act={act} | test_sub={test_subject} | ep={ep}] "
                  f"Loss={losses['total']:.4f} | recon={losses['recon']:.4f} | sep={losses['sep']:.4f} | "
                  f"smooth={losses['smooth']:.4f} | eventBCE={losses['event']:.4f} | "
                  f"countL1={losses['count']:.4f} | cons={losses['cons']:.4f} | "
                  f"gate(chest,ankle,arm)={w.round(3)}")

    # ---------------- eval (on full test trial) ----------------
    model.eval()
    test_trial = test_trials[0]
    gt_count = test_trial.count

    # 0) sensor selection log (final)
    with torch.no_grad():
        gate = torch.softmax(model.gate.logits, dim=0).detach().cpu().numpy()

    # 3) pseudo-label match log (p_hat vs p_pseudo) on the full sequence
    # run sliding windows and stitch by averaging overlaps
    T = test_trial.X.shape[0]
    C = test_trial.X.shape[1]
    p_pseudo = test_trial.p.copy()

    p_accum = np.zeros((T,), dtype=np.float32)
    p_weight= np.zeros((T,), dtype=np.float32)

    z_collect = []
    y_collect = []  # 0 steady / 1 transition by pseudo quantile for viz

    win = CONFIG["win"]
    stride = CONFIG["stride"]

    with torch.no_grad():
        for s in range(0, max(1, T - win + 1), stride):
            e = s + win
            if e > T:
                break
            xw = torch.from_numpy(test_trial.X[s:e]).unsqueeze(0).to(device)  # (1,win,C)
            _, z, p_hat, c_hat, _ = model(xw)
            p_hat = p_hat.squeeze(0).detach().cpu().numpy()  # (win,)
            p_accum[s:e] += p_hat
            p_weight[s:e]+= 1.0

            z_np = z.squeeze(0).detach().cpu().numpy()  # (win,z)
            z_collect.append(z_np)

            # label for viz: top/bot quantiles from pseudo within this window
            pw = p_pseudo[s:e]
            ql = np.quantile(pw, 0.2)
            qh = np.quantile(pw, 0.8)
            yy = np.full((win,), -1, dtype=np.int32)
            yy[pw<=ql]=0
            yy[pw>=qh]=1
            y_collect.append(yy)

    p_model = p_accum / np.maximum(p_weight, 1e-8)

    pseudo_bce = float(-(p_pseudo*np.log(np.clip(p_model,1e-8,1-1e-8)) +
                         (1-p_pseudo)*np.log(np.clip(1-p_model,1e-8,1-1e-8))).mean())
    pseudo_corr = corrcoef_np(p_model, p_pseudo)
    pseudo_auc  = auc_approx(p_model, p_pseudo)

    # 4) GT count match log: use model count head averaged over windows
    c_preds = []
    with torch.no_grad():
        for x, p, c in test_loader:
            x = x.to(device)
            _, _, _, c_hat, _ = model(x)
            c_preds.append(c_hat.detach().cpu().numpy())
    c_pred = float(np.mean(np.concatenate(c_preds, axis=0)))

    mae = abs(c_pred - gt_count)
    mape = mae / (gt_count + 1e-8) * 100.0

    # 2) latent separation visualization (PCA) - show
    if CONFIG["viz_every_fold"]:
        Z = np.concatenate(z_collect, axis=0)  # (N,z)
        Y = np.concatenate(y_collect, axis=0)  # (N,)
        mask = (Y != -1)
        Z = Z[mask]
        Y = Y[mask]

        if len(Z) > 5000:
            idx = np.random.choice(len(Z), 5000, replace=False)
            Z = Z[idx]
            Y = Y[idx]

        # PCA (no sklearn)
        Zc = Z - Z.mean(axis=0, keepdims=True)
        cov = Zc.T @ Zc / max(1, len(Zc)-1)
        eigvals, eigvecs = np.linalg.eigh(cov)
        W = eigvecs[:, np.argsort(eigvals)[::-1][:2]]
        Z2 = Zc @ W

        plt.figure()
        plt.scatter(Z2[Y==0,0], Z2[Y==0,1], s=5, alpha=0.4, label="steady (pseudo low)")
        plt.scatter(Z2[Y==1,0], Z2[Y==1,1], s=5, alpha=0.4, label="transition (pseudo high)")
        plt.title(f"PCA latent | act={act} | test_sub={test_subject}")
        plt.legend()
        plt.show()

    # print summary logs
    print(f"[EVAL act={act} | test_sub={test_subject}] "
          f"gate(chest,ankle,arm)={gate.round(3)} | "
          f"pseudo: BCE={pseudo_bce:.4f}, corr={pseudo_corr:.4f}, AUC(top/bot)={pseudo_auc:.4f} | "
          f"count: pred={c_pred:.2f}, gt={gt_count:.2f}, MAE={mae:.2f}, MAPE={mape:.2f}%")

    return {
        "act": act,
        "test_subject": test_subject,
        "gate": gate,
        "pseudo_bce": pseudo_bce,
        "pseudo_corr": pseudo_corr,
        "pseudo_auc": pseudo_auc,
        "count_pred": c_pred,
        "count_gt": gt_count,
        "mae": mae,
        "mape": mape,
    }

# -----------------------------
# 9) Run all activities + LOSO
# -----------------------------
def main():
    set_seed(42)
    all_results = []

    for act in CONFIG["activities"]:
        print("="*80)
        print(f"[BUILD] activity={act}")
        trials = build_trials_for_activity(act)
        print(f"  trials: {len(trials)} subjects")
        print("="*80)

        for test_sub in CONFIG["subjects"]:
            r = run_fold(act, test_sub, trials)
            if r is not None:
                all_results.append(r)

    # summarize
    if len(all_results) == 0:
        print("[DONE] No results.")
        return

    # aggregate per activity
    print("\n" + "="*80)
    print("[SUMMARY] per-activity (LOSO mean)")
    for act in CONFIG["activities"]:
        rs = [r for r in all_results if r["act"] == act]
        if len(rs) == 0:
            continue
        mae = np.mean([r["mae"] for r in rs])
        mape= np.mean([r["mape"] for r in rs])
        pbce= np.mean([r["pseudo_bce"] for r in rs])
        pcorr=np.mean([r["pseudo_corr"] for r in rs])
        pauc = np.mean([r["pseudo_auc"] for r in rs])
        gate = np.mean(np.stack([r["gate"] for r in rs], axis=0), axis=0)
        print(f"  act={act} | MAE={mae:.2f} | MAPE={mape:.2f}% | pseudo(BCE={pbce:.4f}, corr={pcorr:.4f}, AUC={pauc:.4f}) "
              f"| gate(chest,ankle,arm)~{gate.round(3)}")

    # aggregate overall
    mae = np.mean([r["mae"] for r in all_results])
    mape= np.mean([r["mape"] for r in all_results])
    print(f"\n[OVERALL] MAE={mae:.2f} | MAPE={mape:.2f}% over {len(all_results)} folds")

if __name__ == "__main__":
    main()
```

```python
 """
============================================================
Invariance-Driven Disentanglement (Event-Time, NOT segment)
- mHealth (acts: 6,7,8,12) | subjects 1~10 | LOSO per-activity
- NO GT-count used in training (GT-count is evaluation ONLY)
- Two physics-based pseudo-event definitions (threshold-free):
    (1) rotation-jerk (gyro derivative magnitude)
    (2) gravity-direction instability (acc-derived gravity direction change)
- Stronger disentanglement:
    - pseudo top/bottom mining + contrastive separation
    - sparsity + total-variation to make eventness "spiky"
- Stronger sensor selection:
    - learnable gate over (chest, ankle, arm) with entropy minimization
============================================================
"""

import os
import math
import json
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass
from typing import Dict, List, Tuple, Optional

import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.metrics import roc_auc_score


# ============================================================
# 0) CONFIG
# ============================================================
CONFIG = {
    # data
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "fs": 50,
    "subjects": list(range(1, 11)),
    "activities": [6, 7, 8, 12],

    # windowing
    "window_size": 100,
    "stride": 50,

    # training
    "epochs": 50,
    "batch_size": 128,
    "lr": 1e-3,
    "seed": 42,
    "device": "cuda" if torch.cuda.is_available() else "cpu",

    # pseudo-event
    "beta_sigmoid": 1.0,         # sigmoid sharpness for robust-z
    "robust_eps": 1e-6,
    "grav_lp_win": 25,           # moving-average window for gravity estimation

    # top/bottom mining
    "tb_frac": 0.20,             # pick top 20% as "transition", bottom 20% as "steady"

    # losses (NO count loss!)
    "lambda_recon": 1.0,
    "lambda_bce": 1.0,           # pred eventness vs pseudo (soft BCE)
    "lambda_contrast": 0.3,      # top/bot contrast in latent
    "lambda_sparsity": 0.2,      # L1 on event prob
    "lambda_tv": 0.2,            # total-variation on event prob (encourage short impulses)
    "lambda_gate_entropy": 0.1,  # encourage one-hot sensor selection (min entropy)

    # visualization
    "viz_every_fold": False,     # False -> last fold only
}


# ============================================================
# 1) USER GT COUNT TABLE (EVAL ONLY)
#    (You said 40,42,... are correct; use your corrected table here.)
# ============================================================
COUNT_TABLE: Dict[int, Dict[str, int]] = {
    6: {
        "subject1": 21, "subject2": 19, "subject3": 21, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 21, "subject9": 21, "subject10": 20,
    },
    7: {
        "subject1": 20, "subject2": 20, "subject3": 20, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 19, "subject9": 19, "subject10": 20,
    },
    8: {
        "subject1": 20, "subject2": 21, "subject3": 21, "subject4": 19, "subject5": 20,
        "subject6": 20, "subject7": 21, "subject8": 21, "subject9": 21, "subject10": 21,
    },
    12: {
        # IMPORTANT: replace with your corrected "40,42,44..." counts (examples below)
        "subject1": 40, "subject2": 44, "subject3": 42, "subject4": 42, "subject5": 40,
        "subject6": 42, "subject7": 38, "subject8": 40, "subject9": 40, "subject10": 40,
    },
}


# ============================================================
# 2) UTIL
# ============================================================
def set_seed(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def safe_sigmoid_np(x: np.ndarray):
    x = np.clip(x, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-x))


def robust_z(x: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    """
    robust z-score: (x - median) / (1.4826*MAD + eps)
    """
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    denom = 1.4826 * mad + eps
    return (x - med) / denom


def moving_average_1d(x: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return x
    win = int(win)
    kernel = np.ones(win, dtype=np.float32) / float(win)
    # pad reflect to avoid edge artifacts
    pad = win // 2
    xp = np.pad(x, (pad, pad), mode="reflect")
    return np.convolve(xp, kernel, mode="valid")


def moving_average_vec(X: np.ndarray, win: int) -> np.ndarray:
    # X: (T, C)
    if win <= 1:
        return X
    out = np.zeros_like(X)
    for c in range(X.shape[1]):
        out[:, c] = moving_average_1d(X[:, c], win)
    return out


def l2_normalize(v: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / (n + eps)


def pca_scatter(latent: np.ndarray, pseudo: np.ndarray, title: str, tb_frac: float = 0.2):
    """
    Plot PCA 2D colored by pseudo top/bottom.
    """
    if latent.shape[0] < 5:
        print("[PCA] too few samples")
        return

    p = pseudo.reshape(-1)
    n = len(p)
    k = max(2, int(tb_frac * n))
    idx_sorted = np.argsort(p)
    low_idx = idx_sorted[:k]
    high_idx = idx_sorted[-k:]

    labels = np.zeros(n, dtype=np.int64) - 1
    labels[low_idx] = 0
    labels[high_idx] = 1

    X = latent
    Z = PCA(n_components=2).fit_transform(X)

    plt.figure(figsize=(7, 5))
    plt.scatter(Z[labels == 0, 0], Z[labels == 0, 1], s=8, alpha=0.7, label="steady (pseudo low)")
    plt.scatter(Z[labels == 1, 0], Z[labels == 1, 1], s=8, alpha=0.7, label="transition (pseudo high)")
    plt.title(title)
    plt.legend()
    plt.show()


# ============================================================
# 3) mHealth LOADING + TRIAL EXTRACTION
# ============================================================
def _read_mhealth_subject_log(data_dir: str, subject_id: int) -> np.ndarray:
    """
    mHealth: files like "mHealth_subject1.log" ... inside data_dir
    columns:
      - last col is label
      - signals include: chest acc/gyro, ankle acc/gyro, arm acc/gyro, etc.
    NOTE: mHealth format can vary by source; adjust indices below if needed.
    """
    fname = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    arr = np.loadtxt(fname)
    return arr


def _extract_activity_segments(arr: np.ndarray, activity_code: int) -> List[Tuple[int, int]]:
    """
    Return list of (start_idx, end_idx) for contiguous segments where label == activity_code.
    """
    labels = arr[:, -1].astype(int)
    idx = np.where(labels == activity_code)[0]
    if len(idx) == 0:
        return []
    segments = []
    s = idx[0]
    prev = idx[0]
    for i in idx[1:]:
        if i == prev + 1:
            prev = i
        else:
            segments.append((s, prev + 1))
            s = i
            prev = i
    segments.append((s, prev + 1))
    return segments


def _get_sensor_6ch(arr: np.ndarray, which: str) -> np.ndarray:
    """
    Return 6ch = [acc(3), gyro(3)] for sensor: chest / ankle / arm.
    This index mapping is the common one used in many mHealth codes:
      - chest acc: 0:3, chest gyro: 3:6
      - ankle acc: 6:9, ankle gyro: 9:12
      - arm acc:   12:15, arm gyro: 15:18
    If your file differs, adjust here.
    """
    if which == "chest":
        return arr[:, 0:6]
    if which == "ankle":
        return arr[:, 6:12]
    if which == "arm":
        return arr[:, 12:18]
    raise ValueError(which)


def build_trials_for_activity(data_dir: str, activity: int, subjects: List[int]) -> Dict[str, np.ndarray]:
    """
    For each subject, pick ONE trial segment for the activity (the longest).
    Returns: { "subjectX": np.ndarray(T, 18) } concatenated [chest6, ankle6, arm6]
    """
    trials = {}
    for sid in subjects:
        arr = _read_mhealth_subject_log(data_dir, sid)
        segs = _extract_activity_segments(arr, activity)
        if len(segs) == 0:
            continue
        # pick longest segment
        seg = max(segs, key=lambda x: x[1] - x[0])
        s, e = seg
        chest = _get_sensor_6ch(arr[s:e], "chest")
        ankle = _get_sensor_6ch(arr[s:e], "ankle")
        arm = _get_sensor_6ch(arr[s:e], "arm")
        X = np.concatenate([chest, ankle, arm], axis=1)  # (T, 18)
        trials[f"subject{sid}"] = X.astype(np.float32)
    return trials


# ============================================================
# 4) PSEUDO-EVENTNESS (2 DEFINITIONS, THRESHOLD-FREE)
# ============================================================
def pseudo_eventness_two_defs(
    X18: np.ndarray,
    fs: int,
    beta: float,
    eps: float,
    grav_lp_win: int,
) -> np.ndarray:
    """
    X18: (T, 18) [chest6, ankle6, arm6]
    Returns p(t) in [0,1], shape (T,)

    Def-1: rotation jerk (gyro derivative magnitude)
    Def-2: gravity-direction instability (acc-derived gravity direction change rate)
    Combine: soft-AND via product: p = p1 * p2
    """
    T = X18.shape[0]
    if T < 5:
        return np.zeros((T,), dtype=np.float32)

    # use ALL sensors for pseudo (end-to-end, no manual sensor selection)
    # gyro: chest(3:6), ankle(9:12), arm(15:18)
    gyro = np.concatenate([X18[:, 3:6], X18[:, 9:12], X18[:, 15:18]], axis=1)  # (T, 9)
    # acc: chest(0:3), ankle(6:9), arm(12:15)
    acc = np.concatenate([X18[:, 0:3], X18[:, 6:9], X18[:, 12:15]], axis=1)   # (T, 9)

    # ---------------------------
    # Def-1) rotation jerk proxy
    # ---------------------------
    dgyro = np.diff(gyro, axis=0) * fs  # approx derivative
    jerk1 = np.linalg.norm(dgyro, axis=1)  # (T-1,)
    jerk1 = np.concatenate([[jerk1[0]], jerk1], axis=0)  # (T,)
    z1 = robust_z(jerk1, eps=eps)
    p1 = safe_sigmoid_np(beta * z1)

    # --------------------------------------
    # Def-2) gravity direction instability
    # --------------------------------------
    # estimate gravity direction by low-pass filtering acc and normalizing
    acc_lp = moving_average_vec(acc, grav_lp_win)
    ghat = l2_normalize(acc_lp)  # (T, 9) normalized per-time
    # angle change between successive ghat vectors
    dot = np.sum(ghat[1:] * ghat[:-1], axis=1)
    dot = np.clip(dot, -1.0, 1.0)
    ang = np.arccos(dot)  # (T-1,)
    instab2 = np.concatenate([[ang[0]], ang], axis=0) * fs  # "rate-like"
    z2 = robust_z(instab2, eps=eps)
    p2 = safe_sigmoid_np(beta * z2)

    # combine (soft AND): emphasize coincidence
    p = (p1 * p2).astype(np.float32)
    return p


# ============================================================
# 5) DATASET (WINDOWS)
# ============================================================
@dataclass
class WindowItem:
    x: torch.Tensor        # (C, W)
    p: torch.Tensor        # (W,) pseudo eventness at sample-level
    trial_key: str
    start: int


class TrialWindowDataset(torch.utils.data.Dataset):
    def __init__(self, trials: Dict[str, np.ndarray], cfg: Dict):
        self.cfg = cfg
        self.items: List[WindowItem] = []

        W = cfg["window_size"]
        S = cfg["stride"]

        for trial_key, X18 in trials.items():
            p = pseudo_eventness_two_defs(
                X18=X18,
                fs=cfg["fs"],
                beta=cfg["beta_sigmoid"],
                eps=cfg["robust_eps"],
                grav_lp_win=cfg["grav_lp_win"],
            )  # (T,)

            T = X18.shape[0]
            if T < W:
                continue

            for st in range(0, T - W + 1, S):
                xw = X18[st:st + W]          # (W, 18)
                pw = p[st:st + W]            # (W,)
                # (C, W)
                xw_t = torch.from_numpy(xw).float().transpose(0, 1)
                pw_t = torch.from_numpy(pw).float()
                self.items.append(WindowItem(x=xw_t, p=pw_t, trial_key=trial_key, start=st))

    def __len__(self):
        return len(self.items)

    def __getitem__(self, idx: int):
        it = self.items[idx]
        return it.x, it.p, it.trial_key, it.start


def collate_fn(batch):
    xs, ps, keys, starts = zip(*batch)
    X = torch.stack(xs, dim=0)  # (B, C, W)
    P = torch.stack(ps, dim=0)  # (B, W)
    return X, P, list(keys), torch.tensor(starts, dtype=torch.long)


# ============================================================
# 6) MODEL: gated multi-sensor encoder + recon + event head
# ============================================================
class SensorEncoder(nn.Module):
    def __init__(self, in_ch: int, hid: int, out_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_ch, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_dim),
            nn.GELU(),
        )

    def forward(self, x):  # (B, in_ch, W)
        z = self.net(x)     # (B, out_dim, W)
        return z


class Decoder(nn.Module):
    def __init__(self, in_dim: int, hid: int, out_ch: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_dim, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_ch, kernel_size=3, padding=1),
        )

    def forward(self, z):  # (B, in_dim, W)
        return self.net(z)  # (B, out_ch, W)


class GatedEventModel(nn.Module):
    def __init__(self, latent_dim: int = 64, hid: int = 128):
        super().__init__()
        self.latent_dim = latent_dim

        # 3 sensors: chest(6), ankle(6), arm(6)
        self.enc_chest = SensorEncoder(6, hid, latent_dim)
        self.enc_ankle = SensorEncoder(6, hid, latent_dim)
        self.enc_arm   = SensorEncoder(6, hid, latent_dim)

        # learnable gate logits -> softmax weights
        self.gate_logits = nn.Parameter(torch.zeros(3))

        # recon decoder to full 18ch
        self.dec = Decoder(latent_dim, hid, out_ch=18)

        # event head: per-time logit
        self.event_head = nn.Sequential(
            nn.Conv1d(latent_dim, latent_dim, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv1d(latent_dim, 1, kernel_size=1),
        )

    def gate(self):
        w = F.softmax(self.gate_logits, dim=0)  # (3,)
        return w

    def forward(self, x18):  # (B, 18, W)
        x_chest = x18[:, 0:6, :]
        x_ankle = x18[:, 6:12, :]
        x_arm   = x18[:, 12:18, :]

        zc = self.enc_chest(x_chest)  # (B, D, W)
        za = self.enc_ankle(x_ankle)
        zr = self.enc_arm(x_arm)

        w = self.gate()  # (3,)
        z = w[0] * zc + w[1] * za + w[2] * zr  # (B, D, W)

        xhat = self.dec(z)                   # (B, 18, W)
        elogit = self.event_head(z).squeeze(1)  # (B, W)
        return z, xhat, elogit, w


# ============================================================
# 7) LOSSES (stronger disentangle + spiky eventness + gate sparse)
# ============================================================
def entropy(p: torch.Tensor, eps: float = 1e-8) -> torch.Tensor:
    p = torch.clamp(p, eps, 1.0)
    return -torch.sum(p * torch.log(p), dim=-1)


def contrast_top_bottom(z: torch.Tensor, pseudo_w: torch.Tensor, tb_frac: float, temperature: float = 0.2):
    """
    z: (B, D, W)
    pseudo_w: (B, W) -> we reduce to window score (mean)
    Build two sets: top-k and bottom-k by pseudo, then InfoNCE-like separation.
    """
    B = z.shape[0]
    if B < 8:
        return z.new_tensor(0.0)

    # window-level score: mean pseudo in window
    s = pseudo_w.mean(dim=1)  # (B,)
    k = max(2, int(tb_frac * B))
    idx = torch.argsort(s)
    low = idx[:k]
    high = idx[-k:]

    # pool latent per-window: mean over time
    zw = z.mean(dim=2)  # (B, D)
    z_low = F.normalize(zw[low], dim=1)
    z_high = F.normalize(zw[high], dim=1)

    # positive: within same group should be close, cross-group should be far
    # We'll do a simple "group separation" loss:
    # maximize within-group similarity, minimize cross similarity (soft contrast)
    sim_ll = torch.mm(z_low, z_low.t())
    sim_hh = torch.mm(z_high, z_high.t())
    sim_lh = torch.mm(z_low, z_high.t())

    # remove diagonal from within-group
    eye_l = torch.eye(sim_ll.shape[0], device=sim_ll.device)
    eye_h = torch.eye(sim_hh.shape[0], device=sim_hh.device)
    sim_ll = sim_ll * (1 - eye_l)
    sim_hh = sim_hh * (1 - eye_h)

    # want ll, hh high; want lh low
    loss = (
        -sim_ll.mean()   # maximize
        -sim_hh.mean()
        +sim_lh.mean()   # minimize
    )
    return loss


def total_variation(p: torch.Tensor):
    # p: (B, W)
    return torch.mean(torch.abs(p[:, 1:] - p[:, :-1]))


# ============================================================
# 8) COUNT ESTIMATION (EVAL ONLY, NO GT USED)
#   - We avoid "raw-signal peak detection".
#   - We derive count from MODEL event-prob series:
#       (i) K = round(sum(prob))  (soft, threshold-free)
#       (ii) pick top-K local maxima with simple NMS distance
# ============================================================
def count_from_event_prob(prob: np.ndarray, min_dist: int = 3) -> int:
    """
    prob: (N,) [0,1] over window centers (or time points)
    returns integer count
    """
    if prob.size < 3:
        return 0

    # local maxima candidates (no threshold)
    cand = []
    for i in range(1, len(prob) - 1):
        if prob[i] >= prob[i - 1] and prob[i] >= prob[i + 1]:
            cand.append(i)
    if len(cand) == 0:
        return 0

    # desired K from soft mass
    K = int(np.round(float(np.sum(prob))))
    K = max(0, min(K, len(cand)))
    if K == 0:
        return 0

    # sort candidates by prob descending
    cand_sorted = sorted(cand, key=lambda i: prob[i], reverse=True)

    # NMS pick top-K with min distance
    picked = []
    for i in cand_sorted:
        if len(picked) >= K:
            break
        if all(abs(i - j) >= min_dist for j in picked):
            picked.append(i)

    # if NMS reduces too much, just take first K
    if len(picked) < K:
        picked = cand_sorted[:K]
    return len(picked)


# ============================================================
# 9) TRAIN + EVAL (LOSO per activity)
# ============================================================
@torch.no_grad()
def eval_fold(model: nn.Module, ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )
    device = cfg["device"]

    all_pred = []
    all_pseudo = []
    all_latent = []
    all_keys = []

    bce_sum, n_sum = 0.0, 0

    for X, P, keys, _ in loader:
        X = X.to(device)
        P = P.to(device)

        z, xhat, elogit, w = model(X)
        prob = torch.sigmoid(elogit)  # (B, W)

        # pseudo is sample-level; align to same (B,W)
        pseudo = P

        # soft BCE
        bce = F.binary_cross_entropy(prob, pseudo, reduction="mean")
        bce_sum += float(bce.item()) * X.size(0)
        n_sum += X.size(0)

        all_pred.append(prob.detach().cpu().numpy())
        all_pseudo.append(pseudo.detach().cpu().numpy())
        all_latent.append(z.mean(dim=2).detach().cpu().numpy())  # window pooled
        all_keys += keys

    pred = np.concatenate(all_pred, axis=0)    # (Nwin, W)
    pseudo = np.concatenate(all_pseudo, axis=0)
    latent = np.concatenate(all_latent, axis=0)

    # window-level scores for correlation / AUC top/bot
    pred_w = pred.mean(axis=1)
    pseudo_w = pseudo.mean(axis=1)

    corr = np.corrcoef(pred_w, pseudo_w)[0, 1] if len(pred_w) > 3 else 0.0

    # AUC(top/bot): binary labels from pseudo top/bot windows
    n = len(pseudo_w)
    k = max(5, int(cfg["tb_frac"] * n))
    idx = np.argsort(pseudo_w)
    y = np.zeros(n, dtype=np.int64) - 1
    y[idx[:k]] = 0
    y[idx[-k:]] = 1
    mask = y >= 0
    try:
        auc = roc_auc_score(y[mask], pred_w[mask])
    except Exception:
        auc = 0.5

    # Count eval: aggregate per-trial using window-level pred_w
    # We interpret window-level event prob as "eventness at window center".
    # Then count = round(sum(prob)) with peak selection (threshold-free).
    trial_to_probs = {}
    for i, kkey in enumerate(all_keys):
        trial_to_probs.setdefault(kkey, []).append(pred_w[i])

    # for this evaluation, ds contains only one trial (subject) per fold typically,
    # but keep general.
    pred_counts = {}
    for kkey, plist in trial_to_probs.items():
        series = np.array(plist, dtype=np.float32)
        # light smoothing to stabilize (NOT raw-signal peak detection)
        series_sm = moving_average_1d(series, win=5)
        pred_counts[kkey] = count_from_event_prob(series_sm, min_dist=2)

    # pick the test subject trial
    gt = COUNT_TABLE[activity][test_sub]
    # if multiple trials, average count (rare)
    if test_sub in pred_counts:
        c_pred = pred_counts[test_sub]
    else:
        # fallback: mean over trials
        c_pred = float(np.mean(list(pred_counts.values()))) if len(pred_counts) else 0.0

    mae = abs(float(c_pred) - float(gt))
    mape = mae / (abs(float(gt)) + 1e-8) * 100.0

    return {
        "bce": float(bce_sum / max(1, n_sum)),
        "corr": float(corr),
        "auc": float(auc),
        "count_pred": float(c_pred),
        "count_gt": float(gt),
        "mae": float(mae),
        "mape": float(mape),
        "gate": model.gate().detach().cpu().numpy(),
        "latent": latent,
        "pseudo_w": pseudo_w,
        "pred_w": pred_w,
    }


def train_one_fold(train_ds: TrialWindowDataset, test_ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    device = cfg["device"]
    model = GatedEventModel(latent_dim=64, hid=128).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=cfg["lr"])

    train_loader = torch.utils.data.DataLoader(
        train_ds, batch_size=cfg["batch_size"], shuffle=True, collate_fn=collate_fn, drop_last=True
    )

    for ep in range(1, cfg["epochs"] + 1):
        model.train()

        loss_sum = 0.0
        recon_sum = 0.0
        bce_sum = 0.0
        con_sum = 0.0
        sp_sum = 0.0
        tv_sum = 0.0
        ge_sum = 0.0
        n_batches = 0

        for X, P, _, _ in train_loader:
            X = X.to(device)  # (B, 18, W)
            P = P.to(device)  # (B, W)

            z, xhat, elogit, w = model(X)
            prob = torch.sigmoid(elogit)  # (B, W)

            # recon
            recon = F.mse_loss(xhat, X)

            # pseudo-bce (soft target)
            bce = F.binary_cross_entropy(prob, P)

            # top/bottom contrast for stronger separation
            con = contrast_top_bottom(z, P, tb_frac=cfg["tb_frac"])

            # sparsity + TV for spiky eventness
            sp = prob.mean()
            tv = total_variation(prob)

            # gate entropy (minimize entropy -> sharper selection)
            ge = entropy(w.unsqueeze(0)).mean()

            loss = (
                cfg["lambda_recon"] * recon +
                cfg["lambda_bce"] * bce +
                cfg["lambda_contrast"] * con +
                cfg["lambda_sparsity"] * sp +
                cfg["lambda_tv"] * tv +
                cfg["lambda_gate_entropy"] * ge
            )

            opt.zero_grad()
            loss.backward()
            opt.step()

            loss_sum += float(loss.item())
            recon_sum += float(recon.item())
            bce_sum += float(bce.item())
            con_sum += float(con.item())
            sp_sum += float(sp.item())
            tv_sum += float(tv.item())
            ge_sum += float(ge.item())
            n_batches += 1

        if ep == 1 or ep % 10 == 0:
            w_np = model.gate().detach().cpu().numpy()
            print(
                f"[act={activity} | test_sub={test_sub} | ep={ep}] "
                f"Loss={loss_sum/n_batches:.4f} | "
                f"recon={recon_sum/n_batches:.4f} | bce={bce_sum/n_batches:.4f} | "
                f"contrast={con_sum/n_batches:.4f} | sparsity={sp_sum/n_batches:.4f} | tv={tv_sum/n_batches:.4f} | "
                f"gateEnt={ge_sum/n_batches:.4f} | gate(chest,ankle,arm)={np.round(w_np,3)}"
            )

    # eval fold
    metrics = eval_fold(model, test_ds, cfg, activity, test_sub)
    return model, metrics


# ============================================================
# 10) RUN: per-activity LOSO
# ============================================================
def run_all(cfg: Dict):
    set_seed(cfg["seed"])

    overall_maes = []
    overall_mapes = []

    per_act_summary = {}

    for act in cfg["activities"]:
        print("=" * 80)
        print(f"[BUILD] activity={act}")
        trials = build_trials_for_activity(cfg["data_dir"], act, cfg["subjects"])
        print(f"  trials: {len(trials)} subjects")

        # sanity: require all subjects present
        missing = [f"subject{s}" for s in cfg["subjects"] if f"subject{s}" not in trials]
        if len(missing):
            print("[WARN] missing trials:", missing)

        fold_mae = []
        fold_mape = []
        fold_bce = []
        fold_corr = []
        fold_auc = []
        fold_gates = []

        # LOSO on available subjects for this activity
        keys = sorted(list(trials.keys()), key=lambda x: int(x.replace("subject", "")))

        for fi, test_sub in enumerate(keys):
            train_trials = {k: v for k, v in trials.items() if k != test_sub}
            test_trials  = {test_sub: trials[test_sub]}

            train_ds = TrialWindowDataset(train_trials, cfg)
            test_ds  = TrialWindowDataset(test_trials, cfg)

            model, metrics = train_one_fold(train_ds, test_ds, cfg, act, test_sub)

            fold_mae.append(metrics["mae"])
            fold_mape.append(metrics["mape"])
            fold_bce.append(metrics["bce"])
            fold_corr.append(metrics["corr"])
            fold_auc.append(metrics["auc"])
            fold_gates.append(metrics["gate"])

            print(
                f"[EVAL act={act} | test_sub={test_sub}] "
                f"gate(chest,ankle,arm)={np.round(metrics['gate'],3)} | "
                f"pseudo: BCE={metrics['bce']:.4f}, corr={metrics['corr']:.4f}, AUC(top/bot)={metrics['auc']:.4f} | "
                f"count: pred={metrics['count_pred']:.2f}, gt={metrics['count_gt']:.2f}, "
                f"MAE={metrics['mae']:.2f}, MAPE={metrics['mape']:.2f}%"
            )

            # visualization control
            do_viz = cfg["viz_every_fold"] or (fi == len(keys) - 1)
            if do_viz:
                pca_scatter(
                    latent=metrics["latent"],
                    pseudo=metrics["pseudo_w"],
                    title=f"PCA latent | act={act} | test_sub={test_sub}",
                    tb_frac=cfg["tb_frac"]
                )

        per_act_summary[act] = {
            "MAE": float(np.mean(fold_mae)) if len(fold_mae) else float("nan"),
            "MAPE": float(np.mean(fold_mape)) if len(fold_mape) else float("nan"),
            "pseudo_BCE": float(np.mean(fold_bce)) if len(fold_bce) else float("nan"),
            "pseudo_corr": float(np.mean(fold_corr)) if len(fold_corr) else float("nan"),
            "pseudo_AUC": float(np.mean(fold_auc)) if len(fold_auc) else float("nan"),
            "gate_mean": np.mean(np.stack(fold_gates, axis=0), axis=0) if len(fold_gates) else np.array([np.nan]*3),
        }

        overall_maes += fold_mae
        overall_mapes += fold_mape

    # summary
    print("\n" + "=" * 80)
    print("[SUMMARY] per-activity (LOSO mean)")
    for act, s in per_act_summary.items():
        gm = np.round(s["gate_mean"], 3)
        print(
            f"  act={act} | MAE={s['MAE']:.2f} | MAPE={s['MAPE']:.2f}% | "
            f"pseudo(BCE={s['pseudo_BCE']:.4f}, corr={s['pseudo_corr']:.4f}, AUC={s['pseudo_AUC']:.4f}) | "
            f"gate(chest,ankle,arm)~{gm}"
        )

    overall_mae = float(np.mean(overall_maes)) if len(overall_maes) else float("nan")
    overall_mape = float(np.mean(overall_mapes)) if len(overall_mapes) else float("nan")
    print(f"\n[OVERALL] MAE={overall_mae:.2f} | MAPE={overall_mape:.2f}% over {len(overall_maes)} folds")


# ============================================================
# 11) RUN
# ============================================================
if __name__ == "__main__":
    run_all(CONFIG)
```

```python
"""
============================================================
Invariance-Driven Disentanglement (Event-Time, NOT segment)
- mHealth (acts: 6,7,8,12) | subjects 1~10 | LOSO per-activity
- NO GT-count used in training (GT-count is evaluation ONLY)
- Two physics-based pseudo-event definitions (threshold-free):
    (1) rotation-jerk (gyro derivative magnitude)
    (2) gravity-direction instability (acc-derived gravity direction change)
- Stronger disentanglement:
    - pseudo top/bottom mining + contrastive separation
    - sparsity + total-variation to make eventness "spiky"
- Sensor gate REMOVED:
    - use concatenated 18ch input (chest+ankle+arm)
- Trial-wise normalization ADDED:
    - per trial, per channel: (x - mean) / std
    - pseudo is computed from RAW (unnormalized) signals (physics prior)
============================================================
"""

import os
import json
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass
from typing import Dict, List, Tuple

import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.metrics import roc_auc_score


# ============================================================
# 0) CONFIG
# ============================================================
CONFIG = {
    # data
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "fs": 50,
    "subjects": list(range(1, 11)),
    "activities": [6, 7, 8, 12],

    # windowing
    "window_size": 100,
    "stride": 50,

    # trial-wise normalization
    "trial_norm": True,          # <- 핵심
    "trial_norm_eps": 1e-6,

    # training
    "epochs": 50,
    "batch_size": 128,
    "lr": 1e-3,
    "seed": 42,
    "device": "cuda" if torch.cuda.is_available() else "cpu",

    # pseudo-event
    "beta_sigmoid": 1.0,         # sigmoid sharpness for robust-z
    "robust_eps": 1e-6,
    "grav_lp_win": 25,           # moving-average window for gravity estimation

    # top/bottom mining
    "tb_frac": 0.20,             # pick top 20% as "transition", bottom 20% as "steady"

    # losses (NO count loss!)
    "lambda_recon": 1.0,
    "lambda_bce": 1.0,           # pred eventness vs pseudo (soft BCE)
    "lambda_contrast": 0.3,      # top/bot contrast in latent
    "lambda_sparsity": 0.2,      # L1 on event prob
    "lambda_tv": 0.2,            # total-variation on event prob (encourage short impulses)

    # visualization
    "viz_every_fold": False,     # False -> last fold only
}


# ============================================================
# 1) USER GT COUNT TABLE (EVAL ONLY)
# ============================================================
COUNT_TABLE: Dict[int, Dict[str, int]] = {
    6: {
        "subject1": 21, "subject2": 19, "subject3": 21, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 21, "subject9": 21, "subject10": 20,
    },
    7: {
        "subject1": 20, "subject2": 20, "subject3": 20, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 19, "subject9": 19, "subject10": 20,
    },
    8: {
        "subject1": 20, "subject2": 21, "subject3": 21, "subject4": 19, "subject5": 20,
        "subject6": 20, "subject7": 21, "subject8": 21, "subject9": 21, "subject10": 21,
    },
    12: {
        "subject1": 40, "subject2": 44, "subject3": 42, "subject4": 42, "subject5": 40,
        "subject6": 42, "subject7": 38, "subject8": 40, "subject9": 40, "subject10": 40,
    },
}


# ============================================================
# 2) UTIL
# ============================================================
def set_seed(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def safe_sigmoid_np(x: np.ndarray):
    x = np.clip(x, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-x))


def robust_z(x: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    denom = 1.4826 * mad + eps
    return (x - med) / denom


def moving_average_1d(x: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return x
    win = int(win)
    kernel = np.ones(win, dtype=np.float32) / float(win)
    pad = win // 2
    xp = np.pad(x, (pad, pad), mode="reflect")
    return np.convolve(xp, kernel, mode="valid")


def moving_average_vec(X: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return X
    out = np.zeros_like(X)
    for c in range(X.shape[1]):
        out[:, c] = moving_average_1d(X[:, c], win)
    return out


def l2_normalize(v: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / (n + eps)


def pca_scatter(latent: np.ndarray, pseudo: np.ndarray, title: str, tb_frac: float = 0.2):
    if latent.shape[0] < 5:
        print("[PCA] too few samples")
        return

    p = pseudo.reshape(-1)
    n = len(p)
    k = max(2, int(tb_frac * n))
    idx_sorted = np.argsort(p)
    low_idx = idx_sorted[:k]
    high_idx = idx_sorted[-k:]

    labels = np.zeros(n, dtype=np.int64) - 1
    labels[low_idx] = 0
    labels[high_idx] = 1

    Z = PCA(n_components=2).fit_transform(latent)

    plt.figure(figsize=(7, 5))
    plt.scatter(Z[labels == 0, 0], Z[labels == 0, 1], s=8, alpha=0.7, label="steady (pseudo low)")
    plt.scatter(Z[labels == 1, 0], Z[labels == 1, 1], s=8, alpha=0.7, label="transition (pseudo high)")
    plt.title(title)
    plt.legend()
    plt.show()


def trialwise_normalize(X: np.ndarray, eps: float = 1e-6) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    X: (T, C)
    return Xn, mu, std
    """
    mu = X.mean(axis=0, keepdims=True)
    std = X.std(axis=0, keepdims=True)
    Xn = (X - mu) / (std + eps)
    return Xn.astype(np.float32), mu.astype(np.float32), std.astype(np.float32)


# ============================================================
# 3) mHealth LOADING + TRIAL EXTRACTION
# ============================================================
def _read_mhealth_subject_log(data_dir: str, subject_id: int) -> np.ndarray:
    fname = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    arr = np.loadtxt(fname)
    return arr


def _extract_activity_segments(arr: np.ndarray, activity_code: int) -> List[Tuple[int, int]]:
    labels = arr[:, -1].astype(int)
    idx = np.where(labels == activity_code)[0]
    if len(idx) == 0:
        return []
    segments = []
    s = idx[0]
    prev = idx[0]
    for i in idx[1:]:
        if i == prev + 1:
            prev = i
        else:
            segments.append((s, prev + 1))
            s = i
            prev = i
    segments.append((s, prev + 1))
    return segments


def _get_sensor_6ch(arr: np.ndarray, which: str) -> np.ndarray:
    if which == "chest":
        return arr[:, 0:6]
    if which == "ankle":
        return arr[:, 6:12]
    if which == "arm":
        return arr[:, 12:18]
    raise ValueError(which)


def build_trials_for_activity(data_dir: str, activity: int, subjects: List[int]) -> Dict[str, np.ndarray]:
    """
    Returns RAW trials: { "subjectX": (T,18) }
    """
    trials = {}
    for sid in subjects:
        arr = _read_mhealth_subject_log(data_dir, sid)
        segs = _extract_activity_segments(arr, activity)
        if len(segs) == 0:
            continue
        seg = max(segs, key=lambda x: x[1] - x[0])
        s, e = seg
        chest = _get_sensor_6ch(arr[s:e], "chest")
        ankle = _get_sensor_6ch(arr[s:e], "ankle")
        arm = _get_sensor_6ch(arr[s:e], "arm")
        X = np.concatenate([chest, ankle, arm], axis=1)  # (T,18)
        trials[f"subject{sid}"] = X.astype(np.float32)
    return trials


# ============================================================
# 4) PSEUDO-EVENTNESS (computed on RAW)
# ============================================================
def pseudo_eventness_two_defs(
    X18_raw: np.ndarray,
    fs: int,
    beta: float,
    eps: float,
    grav_lp_win: int,
) -> np.ndarray:
    """
    X18_raw: (T,18) RAW
    Returns p(t) in [0,1], shape (T,)
    """
    T = X18_raw.shape[0]
    if T < 5:
        return np.zeros((T,), dtype=np.float32)

    gyro = np.concatenate([X18_raw[:, 3:6], X18_raw[:, 9:12], X18_raw[:, 15:18]], axis=1)  # (T,9)
    acc  = np.concatenate([X18_raw[:, 0:3], X18_raw[:, 6:9], X18_raw[:, 12:15]], axis=1)   # (T,9)

    # Def-1) rotation jerk proxy
    dgyro = np.diff(gyro, axis=0) * fs
    jerk1 = np.linalg.norm(dgyro, axis=1)
    jerk1 = np.concatenate([[jerk1[0]], jerk1], axis=0)
    z1 = robust_z(jerk1, eps=eps)
    p1 = safe_sigmoid_np(beta * z1)

    # Def-2) gravity direction instability
    acc_lp = moving_average_vec(acc, grav_lp_win)
    ghat = l2_normalize(acc_lp)
    dot = np.sum(ghat[1:] * ghat[:-1], axis=1)
    dot = np.clip(dot, -1.0, 1.0)
    ang = np.arccos(dot)
    instab2 = np.concatenate([[ang[0]], ang], axis=0) * fs
    z2 = robust_z(instab2, eps=eps)
    p2 = safe_sigmoid_np(beta * z2)

    return (p1 * p2).astype(np.float32)


# ============================================================
# 5) DATASET (WINDOWS) + TRIAL-WISE NORM
# ============================================================
@dataclass
class WindowItem:
    x: torch.Tensor
    p: torch.Tensor
    trial_key: str
    start: int


class TrialWindowDataset(torch.utils.data.Dataset):
    def __init__(self, trials_raw: Dict[str, np.ndarray], cfg: Dict):
        self.cfg = cfg
        self.items: List[WindowItem] = []

        W = cfg["window_size"]
        S = cfg["stride"]

        for trial_key, X18_raw in trials_raw.items():
            # pseudo from RAW (physics)
            p = pseudo_eventness_two_defs(
                X18_raw=X18_raw,
                fs=cfg["fs"],
                beta=cfg["beta_sigmoid"],
                eps=cfg["robust_eps"],
                grav_lp_win=cfg["grav_lp_win"],
            )

            # model input normalization (trial-wise)
            if cfg.get("trial_norm", True):
                X18, _, _ = trialwise_normalize(X18_raw, eps=cfg["trial_norm_eps"])
            else:
                X18 = X18_raw.astype(np.float32)

            T = X18.shape[0]
            if T < W:
                continue

            for st in range(0, T - W + 1, S):
                xw = X18[st:st + W]       # (W,18) normalized (if enabled)
                pw = p[st:st + W]         # (W,) pseudo from RAW
                xw_t = torch.from_numpy(xw).float().transpose(0, 1)  # (18,W)
                pw_t = torch.from_numpy(pw).float()                  # (W,)
                self.items.append(WindowItem(x=xw_t, p=pw_t, trial_key=trial_key, start=st))

    def __len__(self):
        return len(self.items)

    def __getitem__(self, idx: int):
        it = self.items[idx]
        return it.x, it.p, it.trial_key, it.start


def collate_fn(batch):
    xs, ps, keys, starts = zip(*batch)
    X = torch.stack(xs, dim=0)
    P = torch.stack(ps, dim=0)
    return X, P, list(keys), torch.tensor(starts, dtype=torch.long)


# ============================================================
# 6) MODEL: single 18ch encoder (NO GATE)
# ============================================================
class Encoder18(nn.Module):
    def __init__(self, in_ch: int, hid: int, out_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_ch, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_dim),
            nn.GELU(),
        )

    def forward(self, x):
        return self.net(x)


class Decoder(nn.Module):
    def __init__(self, in_dim: int, hid: int, out_ch: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_dim, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_ch, kernel_size=3, padding=1),
        )

    def forward(self, z):
        return self.net(z)


class EventModel(nn.Module):
    def __init__(self, latent_dim: int = 64, hid: int = 128):
        super().__init__()
        self.enc = Encoder18(18, hid, latent_dim)
        self.dec = Decoder(latent_dim, hid, out_ch=18)
        self.event_head = nn.Sequential(
            nn.Conv1d(latent_dim, latent_dim, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv1d(latent_dim, 1, kernel_size=1),
        )

    def forward(self, x18):  # (B,18,W)
        z = self.enc(x18)                     # (B,D,W)
        xhat = self.dec(z)                    # (B,18,W)
        elogit = self.event_head(z).squeeze(1)  # (B,W)
        return z, xhat, elogit


# ============================================================
# 7) LOSSES (NO gate entropy)
# ============================================================
def contrast_top_bottom(z: torch.Tensor, pseudo_w: torch.Tensor, tb_frac: float):
    B = z.shape[0]
    if B < 8:
        return z.new_tensor(0.0)

    s = pseudo_w.mean(dim=1)
    k = max(2, int(tb_frac * B))
    idx = torch.argsort(s)
    low = idx[:k]
    high = idx[-k:]

    zw = z.mean(dim=2)  # (B,D)
    z_low = F.normalize(zw[low], dim=1)
    z_high = F.normalize(zw[high], dim=1)

    sim_ll = torch.mm(z_low, z_low.t())
    sim_hh = torch.mm(z_high, z_high.t())
    sim_lh = torch.mm(z_low, z_high.t())

    eye_l = torch.eye(sim_ll.shape[0], device=sim_ll.device)
    eye_h = torch.eye(sim_hh.shape[0], device=sim_hh.device)
    sim_ll = sim_ll * (1 - eye_l)
    sim_hh = sim_hh * (1 - eye_h)

    loss = (-sim_ll.mean() - sim_hh.mean() + sim_lh.mean())
    return loss


def total_variation(p: torch.Tensor):
    return torch.mean(torch.abs(p[:, 1:] - p[:, :-1]))


# ============================================================
# 8) COUNT ESTIMATION (EVAL ONLY)
# ============================================================
def count_from_event_prob(prob: np.ndarray, min_dist: int = 3) -> int:
    if prob.size < 3:
        return 0

    cand = []
    for i in range(1, len(prob) - 1):
        if prob[i] >= prob[i - 1] and prob[i] >= prob[i + 1]:
            cand.append(i)
    if len(cand) == 0:
        return 0

    K = int(np.round(float(np.sum(prob))))
    K = max(0, min(K, len(cand)))
    if K == 0:
        return 0

    cand_sorted = sorted(cand, key=lambda i: prob[i], reverse=True)

    picked = []
    for i in cand_sorted:
        if len(picked) >= K:
            break
        if all(abs(i - j) >= min_dist for j in picked):
            picked.append(i)

    if len(picked) < K:
        picked = cand_sorted[:K]
    return len(picked)


# ============================================================
# 9) TRAIN + EVAL (LOSO per activity)
# ============================================================
@torch.no_grad()
def eval_fold(model: nn.Module, ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )
    device = cfg["device"]

    all_pred, all_pseudo, all_latent, all_keys = [], [], [], []
    bce_sum, n_sum = 0.0, 0

    for X, P, keys, _ in loader:
        X = X.to(device)
        P = P.to(device)

        z, xhat, elogit = model(X)
        prob = torch.sigmoid(elogit)

        bce = F.binary_cross_entropy(prob, P, reduction="mean")
        bce_sum += float(bce.item()) * X.size(0)
        n_sum += X.size(0)

        all_pred.append(prob.detach().cpu().numpy())
        all_pseudo.append(P.detach().cpu().numpy())
        all_latent.append(z.mean(dim=2).detach().cpu().numpy())
        all_keys += keys

    pred = np.concatenate(all_pred, axis=0)
    pseudo = np.concatenate(all_pseudo, axis=0)
    latent = np.concatenate(all_latent, axis=0)

    pred_w = pred.mean(axis=1)
    pseudo_w = pseudo.mean(axis=1)

    corr = np.corrcoef(pred_w, pseudo_w)[0, 1] if len(pred_w) > 3 else 0.0

    n = len(pseudo_w)
    k = max(5, int(cfg["tb_frac"] * n))
    idx = np.argsort(pseudo_w)
    y = np.zeros(n, dtype=np.int64) - 1
    y[idx[:k]] = 0
    y[idx[-k:]] = 1
    mask = y >= 0
    try:
        auc = roc_auc_score(y[mask], pred_w[mask])
    except Exception:
        auc = 0.5

    trial_to_probs = {}
    for i, kkey in enumerate(all_keys):
        trial_to_probs.setdefault(kkey, []).append(pred_w[i])

    pred_counts = {}
    for kkey, plist in trial_to_probs.items():
        series = np.array(plist, dtype=np.float32)
        series_sm = moving_average_1d(series, win=5)
        pred_counts[kkey] = count_from_event_prob(series_sm, min_dist=2)

    gt = COUNT_TABLE[activity][test_sub]
    c_pred = pred_counts.get(test_sub, float(np.mean(list(pred_counts.values()))) if len(pred_counts) else 0.0)

    mae = abs(float(c_pred) - float(gt))
    mape = mae / (abs(float(gt)) + 1e-8) * 100.0

    return {
        "bce": float(bce_sum / max(1, n_sum)),
        "corr": float(corr),
        "auc": float(auc),
        "count_pred": float(c_pred),
        "count_gt": float(gt),
        "mae": float(mae),
        "mape": float(mape),
        "latent": latent,
        "pseudo_w": pseudo_w,
        "pred_w": pred_w,
    }


def train_one_fold(train_ds: TrialWindowDataset, test_ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    device = cfg["device"]
    model = EventModel(latent_dim=64, hid=128).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=cfg["lr"])

    train_loader = torch.utils.data.DataLoader(
        train_ds, batch_size=cfg["batch_size"], shuffle=True, collate_fn=collate_fn, drop_last=True
    )

    for ep in range(1, cfg["epochs"] + 1):
        model.train()

        loss_sum = recon_sum = bce_sum = con_sum = sp_sum = tv_sum = 0.0
        n_batches = 0

        for X, P, _, _ in train_loader:
            X = X.to(device)
            P = P.to(device)

            z, xhat, elogit = model(X)
            prob = torch.sigmoid(elogit)

            recon = F.mse_loss(xhat, X)
            bce = F.binary_cross_entropy(prob, P)
            con = contrast_top_bottom(z, P, tb_frac=cfg["tb_frac"])
            sp = prob.mean()
            tv = total_variation(prob)

            loss = (
                cfg["lambda_recon"] * recon +
                cfg["lambda_bce"] * bce +
                cfg["lambda_contrast"] * con +
                cfg["lambda_sparsity"] * sp +
                cfg["lambda_tv"] * tv
            )

            opt.zero_grad()
            loss.backward()
            opt.step()

            loss_sum += float(loss.item())
            recon_sum += float(recon.item())
            bce_sum += float(bce.item())
            con_sum += float(con.item())
            sp_sum += float(sp.item())
            tv_sum += float(tv.item())
            n_batches += 1

        if ep == 1 or ep % 10 == 0:
            print(
                f"[act={activity} | test_sub={test_sub} | ep={ep}] "
                f"Loss={loss_sum/n_batches:.4f} | "
                f"recon={recon_sum/n_batches:.4f} | bce={bce_sum/n_batches:.4f} | "
                f"contrast={con_sum/n_batches:.4f} | sparsity={sp_sum/n_batches:.4f} | tv={tv_sum/n_batches:.4f}"
            )

    metrics = eval_fold(model, test_ds, cfg, activity, test_sub)
    return model, metrics


# ============================================================
# 10) RUN: per-activity LOSO
# ============================================================
def run_all(cfg: Dict):
    set_seed(cfg["seed"])

    overall_maes, overall_mapes = [], []
    per_act_summary = {}

    for act in cfg["activities"]:
        print("=" * 80)
        print(f"[BUILD] activity={act}")
        trials = build_trials_for_activity(cfg["data_dir"], act, cfg["subjects"])
        print(f"  trials: {len(trials)} subjects")

        missing = [f"subject{s}" for s in cfg["subjects"] if f"subject{s}" not in trials]
        if len(missing):
            print("[WARN] missing trials:", missing)

        fold_mae = []
        fold_mape = []
        fold_bce = []
        fold_corr = []
        fold_auc = []

        keys = sorted(list(trials.keys()), key=lambda x: int(x.replace("subject", "")))

        for fi, test_sub in enumerate(keys):
            train_trials = {k: v for k, v in trials.items() if k != test_sub}
            test_trials  = {test_sub: trials[test_sub]}

            train_ds = TrialWindowDataset(train_trials, cfg)
            test_ds  = TrialWindowDataset(test_trials, cfg)

            model, metrics = train_one_fold(train_ds, test_ds, cfg, act, test_sub)

            fold_mae.append(metrics["mae"])
            fold_mape.append(metrics["mape"])
            fold_bce.append(metrics["bce"])
            fold_corr.append(metrics["corr"])
            fold_auc.append(metrics["auc"])

            print(
                f"[EVAL act={act} | test_sub={test_sub}] "
                f"pseudo: BCE={metrics['bce']:.4f}, corr={metrics['corr']:.4f}, AUC(top/bot)={metrics['auc']:.4f} | "
                f"count: pred={metrics['count_pred']:.2f}, gt={metrics['count_gt']:.2f}, "
                f"MAE={metrics['mae']:.2f}, MAPE={metrics['mape']:.2f}%"
            )

            do_viz = cfg["viz_every_fold"] or (fi == len(keys) - 1)
            if do_viz:
                pca_scatter(
                    latent=metrics["latent"],
                    pseudo=metrics["pseudo_w"],
                    title=f"PCA latent | act={act} | test_sub={test_sub}",
                    tb_frac=cfg["tb_frac"]
                )

        per_act_summary[act] = {
            "MAE": float(np.mean(fold_mae)) if len(fold_mae) else float("nan"),
            "MAPE": float(np.mean(fold_mape)) if len(fold_mape) else float("nan"),
            "pseudo_BCE": float(np.mean(fold_bce)) if len(fold_bce) else float("nan"),
            "pseudo_corr": float(np.mean(fold_corr)) if len(fold_corr) else float("nan"),
            "pseudo_AUC": float(np.mean(fold_auc)) if len(fold_auc) else float("nan"),
        }

        overall_maes += fold_mae
        overall_mapes += fold_mape

    print("\n" + "=" * 80)
    print("[SUMMARY] per-activity (LOSO mean)")
    for act, s in per_act_summary.items():
        print(
            f"  act={act} | MAE={s['MAE']:.2f} | MAPE={s['MAPE']:.2f}% | "
            f"pseudo(BCE={s['pseudo_BCE']:.4f}, corr={s['pseudo_corr']:.4f}, AUC={s['pseudo_AUC']:.4f})"
        )

    overall_mae = float(np.mean(overall_maes)) if len(overall_maes) else float("nan")
    overall_mape = float(np.mean(overall_mapes)) if len(overall_mapes) else float("nan")
    print(f"\n[OVERALL] MAE={overall_mae:.2f} | MAPE={overall_mape:.2f}% over {len(overall_maes)} folds")


# ============================================================
# 11) RUN
# ============================================================
if __name__ == "__main__":
    run_all(CONFIG)
```

```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass
from typing import Dict, List, Tuple

import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.metrics import roc_auc_score


# ============================================================
# 0) CONFIG
#   - autocorr-only counting (no count_mode / no method switch)
# ============================================================
CONFIG = {
    # data
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "fs": 50,
    "subjects": list(range(1, 11)),
    "activities": [6, 7, 8, 12],

    # windowing
    "window_size": 100,
    "stride": 50,

    # trial-wise normalization
    "trial_norm": True,
    "trial_norm_eps": 1e-6,

    # training
    "epochs": 50,
    "batch_size": 128,
    "lr": 1e-3,
    "seed": 42,
    "device": "cuda" if torch.cuda.is_available() else "cpu",

    # pseudo-event
    "beta_sigmoid": 1.0,
    "robust_eps": 1e-6,
    "grav_lp_win": 25,

    # top/bottom mining
    "tb_frac": 0.20,

    # losses (NO count loss!)
    "lambda_recon": 1.0,
    "lambda_bce": 1.0,
    "lambda_contrast": 0.3,
    "lambda_sparsity": 0.2,
    "lambda_tv": 0.2,

    # count estimation (EVAL ONLY) - autocorr only
    "count_smooth_win": 9,     # smoothing for frame-wise eventness before AC
    "min_bpm": 10,             # plausible repetition rate range (per minute)
    "max_bpm": 120,

    # visualization
    "viz_every_fold": False,   # False -> last fold only
    "viz_raw_channel": "ankle_gyro_z",  # which raw channel to plot
    "viz_show_seconds": True,
}


# ============================================================
# 1) USER GT COUNT TABLE (EVAL ONLY)
# ============================================================
COUNT_TABLE: Dict[int, Dict[str, int]] = {
    6: {
        "subject1": 21, "subject2": 19, "subject3": 21, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 21, "subject9": 21, "subject10": 20,
    },
    7: {
        "subject1": 20, "subject2": 20, "subject3": 20, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 19, "subject9": 19, "subject10": 20,
    },
    8: {
        "subject1": 20, "subject2": 21, "subject3": 21, "subject4": 19, "subject5": 20,
        "subject6": 20, "subject7": 21, "subject8": 21, "subject9": 21, "subject10": 21,
    },
    12: {
        "subject1": 40, "subject2": 44, "subject3": 42, "subject4": 42, "subject5": 40,
        "subject6": 42, "subject7": 38, "subject8": 40, "subject9": 40, "subject10": 40,
    },
}


# ============================================================
# 2) UTIL
# ============================================================
def set_seed(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def safe_sigmoid_np(x: np.ndarray):
    x = np.clip(x, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-x))


def robust_z(x: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    denom = 1.4826 * mad + eps
    return (x - med) / denom


def moving_average_1d(x: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return x
    win = int(win)
    kernel = np.ones(win, dtype=np.float32) / float(win)
    pad = win // 2
    xp = np.pad(x, (pad, pad), mode="reflect")
    return np.convolve(xp, kernel, mode="valid")


def moving_average_vec(X: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return X
    out = np.zeros_like(X)
    for c in range(X.shape[1]):
        out[:, c] = moving_average_1d(X[:, c], win)
    return out


def l2_normalize(v: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / (n + eps)


def pca_scatter(latent: np.ndarray, pseudo: np.ndarray, title: str, tb_frac: float = 0.2):
    if latent.shape[0] < 5:
        print("[PCA] too few samples")
        return

    p = pseudo.reshape(-1)
    n = len(p)
    k = max(2, int(tb_frac * n))
    idx_sorted = np.argsort(p)
    low_idx = idx_sorted[:k]
    high_idx = idx_sorted[-k:]

    labels = np.zeros(n, dtype=np.int64) - 1
    labels[low_idx] = 0
    labels[high_idx] = 1

    Z = PCA(n_components=2).fit_transform(latent)

    plt.figure(figsize=(7, 5))
    plt.scatter(Z[labels == 0, 0], Z[labels == 0, 1], s=8, alpha=0.7, label="steady (pseudo low)")
    plt.scatter(Z[labels == 1, 0], Z[labels == 1, 1], s=8, alpha=0.7, label="transition (pseudo high)")
    plt.title(title)
    plt.legend()
    plt.show()


def trialwise_normalize(X: np.ndarray, eps: float = 1e-6) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    mu = X.mean(axis=0, keepdims=True)
    std = X.std(axis=0, keepdims=True)
    Xn = (X - mu) / (std + eps)
    return Xn.astype(np.float32), mu.astype(np.float32), std.astype(np.float32)


# ============================================================
# 3) mHealth LOADING + TRIAL EXTRACTION
# ============================================================
def _read_mhealth_subject_log(data_dir: str, subject_id: int) -> np.ndarray:
    fname = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    arr = np.loadtxt(fname)
    return arr


def _extract_activity_segments(arr: np.ndarray, activity_code: int) -> List[Tuple[int, int]]:
    labels = arr[:, -1].astype(int)
    idx = np.where(labels == activity_code)[0]
    if len(idx) == 0:
        return []
    segments = []
    s = idx[0]
    prev = idx[0]
    for i in idx[1:]:
        if i == prev + 1:
            prev = i
        else:
            segments.append((s, prev + 1))
            s = i
            prev = i
    segments.append((s, prev + 1))
    return segments


def _get_sensor_6ch(arr: np.ndarray, which: str) -> np.ndarray:
    if which == "chest":
        return arr[:, 0:6]
    if which == "ankle":
        return arr[:, 6:12]
    if which == "arm":
        return arr[:, 12:18]
    raise ValueError(which)


def build_trials_for_activity(data_dir: str, activity: int, subjects: List[int]) -> Dict[str, np.ndarray]:
    """
    Returns RAW trials: { "subjectX": (T,18) }
    """
    trials = {}
    for sid in subjects:
        arr = _read_mhealth_subject_log(data_dir, sid)
        segs = _extract_activity_segments(arr, activity)
        if len(segs) == 0:
            continue
        seg = max(segs, key=lambda x: x[1] - x[0])  # longest segment
        s, e = seg
        chest = _get_sensor_6ch(arr[s:e], "chest")
        ankle = _get_sensor_6ch(arr[s:e], "ankle")
        arm = _get_sensor_6ch(arr[s:e], "arm")
        X = np.concatenate([chest, ankle, arm], axis=1)  # (T,18)
        trials[f"subject{sid}"] = X.astype(np.float32)
    return trials


# ============================================================
# 4) PSEUDO-EVENTNESS (computed on RAW)
# ============================================================
def pseudo_eventness_two_defs(
    X18_raw: np.ndarray,
    fs: int,
    beta: float,
    eps: float,
    grav_lp_win: int,
) -> np.ndarray:
    T = X18_raw.shape[0]
    if T < 5:
        return np.zeros((T,), dtype=np.float32)

    # 9ch gyro, 9ch acc (3 sensors x 3 axes)
    gyro = np.concatenate([X18_raw[:, 3:6], X18_raw[:, 9:12], X18_raw[:, 15:18]], axis=1)  # (T,9)
    acc  = np.concatenate([X18_raw[:, 0:3], X18_raw[:, 6:9], X18_raw[:, 12:15]], axis=1)   # (T,9)

    # Def-1) rotation jerk proxy (|d gyro / dt|)
    dgyro = np.diff(gyro, axis=0) * fs                          # (T-1, 9)
    jerk1 = np.linalg.norm(dgyro, axis=1).astype(np.float32)    # (T-1,)
    jerk1 = np.concatenate([[jerk1[0]], jerk1], axis=0)         # (T,)
    z1 = robust_z(jerk1, eps=eps).astype(np.float32)

    # Def-2) gravity direction instability (angle between LP acc dirs)
    acc_lp = moving_average_vec(acc.astype(np.float32), grav_lp_win)  # (T,9)
    ghat = l2_normalize(acc_lp.astype(np.float32))                    # (T,9)

    dot = np.sum(ghat[1:] * ghat[:-1], axis=1).astype(np.float32)     # (T-1,)
    dot = np.clip(dot, -1.0, 1.0)
    ang = np.arccos(dot).astype(np.float32)                           # (T-1,)
    instab2 = np.concatenate([[ang[0]], ang], axis=0) * fs            # (T,)
    z2 = robust_z(instab2, eps=eps).astype(np.float32)

    z = (z1 + z2).astype(np.float32)
    p = safe_sigmoid_np(beta * z).astype(np.float32)

    return p



# ============================================================
# 5) DATASET (WINDOWS) + TRIAL-WISE NORM
# ============================================================
@dataclass
class WindowItem:
    x: torch.Tensor
    p: torch.Tensor
    trial_key: str
    start: int
    W: int


class TrialWindowDataset(torch.utils.data.Dataset):
    def __init__(self, trials_raw: Dict[str, np.ndarray], cfg: Dict):
        self.cfg = cfg
        self.items: List[WindowItem] = []
        self.trials_raw = trials_raw  # keep for qualitative viz
        self.trial_len = {k: v.shape[0] for k, v in trials_raw.items()}

        W = cfg["window_size"]
        S = cfg["stride"]

        for trial_key, X18_raw in trials_raw.items():
            # pseudo from RAW (physics)
            p = pseudo_eventness_two_defs(
                X18_raw=X18_raw,
                fs=cfg["fs"],
                beta=cfg["beta_sigmoid"],
                eps=cfg["robust_eps"],
                grav_lp_win=cfg["grav_lp_win"],
            )

            # model input normalization (trial-wise)
            if cfg.get("trial_norm", True):
                X18, _, _ = trialwise_normalize(X18_raw, eps=cfg["trial_norm_eps"])
            else:
                X18 = X18_raw.astype(np.float32)

            T = X18.shape[0]
            if T < W:
                continue

            for st in range(0, T - W + 1, S):
                xw = X18[st:st + W]       # (W,18) normalized
                pw = p[st:st + W]         # (W,) pseudo from RAW
                xw_t = torch.from_numpy(xw).float().transpose(0, 1)  # (18,W)
                pw_t = torch.from_numpy(pw).float()                  # (W,)
                self.items.append(WindowItem(x=xw_t, p=pw_t, trial_key=trial_key, start=st, W=W))

    def __len__(self):
        return len(self.items)

    def __getitem__(self, idx: int):
        it = self.items[idx]
        return it.x, it.p, it.trial_key, it.start, it.W


def collate_fn(batch):
    xs, ps, keys, starts, Ws = zip(*batch)
    X = torch.stack(xs, dim=0)
    P = torch.stack(ps, dim=0)
    return X, P, list(keys), torch.tensor(starts, dtype=torch.long), torch.tensor(Ws, dtype=torch.long)


# ============================================================
# 6) MODEL: single 18ch encoder (NO GATE)
# ============================================================
class Encoder18(nn.Module):
    def __init__(self, in_ch: int, hid: int, out_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_ch, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_dim),
            nn.GELU(),
        )

    def forward(self, x):
        return self.net(x)


class Decoder(nn.Module):
    def __init__(self, in_dim: int, hid: int, out_ch: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_dim, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_ch, kernel_size=3, padding=1),
        )

    def forward(self, z):
        return self.net(z)


class EventModel(nn.Module):
    def __init__(self, latent_dim: int = 64, hid: int = 128):
        super().__init__()
        self.enc = Encoder18(18, hid, latent_dim)
        self.dec = Decoder(latent_dim, hid, out_ch=18)
        self.event_head = nn.Sequential(
            nn.Conv1d(latent_dim, latent_dim, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv1d(latent_dim, 1, kernel_size=1),
        )

    def forward(self, x18):  # (B,18,W)
        z = self.enc(x18)                       # (B,D,W)
        xhat = self.dec(z)                      # (B,18,W)
        elogit = self.event_head(z).squeeze(1)  # (B,W)
        return z, xhat, elogit


# ============================================================
# 7) LOSSES
# ============================================================
def contrast_top_bottom(z: torch.Tensor, pseudo_w: torch.Tensor, tb_frac: float):
    B = z.shape[0]
    if B < 8:
        return z.new_tensor(0.0)

    s = pseudo_w.mean(dim=1)
    k = max(2, int(tb_frac * B))
    idx = torch.argsort(s)
    low = idx[:k]
    high = idx[-k:]

    zw = z.mean(dim=2)  # (B,D)
    z_low = F.normalize(zw[low], dim=1)
    z_high = F.normalize(zw[high], dim=1)

    sim_ll = torch.mm(z_low, z_low.t())
    sim_hh = torch.mm(z_high, z_high.t())
    sim_lh = torch.mm(z_low, z_high.t())

    eye_l = torch.eye(sim_ll.shape[0], device=sim_ll.device)
    eye_h = torch.eye(sim_hh.shape[0], device=sim_hh.device)
    sim_ll = sim_ll * (1 - eye_l)
    sim_hh = sim_hh * (1 - eye_h)

    loss = (-sim_ll.mean() - sim_hh.mean() + sim_lh.mean())
    return loss


def total_variation(p: torch.Tensor):
    return torch.mean(torch.abs(p[:, 1:] - p[:, :-1]))


# ============================================================
# 8) FRAME-WISE RECONSTRUCTION (overlap-add)
# ============================================================
def overlap_add_series(T: int, starts: List[int], win_series: List[np.ndarray]) -> np.ndarray:
    """
    Reconstruct frame-wise series of length T from many window-level sequences.
    win_series: list of (W,) arrays aligned to each start.
    """
    acc = np.zeros((T,), dtype=np.float32)
    cnt = np.zeros((T,), dtype=np.float32)
    for st, wseq in zip(starts, win_series):
        W = len(wseq)
        ed = min(T, st + W)
        seg = wseq[:(ed - st)]
        acc[st:ed] += seg
        cnt[st:ed] += 1.0
    out = acc / (cnt + 1e-8)
    return out


# ============================================================
# 9) COUNT ESTIMATION (EVAL ONLY) : Autocorr ONLY
# ============================================================
def _valid_freq_range(fs: int, min_bpm: float, max_bpm: float) -> Tuple[float, float]:
    fmin = (min_bpm / 60.0)
    fmax = (max_bpm / 60.0)
    fmax = min(fmax, fs / 2.0 - 1e-6)
    return fmin, max(fmin + 1e-6, fmax)


def estimate_count_autocorr(x: np.ndarray, fs: int, min_bpm: float, max_bpm: float) -> Tuple[float, float, float]:
    """
    Returns (count_est, score, period_sec)
    score ~ peak / mean(ac_lags_range)
    """
    x = x.astype(np.float32)
    x = x - np.mean(x)
    T = len(x)
    if T < fs * 2:
        return 0.0, 0.0, 0.0

    # normalized autocorr (biased)
    ac = np.correlate(x, x, mode="full")[T-1:]
    ac = ac / (ac[0] + 1e-8)

    # lag range
    fmin, fmax = _valid_freq_range(fs, min_bpm, max_bpm)
    lag_max = int(np.floor(fs / fmin))  # longest period
    lag_min = int(np.ceil(fs / fmax))   # shortest period
    lag_max = min(lag_max, T - 2)
    lag_min = max(1, min(lag_min, lag_max - 1))

    seg = ac[lag_min:lag_max+1]
    if len(seg) < 5:
        return 0.0, 0.0, 0.0

    k = int(np.argmax(seg))
    lag = lag_min + k
    peak = float(seg[k])
    meanv = float(np.mean(seg) + 1e-8)
    score = peak / meanv

    period_sec = lag / float(fs)
    duration_sec = T / float(fs)
    count = duration_sec / max(1e-8, period_sec)
    return float(count), float(score), float(period_sec)


def estimate_count_autocorr_only(x: np.ndarray, fs: int, cfg: Dict) -> Dict[str, float]:
    """
    Autocorr-only counting wrapper.
    x: frame-wise eventness
    """
    xs = moving_average_1d(x, win=cfg["count_smooth_win"])
    c, s, p0 = estimate_count_autocorr(xs, fs, cfg["min_bpm"], cfg["max_bpm"])
    return {"count": c, "score": s, "period_sec": p0}


# ============================================================
# 10) TRAIN + EVAL (LOSO per activity)
# ============================================================
@torch.no_grad()
def eval_fold(model: nn.Module, ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )
    device = cfg["device"]

    all_pred_w, all_pseudo_w, all_latent = [], [], []
    all_keys, all_starts = [], []
    all_pred_seq, all_pseudo_seq = [], []

    bce_sum, n_sum = 0.0, 0

    for X, P, keys, starts, _ in loader:
        X = X.to(device)
        P = P.to(device)

        z, _, elogit = model(X)
        prob = torch.sigmoid(elogit)

        bce = F.binary_cross_entropy(prob, P, reduction="mean")
        bce_sum += float(bce.item()) * X.size(0)
        n_sum += X.size(0)

        # store window-mean for pseudo-alignment metrics
        all_pred_w.append(prob.mean(dim=1).detach().cpu().numpy())
        all_pseudo_w.append(P.mean(dim=1).detach().cpu().numpy())
        all_latent.append(z.mean(dim=2).detach().cpu().numpy())

        # store sequences for frame-wise reconstruction
        for bi in range(prob.size(0)):
            all_keys.append(keys[bi])
            all_starts.append(int(starts[bi].item()))
            all_pred_seq.append(prob[bi].detach().cpu().numpy().astype(np.float32))
            all_pseudo_seq.append(P[bi].detach().cpu().numpy().astype(np.float32))

    pred_w = np.concatenate(all_pred_w, axis=0)
    pseudo_w = np.concatenate(all_pseudo_w, axis=0)
    latent = np.concatenate(all_latent, axis=0)

    corr = np.corrcoef(pred_w, pseudo_w)[0, 1] if len(pred_w) > 3 else 0.0

    n = len(pseudo_w)
    k = max(5, int(cfg["tb_frac"] * n))
    idx = np.argsort(pseudo_w)
    y = np.zeros(n, dtype=np.int64) - 1
    y[idx[:k]] = 0
    y[idx[-k:]] = 1
    mask = y >= 0
    try:
        auc = roc_auc_score(y[mask], pred_w[mask])
    except Exception:
        auc = 0.5

    # -------- frame-wise reconstruction for the single test trial --------
    X18_raw = ds.trials_raw[test_sub]
    T = X18_raw.shape[0]

    starts, pred_seqs, pseudo_seqs = [], [], []
    for key, st, pseq, qseq in zip(all_keys, all_starts, all_pred_seq, all_pseudo_seq):
        if key == test_sub:
            starts.append(st)
            pred_seqs.append(pseq)
            pseudo_seqs.append(qseq)

    pred_frame = overlap_add_series(T, starts, pred_seqs)
    pseudo_frame = overlap_add_series(T, starts, pseudo_seqs)

    # autocorr-only count
    count_info = estimate_count_autocorr_only(pred_frame, cfg["fs"], cfg)
    c_pred = float(count_info["count"])
    gt = float(COUNT_TABLE[activity][test_sub])
    mae = abs(c_pred - gt)
    mape = mae / (abs(gt) + 1e-8) * 100.0

    return {
        "bce": float(bce_sum / max(1, n_sum)),
        "corr": float(corr),
        "auc": float(auc),

        "count_pred": float(c_pred),
        "count_gt": float(gt),
        "mae": float(mae),
        "mape": float(mape),

        "count_score": float(count_info["score"]),
        "count_period_sec": float(count_info["period_sec"]),

        "latent": latent,
        "pseudo_w": pseudo_w,
        "pred_w": pred_w,

        "pred_frame": pred_frame,
        "pseudo_frame": pseudo_frame,
        "X18_raw": X18_raw,
    }


def train_one_fold(train_ds: TrialWindowDataset, test_ds: TrialWindowDataset, cfg: Dict, activity: int, test_sub: str):
    device = cfg["device"]
    model = EventModel(latent_dim=64, hid=128).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=cfg["lr"])

    train_loader = torch.utils.data.DataLoader(
        train_ds, batch_size=cfg["batch_size"], shuffle=True, collate_fn=collate_fn, drop_last=True
    )

    for ep in range(1, cfg["epochs"] + 1):
        model.train()

        loss_sum = recon_sum = bce_sum = con_sum = sp_sum = tv_sum = 0.0
        n_batches = 0

        for X, P, _, _, _ in train_loader:
            X = X.to(device)
            P = P.to(device)

            z, xhat, elogit = model(X)
            prob = torch.sigmoid(elogit)

            recon = F.mse_loss(xhat, X)
            bce = F.binary_cross_entropy(prob, P)
            con = contrast_top_bottom(z, P, tb_frac=cfg["tb_frac"])
            sp = prob.mean()
            tv = total_variation(prob)

            loss = (
                cfg["lambda_recon"] * recon +
                cfg["lambda_bce"] * bce +
                cfg["lambda_contrast"] * con +
                cfg["lambda_sparsity"] * sp +
                cfg["lambda_tv"] * tv
            )

            opt.zero_grad()
            loss.backward()
            opt.step()

            loss_sum += float(loss.item())
            recon_sum += float(recon.item())
            bce_sum += float(bce.item())
            con_sum += float(con.item())
            sp_sum += float(sp.item())
            tv_sum += float(tv.item())
            n_batches += 1

        if ep == 1 or ep % 10 == 0:
            print(
                f"[act={activity} | test_sub={test_sub} | ep={ep}] "
                f"Loss={loss_sum/n_batches:.4f} | "
                f"recon={recon_sum/n_batches:.4f} | bce={bce_sum/n_batches:.4f} | "
                f"contrast={con_sum/n_batches:.4f} | sparsity={sp_sum/n_batches:.4f} | tv={tv_sum/n_batches:.4f}"
            )

    metrics = eval_fold(model, test_ds, cfg, activity, test_sub)
    return model, metrics


# ============================================================
# 11) RUN: per-activity LOSO
# ============================================================
def run_all(cfg: Dict):
    set_seed(cfg["seed"])

    overall_maes, overall_mapes = [], []
    per_act_summary = {}

    for act in cfg["activities"]:
        print("=" * 80)
        print(f"[BUILD] activity={act}")
        trials = build_trials_for_activity(cfg["data_dir"], act, cfg["subjects"])
        print(f"  trials: {len(trials)} subjects")

        missing = [f"subject{s}" for s in cfg["subjects"] if f"subject{s}" not in trials]
        if len(missing):
            print("[WARN] missing trials:", missing)

        fold_mae, fold_mape = [], []
        fold_bce, fold_corr, fold_auc = [], [], []

        keys = sorted(list(trials.keys()), key=lambda x: int(x.replace("subject", "")))

        for fi, test_sub in enumerate(keys):
            train_trials = {k: v for k, v in trials.items() if k != test_sub}
            test_trials  = {test_sub: trials[test_sub]}

            train_ds = TrialWindowDataset(train_trials, cfg)
            test_ds  = TrialWindowDataset(test_trials, cfg)

            _, metrics = train_one_fold(train_ds, test_ds, cfg, act, test_sub)

            fold_mae.append(metrics["mae"])
            fold_mape.append(metrics["mape"])
            fold_bce.append(metrics["bce"])
            fold_corr.append(metrics["corr"])
            fold_auc.append(metrics["auc"])

            print(
                f"[EVAL act={act} | test_sub={test_sub}] "
                f"pseudo: BCE={metrics['bce']:.4f}, corr={metrics['corr']:.4f}, AUC(top/bot)={metrics['auc']:.4f} | "
                f"count(autocorr score={metrics['count_score']:.2f}, period={metrics['count_period_sec']:.3f}s): "
                f"pred={metrics['count_pred']:.2f}, gt={metrics['count_gt']:.2f}, "
                f"MAE={metrics['mae']:.2f}, MAPE={metrics['mape']:.2f}%"
            )

            do_viz = cfg["viz_every_fold"] or (fi == len(keys) - 1)
            if do_viz:
                pca_scatter(
                    latent=metrics["latent"],
                    pseudo=metrics["pseudo_w"],
                    title=f"PCA latent | act={act} | test_sub={test_sub}",
                    tb_frac=cfg["tb_frac"]
                )

        per_act_summary[act] = {
            "MAE": float(np.mean(fold_mae)) if len(fold_mae) else float("nan"),
            "MAPE": float(np.mean(fold_mape)) if len(fold_mape) else float("nan"),
            "pseudo_BCE": float(np.mean(fold_bce)) if len(fold_bce) else float("nan"),
            "pseudo_corr": float(np.mean(fold_corr)) if len(fold_corr) else float("nan"),
            "pseudo_AUC": float(np.mean(fold_auc)) if len(fold_auc) else float("nan"),
        }

        overall_maes += fold_mae
        overall_mapes += fold_mape

    print("\n" + "=" * 80)
    print("[SUMMARY] per-activity (LOSO mean)")
    for act, s in per_act_summary.items():
        print(
            f"  act={act} | MAE={s['MAE']:.2f} | MAPE={s['MAPE']:.2f}% | "
            f"pseudo(BCE={s['pseudo_BCE']:.4f}, corr={s['pseudo_corr']:.4f}, AUC={s['pseudo_AUC']:.4f})"
        )

    overall_mae = float(np.mean(overall_maes)) if len(overall_maes) else float("nan")
    overall_mape = float(np.mean(overall_mapes)) if len(overall_mapes) else float("nan")
    print(f"\n[OVERALL] MAE={overall_mae:.2f} | MAPE={overall_mape:.2f}% over {len(overall_maes)} folds")


# ============================================================
# 12) RUN
# ============================================================
if __name__ == "__main__":
    run_all(CONFIG)
```
