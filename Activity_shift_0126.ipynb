```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from dataclasses import dataclass
from typing import Dict, List, Tuple

import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score


# =============================================================================
# 0) CONFIG
# =============================================================================
CONFIG = {
    "data_dir": "/content/drive/MyDrive/Colab Notebooks/HAR_data/MHEALTHDATASET",
    "fs": 50,
    "subjects": list(range(1, 11)),
    "activities": [6, 7, 8, 12],

    "window_size": 50,
    "stride": 25,

    "trial_norm": True,
    "trial_norm_eps": 1e-6,

    "epochs": 50,
    "batch_size": 128,
    "lr": 1e-3,
    "seed": 42,
    "device": "cuda" if torch.cuda.is_available() else "cpu",

    "beta_sigmoid": 1.0,
    "robust_eps": 1e-6,
    "grav_lp_win": 25,

    "tb_frac": 0.20,

    "lambda_recon": 1.0,
    "lambda_bce": 1.0,
    "lambda_contrast": 0.3,
    "lambda_sparsity": 0.2,
    "lambda_tv": 0.2,

    "count_smooth_win": 9,
    "min_bpm": 10,
    "max_bpm": 120,
}


# =============================================================================
# 1) USER GT COUNT TABLE (EVAL ONLY)
# =============================================================================
COUNT_TABLE: Dict[int, Dict[str, int]] = {
    6: {
        "subject1": 21, "subject2": 19, "subject3": 21, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 21, "subject9": 21, "subject10": 20,
    },
    7: {
        "subject1": 20, "subject2": 20, "subject3": 20, "subject4": 20, "subject5": 20,
        "subject6": 20, "subject7": 20, "subject8": 19, "subject9": 19, "subject10": 20,
    },
    8: {
        "subject1": 20, "subject2": 21, "subject3": 21, "subject4": 19, "subject5": 20,
        "subject6": 20, "subject7": 21, "subject8": 21, "subject9": 21, "subject10": 21,
    },
    12: {
        "subject1": 40, "subject2": 44, "subject3": 42, "subject4": 42, "subject5": 40,
        "subject6": 42, "subject7": 38, "subject8": 40, "subject9": 40, "subject10": 40,
    },
}


# =============================================================================
# 2) UTILS (PURE FUNCTIONS)
# =============================================================================
def set_seed(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def safe_sigmoid_np(x: np.ndarray):
    x = np.clip(x, -30.0, 30.0)
    return 1.0 / (1.0 + np.exp(-x))


def robust_z(x: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    denom = 1.4826 * mad + eps
    return (x - med) / denom


def moving_average_1d(x: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return x
    win = int(win)
    kernel = np.ones(win, dtype=np.float32) / float(win)
    pad = win // 2
    xp = np.pad(x, (pad, pad), mode="reflect")
    return np.convolve(xp, kernel, mode="valid")


def moving_average_vec(X: np.ndarray, win: int) -> np.ndarray:
    if win <= 1:
        return X
    out = np.zeros_like(X)
    for c in range(X.shape[1]):
        out[:, c] = moving_average_1d(X[:, c], win)
    return out


def l2_normalize(v: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / (n + eps)


def trialwise_normalize(X: np.ndarray, eps: float = 1e-6) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    mu = X.mean(axis=0, keepdims=True)
    std = X.std(axis=0, keepdims=True)
    Xn = (X - mu) / (std + eps)
    return Xn.astype(np.float32), mu.astype(np.float32), std.astype(np.float32)


def chest_acc_mag(X18_raw: np.ndarray) -> np.ndarray:
    v = X18_raw[:, 0:3].astype(np.float32)
    return np.linalg.norm(v, axis=1).astype(np.float32)


def mask_to_spans(mask: np.ndarray) -> List[Tuple[int, int]]:
    mask = mask.astype(bool)
    if mask.sum() == 0:
        return []
    idx = np.where(mask)[0]
    spans = []
    s = idx[0]
    prev = idx[0]
    for i in idx[1:]:
        if i == prev + 1:
            prev = i
        else:
            spans.append((s, prev + 1))
            s = i
            prev = i
    spans.append((s, prev + 1))
    return spans


# =============================================================================
# 3) mHealth LOADING + TRIAL EXTRACTION (RAW TRIALS ONLY)
# =============================================================================
def _read_mhealth_subject_log(data_dir: str, subject_id: int) -> np.ndarray:
    fname = os.path.join(data_dir, f"mHealth_subject{subject_id}.log")
    return np.loadtxt(fname)


def _extract_activity_segments(arr: np.ndarray, activity_code: int) -> List[Tuple[int, int]]:
    labels = arr[:, -1].astype(int)
    idx = np.where(labels == activity_code)[0]
    if len(idx) == 0:
        return []
    segments = []
    s = idx[0]
    prev = idx[0]
    for i in idx[1:]:
        if i == prev + 1:
            prev = i
        else:
            segments.append((s, prev + 1))
            s = i
            prev = i
    segments.append((s, prev + 1))
    return segments


def _get_sensor_6ch(arr: np.ndarray, which: str) -> np.ndarray:
    if which == "chest":
        return arr[:, 0:6]
    if which == "ankle":
        return arr[:, 6:12]
    if which == "arm":
        return arr[:, 12:18]
    raise ValueError(which)


def build_trials_for_activity(data_dir: str, activity: int, subjects: List[int]) -> Dict[str, np.ndarray]:
    trials = {}
    for sid in subjects:
        arr = _read_mhealth_subject_log(data_dir, sid)
        segs = _extract_activity_segments(arr, activity)
        if len(segs) == 0:
            continue
        s, e = max(segs, key=lambda x: x[1] - x[0])  # longest segment (same as your code)
        chest = _get_sensor_6ch(arr[s:e], "chest")
        ankle = _get_sensor_6ch(arr[s:e], "ankle")
        arm = _get_sensor_6ch(arr[s:e], "arm")
        X = np.concatenate([chest, ankle, arm], axis=1)
        trials[f"subject{sid}"] = X.astype(np.float32)
    return trials


# =============================================================================
# 4) PSEUDO-EVENTNESS (STATE-CHANGE MAGNITUDE)  [UNCHANGED]
# =============================================================================
def pseudo_eventness_two_defs(
    X18_raw: np.ndarray,
    fs: int,
    beta: float,
    eps: float,
    grav_lp_win: int,
) -> np.ndarray:
    T = X18_raw.shape[0]
    if T < 5:
        return np.zeros((T,), dtype=np.float32)

    gyro = np.concatenate([X18_raw[:, 3:6], X18_raw[:, 9:12], X18_raw[:, 15:18]], axis=1)
    acc = np.concatenate([X18_raw[:, 0:3], X18_raw[:, 6:9], X18_raw[:, 12:15]], axis=1)

    dgyro = np.diff(gyro, axis=0) * fs
    jerk1 = np.linalg.norm(dgyro, axis=1).astype(np.float32)
    jerk1 = np.concatenate([[jerk1[0]], jerk1], axis=0)
    z1 = robust_z(jerk1, eps=eps).astype(np.float32)

    acc_lp = moving_average_vec(acc.astype(np.float32), grav_lp_win)
    ghat = l2_normalize(acc_lp.astype(np.float32))

    dot = np.sum(ghat[1:] * ghat[:-1], axis=1).astype(np.float32)
    dot = np.clip(dot, -1.0, 1.0)
    ang = np.arccos(dot).astype(np.float32)
    instab2 = np.concatenate([[ang[0]], ang], axis=0) * fs
    z2 = robust_z(instab2, eps=eps).astype(np.float32)

    z = (z1 + z2).astype(np.float32)
    p = safe_sigmoid_np(beta * z).astype(np.float32)
    return p


# =============================================================================
# 5) DATASET (WINDOWS)
# =============================================================================
@dataclass
class WindowItem:
    x: torch.Tensor
    p: torch.Tensor
    trial_key: str
    start: int
    W: int


class TrialWindowDataset(torch.utils.data.Dataset):
    def __init__(self, trials_raw: Dict[str, np.ndarray], cfg: Dict):
        self.cfg = cfg
        self.items: List[WindowItem] = []
        self.trials_raw = trials_raw  # keep raw for eval/viz

        W = cfg["window_size"]
        S = cfg["stride"]

        for trial_key, X18_raw in trials_raw.items():
            p = pseudo_eventness_two_defs(
                X18_raw=X18_raw,
                fs=cfg["fs"],
                beta=cfg["beta_sigmoid"],
                eps=cfg["robust_eps"],
                grav_lp_win=cfg["grav_lp_win"],
            )

            if cfg.get("trial_norm", True):
                X18, _, _ = trialwise_normalize(X18_raw, eps=cfg["trial_norm_eps"])
            else:
                X18 = X18_raw.astype(np.float32)

            T = X18.shape[0]
            if T < W:
                continue

            for st in range(0, T - W + 1, S):
                xw = X18[st:st + W]
                pw = p[st:st + W]
                xw_t = torch.from_numpy(xw).float().transpose(0, 1)  # (C, W)
                pw_t = torch.from_numpy(pw).float()                  # (W,)
                self.items.append(WindowItem(x=xw_t, p=pw_t, trial_key=trial_key, start=st, W=W))

    def __len__(self):
        return len(self.items)

    def __getitem__(self, idx: int):
        it = self.items[idx]
        return it.x, it.p, it.trial_key, it.start, it.W


def collate_fn(batch):
    xs, ps, keys, starts, Ws = zip(*batch)
    X = torch.stack(xs, dim=0)  # (B, C, W)
    P = torch.stack(ps, dim=0)  # (B, W)
    return X, P, list(keys), torch.tensor(starts, dtype=torch.long), torch.tensor(Ws, dtype=torch.long)


# =============================================================================
# 6) MODEL
# =============================================================================
class Encoder18(nn.Module):
    def __init__(self, in_ch: int, hid: int, out_dim: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_ch, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_dim, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_dim),
            nn.GELU(),
        )

    def forward(self, x):
        return self.net(x)


class Decoder(nn.Module):
    def __init__(self, in_dim: int, hid: int, out_ch: int):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv1d(in_dim, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, hid, kernel_size=5, padding=2),
            nn.BatchNorm1d(hid),
            nn.GELU(),
            nn.Conv1d(hid, out_ch, kernel_size=3, padding=1),
        )

    def forward(self, z):
        return self.net(z)


class EventModel(nn.Module):
    def __init__(self, latent_dim: int = 64, hid: int = 128):
        super().__init__()
        self.enc = Encoder18(18, hid, latent_dim)
        self.dec = Decoder(latent_dim, hid, out_ch=18)
        self.event_head = nn.Sequential(
            nn.Conv1d(latent_dim, latent_dim, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv1d(latent_dim, 1, kernel_size=1),
        )

    def forward(self, x18):
        z = self.enc(x18)
        xhat = self.dec(z)
        elogit = self.event_head(z).squeeze(1)  # (B, W)
        return z, xhat, elogit


# =============================================================================
# 7) LOSSES (REGISTRY STYLE)  [NUMERICALLY IDENTICAL]
# =============================================================================
def contrast_top_bottom(z: torch.Tensor, pseudo_w: torch.Tensor, tb_frac: float):
    B = z.shape[0]
    if B < 8:
        return z.new_tensor(0.0)

    s = pseudo_w.mean(dim=1)
    k = max(2, int(tb_frac * B))
    idx = torch.argsort(s)
    low = idx[:k]
    high = idx[-k:]

    zw = z.mean(dim=2)
    z_low = F.normalize(zw[low], dim=1)
    z_high = F.normalize(zw[high], dim=1)

    sim_ll = torch.mm(z_low, z_low.t())
    sim_hh = torch.mm(z_high, z_high.t())
    sim_lh = torch.mm(z_low, z_high.t())

    eye_l = torch.eye(sim_ll.shape[0], device=sim_ll.device)
    eye_h = torch.eye(sim_hh.shape[0], device=sim_hh.device)
    sim_ll = sim_ll * (1 - eye_l)
    sim_hh = sim_hh * (1 - eye_h)

    return (-sim_ll.mean() - sim_hh.mean() + sim_lh.mean())


def total_variation(p: torch.Tensor):
    return torch.mean(torch.abs(p[:, 1:] - p[:, :-1]))


def compute_loss_dict(z, xhat, elogit, X, P, cfg: Dict) -> Dict[str, torch.Tensor]:
    prob = torch.sigmoid(elogit)

    recon = F.mse_loss(xhat, X)
    bce = F.binary_cross_entropy(prob, P)
    con = contrast_top_bottom(z, P, tb_frac=cfg["tb_frac"])
    sp = prob.mean()
    tv = total_variation(prob)

    total = (
        cfg["lambda_recon"] * recon +
        cfg["lambda_bce"] * bce +
        cfg["lambda_contrast"] * con +
        cfg["lambda_sparsity"] * sp +
        cfg["lambda_tv"] * tv
    )

    return {
        "total": total,
        "recon": recon,
        "bce": bce,
        "contrast": con,
        "sparsity": sp,
        "tv": tv,
    }


# =============================================================================
# 8) FRAME-WISE RECONSTRUCTION
# =============================================================================
def overlap_add_series(T: int, starts: List[int], win_series: List[np.ndarray]) -> np.ndarray:
    acc = np.zeros((T,), dtype=np.float32)
    cnt = np.zeros((T,), dtype=np.float32)
    for st, wseq in zip(starts, win_series):
        W = len(wseq)
        ed = min(T, st + W)
        seg = wseq[:(ed - st)]
        acc[st:ed] += seg
        cnt[st:ed] += 1.0
    return acc / (cnt + 1e-8)


@torch.no_grad()
def infer_one_trial_framewise(model: nn.Module, ds_one: TrialWindowDataset, cfg: Dict, sub: str):
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds_one, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )

    device = cfg["device"]
    all_starts, all_pred_seq, all_pseudo_seq = [], [], []

    for X, P, keys, starts, _ in loader:
        X = X.to(device)
        P = P.to(device)
        _, _, elogit = model(X)
        prob = torch.sigmoid(elogit)

        for bi in range(prob.size(0)):
            if keys[bi] != sub:
                continue
            all_starts.append(int(starts[bi].item()))
            all_pred_seq.append(prob[bi].detach().cpu().numpy().astype(np.float32))
            all_pseudo_seq.append(P[bi].detach().cpu().numpy().astype(np.float32))

    X18_raw = ds_one.trials_raw[sub]
    T = X18_raw.shape[0]
    pred_frame = overlap_add_series(T, all_starts, all_pred_seq)
    pseudo_frame = overlap_add_series(T, all_starts, all_pseudo_seq)
    return pred_frame, pseudo_frame, X18_raw


@torch.no_grad()
def infer_all_trials_framewise(model: nn.Module, ds: TrialWindowDataset, cfg: Dict):
    """
    Returns:
      pred_frame_map[key], pseudo_frame_map[key], raw_map[key]
    """
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )
    device = cfg["device"]

    # collect windows per trial
    starts_map: Dict[str, List[int]] = {}
    predseq_map: Dict[str, List[np.ndarray]] = {}
    pseudoseq_map: Dict[str, List[np.ndarray]] = {}

    for X, P, keys, starts, _ in loader:
        X = X.to(device)
        P = P.to(device)
        _, _, elogit = model(X)
        prob = torch.sigmoid(elogit)

        for bi in range(prob.size(0)):
            k = keys[bi]
            starts_map.setdefault(k, []).append(int(starts[bi].item()))
            predseq_map.setdefault(k, []).append(prob[bi].detach().cpu().numpy().astype(np.float32))
            pseudoseq_map.setdefault(k, []).append(P[bi].detach().cpu().numpy().astype(np.float32))

    pred_frame_map = {}
    pseudo_frame_map = {}
    raw_map = {}

    for k, X18_raw in ds.trials_raw.items():
        raw_map[k] = X18_raw
        T = X18_raw.shape[0]
        st_list = starts_map.get(k, [])
        pr_list = predseq_map.get(k, [])
        ps_list = pseudoseq_map.get(k, [])
        pred_frame_map[k] = overlap_add_series(T, st_list, pr_list) if len(st_list) else np.zeros((T,), np.float32)
        pseudo_frame_map[k] = overlap_add_series(T, st_list, ps_list) if len(st_list) else np.zeros((T,), np.float32)

    return pred_frame_map, pseudo_frame_map, raw_map


# =============================================================================
# 9) COUNT ESTIMATION (AUTOCORR ONLY)
# =============================================================================
def _valid_freq_range(fs: int, min_bpm: float, max_bpm: float) -> Tuple[float, float]:
    fmin = (min_bpm / 60.0)
    fmax = (max_bpm / 60.0)
    fmax = min(fmax, fs / 2.0 - 1e-6)
    return fmin, max(fmin + 1e-6, fmax)


def estimate_count_autocorr_harmonic(
    x: np.ndarray,
    fs: int,
    min_bpm: float,
    max_bpm: float,
    h2_weight: float = 0.35,
    hhalf_weight: float = 0.15,
    fix_iters: int = 2,
    half_prefer_ratio: float = 0.92,
    double_prefer_ratio: float = 1.08,
    peak_margin: float = 0.02,
) -> Tuple[float, float, float]:
    x = x.astype(np.float32)
    x = x - np.mean(x)
    T = len(x)
    if T < fs * 2:
        return 0.0, 0.0, 0.0

    ac = np.correlate(x, x, mode="full")[T-1:]
    ac = ac / (ac[0] + 1e-8)

    fmin, fmax = _valid_freq_range(fs, min_bpm, max_bpm)
    lag_max = int(np.floor(fs / fmin))
    lag_min = int(np.ceil(fs / fmax))
    lag_max = min(lag_max, T - 2)
    lag_min = max(1, min(lag_min, lag_max - 1))

    seg = ac[lag_min:lag_max+1]
    if len(seg) < 5:
        return 0.0, 0.0, 0.0

    def ac_at(l: int) -> float:
        if l <= 0 or l >= len(ac):
            return -1e9
        return float(ac[l])

    best_lag = lag_min
    best_comb = -1e18
    for lag in range(lag_min, lag_max + 1):
        base = ac_at(lag)
        if base < -0.3:
            continue

        comb = base
        l2 = 2 * lag
        if l2 <= lag_max:
            comb += h2_weight * ac_at(l2)

        if (lag % 2 == 0):
            lhalf = lag // 2
            if lhalf >= lag_min:
                comb += hhalf_weight * ac_at(lhalf)

        if comb > best_comb:
            best_comb = comb
            best_lag = lag

    L = int(best_lag)
    for _ in range(int(fix_iters)):
        aL = ac_at(L)

        if L % 2 == 0:
            H = L // 2
            if H >= lag_min:
                aH = ac_at(H)
                if (aH >= half_prefer_ratio * aL - peak_margin):
                    L = H
                    continue

        D = 2 * L
        if D <= lag_max:
            aD = ac_at(D)
            if (aD >= double_prefer_ratio * aL + peak_margin):
                L = D
                continue

        break

    meanv = float(np.mean(seg) + 1e-8)
    aL = ac_at(L)
    a2 = ac_at(2 * L) if 2 * L <= lag_max else 0.0
    ah = ac_at(L // 2) if (L % 2 == 0 and (L // 2) >= lag_min) else 0.0
    final_comb = aL + h2_weight * a2 + hhalf_weight * ah
    score = float(final_comb / meanv)

    period_sec = L / float(fs)
    duration_sec = T / float(fs)
    count = duration_sec / max(1e-8, period_sec)

    return float(count), float(score), float(period_sec)


def estimate_count_autocorr_only(x: np.ndarray, fs: int, cfg: Dict) -> Dict[str, float]:
    xs = moving_average_1d(x, win=cfg["count_smooth_win"])
    c, s, p0 = estimate_count_autocorr_harmonic(
        xs, fs, cfg["min_bpm"], cfg["max_bpm"],
        h2_weight=0.35,
        hhalf_weight=0.15,
        fix_iters=2,
        half_prefer_ratio=0.92,
        double_prefer_ratio=1.08,
        peak_margin=0.02
    )
    return {"count": c, "score": s, "period_sec": p0}


# =============================================================================
# 10) EVAL HELPERS (METRICS + LATENT SPEED)
# =============================================================================
def latent_speed_window(z_win: torch.Tensor) -> torch.Tensor:
    dz = z_win[:, :, 1:] - z_win[:, :, :-1]
    s = torch.norm(dz, dim=1).mean(dim=1)
    return s


@torch.no_grad()
def eval_activity_dataset(model: nn.Module, ds: TrialWindowDataset, cfg: Dict, activity_eval: int):
    """
    Evaluate on a dataset that may contain multiple trials (subjects).
    Returns:
      - pseudo metrics over ALL windows (BCE, corr, AUC top/bot)
      - per-trial count metrics dict
      - plus last-batch latent/speed arrays for visualization (window-level)
    """
    model.eval()
    loader = torch.utils.data.DataLoader(
        ds, batch_size=cfg["batch_size"], shuffle=False, collate_fn=collate_fn
    )
    device = cfg["device"]

    all_pred_w, all_pseudo_w, all_speed = [], [], []
    all_latent = []

    bce_sum, n_sum = 0.0, 0

    for X, P, _, _, _ in loader:
        X = X.to(device)
        P = P.to(device)

        z, _, elogit = model(X)
        prob = torch.sigmoid(elogit)

        bce = F.binary_cross_entropy(prob, P, reduction="mean")
        bce_sum += float(bce.item()) * X.size(0)
        n_sum += X.size(0)

        all_pred_w.append(prob.mean(dim=1).detach().cpu().numpy())
        all_pseudo_w.append(P.mean(dim=1).detach().cpu().numpy())
        all_latent.append(z.mean(dim=2).detach().cpu().numpy())
        all_speed.append(latent_speed_window(z).detach().cpu().numpy())

    if n_sum == 0:
        # no windows
        return {
            "pseudo_BCE": float("nan"),
            "pseudo_corr": float("nan"),
            "pseudo_AUC": float("nan"),
            "pred_w": np.array([], dtype=np.float32),
            "pseudo_w": np.array([], dtype=np.float32),
            "latent": np.zeros((0, 64), dtype=np.float32),
            "speed": np.zeros((0,), dtype=np.float32),
            "per_trial": {},
        }

    pred_w = np.concatenate(all_pred_w, axis=0)
    pseudo_w = np.concatenate(all_pseudo_w, axis=0)
    latent = np.concatenate(all_latent, axis=0)
    speed = np.concatenate(all_speed, axis=0)

    corr = np.corrcoef(pred_w, pseudo_w)[0, 1] if len(pred_w) > 3 else 0.0

    n = len(pseudo_w)
    k = max(5, int(cfg["tb_frac"] * n))
    idx = np.argsort(pseudo_w)
    y = np.zeros(n, dtype=np.int64) - 1
    y[idx[:k]] = 0
    y[idx[-k:]] = 1
    mask = y >= 0
    try:
        auc = roc_auc_score(y[mask], pred_w[mask])
    except Exception:
        auc = 0.5

    # per-trial framewise + count
    pred_frame_map, _, _ = infer_all_trials_framewise(model, ds, cfg)

    per_trial = {}
    for trial_key, pred_frame in pred_frame_map.items():
        info = estimate_count_autocorr_only(pred_frame, cfg["fs"], cfg)
        c_pred = float(info["count"])
        gt = float(COUNT_TABLE[activity_eval][trial_key])
        mae = abs(c_pred - gt)
        mape = mae / (abs(gt) + 1e-8) * 100.0
        per_trial[trial_key] = {
            "count_pred": c_pred,
            "count_gt": gt,
            "mae": mae,
            "mape": mape,
            "count_score": float(info["score"]),
            "count_period_sec": float(info["period_sec"]),
        }

    return {
        "pseudo_BCE": float(bce_sum / max(1, n_sum)),
        "pseudo_corr": float(corr),
        "pseudo_AUC": float(auc),
        "pred_w": pred_w,
        "pseudo_w": pseudo_w,
        "latent": latent,
        "speed": speed,
        "per_trial": per_trial,
    }


# =============================================================================
# 11) TRAIN (ONCE PER SOURCE ACTIVITY)
# =============================================================================
def train_one_activity(train_ds: TrialWindowDataset, cfg: Dict, activity_train: int):
    device = cfg["device"]
    model = EventModel(latent_dim=64, hid=128).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=cfg["lr"])

    train_loader = torch.utils.data.DataLoader(
        train_ds, batch_size=cfg["batch_size"], shuffle=True, collate_fn=collate_fn, drop_last=True
    )

    for ep in range(1, cfg["epochs"] + 1):
        model.train()

        loss_sum = recon_sum = bce_sum = con_sum = sp_sum = tv_sum = 0.0
        n_batches = 0

        for X, P, _, _, _ in train_loader:
            X = X.to(device)
            P = P.to(device)

            z, xhat, elogit = model(X)
            loss_d = compute_loss_dict(z, xhat, elogit, X, P, cfg)

            opt.zero_grad()
            loss_d["total"].backward()
            opt.step()

            loss_sum += float(loss_d["total"].item())
            recon_sum += float(loss_d["recon"].item())
            bce_sum += float(loss_d["bce"].item())
            con_sum += float(loss_d["contrast"].item())
            sp_sum += float(loss_d["sparsity"].item())
            tv_sum += float(loss_d["tv"].item())
            n_batches += 1

        if ep == 1 or ep % 10 == 0:
            print(
                f"[TRAIN actA={activity_train} | ep={ep}] "
                f"Loss={loss_sum/n_batches:.4f} | "
                f"recon={recon_sum/n_batches:.4f} | bce={bce_sum/n_batches:.4f} | "
                f"contrast={con_sum/n_batches:.4f} | sparsity={sp_sum/n_batches:.4f} | tv={tv_sum/n_batches:.4f}"
            )

    return model


# =============================================================================
# 12) VISUALIZATION (TOP/BOTTOM ONLY)
# =============================================================================
def _top_bottom_masks(s: np.ndarray, tb_frac: float) -> Tuple[np.ndarray, np.ndarray]:
    s = s.reshape(-1)
    n = len(s)
    k = max(2, int(tb_frac * n))
    idx = np.argsort(s)
    low = np.zeros(n, dtype=bool)
    high = np.zeros(n, dtype=bool)
    low[idx[:k]] = True
    high[idx[-k:]] = True
    return low, high


def plot_raw_with_top_bottom_shades(
    act: int,
    sub: str,
    fs: int,
    X18_raw: np.ndarray,
    series: np.ndarray,
    tb_frac: float,
    title_suffix: str,
    smooth_win: int = 9,
    alpha: float = 0.20,
):
    T = X18_raw.shape[0]
    t = np.arange(T, dtype=np.float32) / float(fs)
    raw = chest_acc_mag(X18_raw)

    s = moving_average_1d(series.astype(np.float32), win=smooth_win)
    low_mask, high_mask = _top_bottom_masks(s, tb_frac)

    fig, ax = plt.subplots(1, 1, figsize=(18, 4))
    ax.plot(t, raw, lw=1.0)

    for a, b in mask_to_spans(low_mask):
        x0 = t[a]
        x1 = t[b] if b < T else t[-1]
        ax.axvspan(x0, x1, alpha=alpha, label="_nolegend_")

    for a, b in mask_to_spans(high_mask):
        x0 = t[a]
        x1 = t[b] if b < T else t[-1]
        ax.axvspan(x0, x1, alpha=alpha, hatch="///", label="_nolegend_")

    from matplotlib.patches import Patch
    handles = [
        Patch(alpha=alpha, label=f"steady: bottom {int(tb_frac*100)}%"),
        Patch(alpha=alpha, hatch="///", label=f"transition: top {int(tb_frac*100)}%"),
    ]
    ax.legend(handles=handles, loc="upper right")
    ax.set_title(f"act={act} | {sub} | {title_suffix}")
    ax.set_xlabel("time (sec)")
    ax.set_ylabel("chest_acc_mag")
    ax.grid(True, alpha=0.25)
    plt.tight_layout()
    plt.show()


def plot_latent_speed_boxplot(speed: np.ndarray, pseudo_w: np.ndarray, tb_frac: float, title: str):
    n = len(pseudo_w)
    k = max(2, int(tb_frac * n))
    idx = np.argsort(pseudo_w.reshape(-1))
    low = idx[:k]
    high = idx[-k:]

    data = [speed[low], speed[high]]
    plt.figure(figsize=(5.5, 4))
    plt.boxplot(data, labels=[f"steady\n(bottom {int(tb_frac*100)}%)", f"transition\n(top {int(tb_frac*100)}%)"])
    plt.title(title)
    plt.ylabel("latent speed (mean ||z_t - z_{t-1}||)")
    plt.grid(True, alpha=0.25)
    plt.tight_layout()
    plt.show()


def plot_count_scatter(preds: List[float], gts: List[float], title: str):
    plt.figure(figsize=(5.2, 5.2))
    plt.scatter(gts, preds, s=22, alpha=0.8)
    mn = min(min(gts), min(preds))
    mx = max(max(gts), max(preds))
    plt.plot([mn, mx], [mn, mx])
    plt.title(title)
    plt.xlabel("GT count")
    plt.ylabel("Pred count")
    plt.grid(True, alpha=0.25)
    plt.tight_layout()
    plt.show()


# =============================================================================
# 13) RUN (ACTIVITY SHIFT: Train on A (10 subj) -> Eval on B (10 subj))
#     - all ordered pairs A->B, exclude A==B
# =============================================================================
def run_all_activity_shift(cfg: Dict):
    set_seed(cfg["seed"])

    # pre-build trials for each activity once
    trials_by_act = {}
    for act in cfg["activities"]:
        trials = build_trials_for_activity(cfg["data_dir"], act, cfg["subjects"])
        keys = sorted(list(trials.keys()), key=lambda x: int(x.replace("subject", "")))
        trials_by_act[act] = (trials, keys)
        print("=" * 80)
        print(f"[BUILD] activity={act} | trials={len(trials)} subjects")

    pair_summary = {}  # (A,B) -> metrics

    for actA in cfg["activities"]:
        print("\n" + "#" * 92)
        print(f"[TRAIN ON] actA={actA} (10 subjects)")
        train_trials, train_keys = trials_by_act[actA]
        train_ds = TrialWindowDataset(train_trials, cfg)

        model = train_one_activity(train_ds, cfg, activity_train=actA)

        # evaluate on all B != A
        for actB in cfg["activities"]:
            if actB == actA:
                continue

            print("\n" + "-" * 92)
            print(f"[EVAL] actA={actA} -> actB={actB} (10 subjects)")

            test_trials, test_keys = trials_by_act[actB]
            test_ds = TrialWindowDataset(test_trials, cfg)

            metrics = eval_activity_dataset(model, test_ds, cfg, activity_eval=actB)

            # aggregate per-trial count results
            per_trial = metrics["per_trial"]
            maes = [per_trial[k]["mae"] for k in test_keys if k in per_trial]
            mapes = [per_trial[k]["mape"] for k in test_keys if k in per_trial]
            mean_mae = float(np.mean(maes)) if len(maes) else float("nan")
            mean_mape = float(np.mean(mapes)) if len(mapes) else float("nan")

            pair_summary[(actA, actB)] = {
                "MAE": mean_mae,
                "MAPE": mean_mape,
                "pseudo_BCE": metrics["pseudo_BCE"],
                "pseudo_corr": metrics["pseudo_corr"],
                "pseudo_AUC": metrics["pseudo_AUC"],
            }

            # print per-subject lines (optional but useful)
            for k in test_keys:
                if k not in per_trial:
                    continue
                r = per_trial[k]
                print(
                    f"  [actA={actA} -> actB={actB} | {k}] "
                    f"count(score={r['count_score']:.2f}, period={r['count_period_sec']:.3f}s): "
                    f"pred={r['count_pred']:.2f}, gt={r['count_gt']:.2f}, "
                    f"MAE={r['mae']:.2f}, MAPE={r['mape']:.2f}%"
                )

            print(
                f"[PAIR SUMMARY actA={actA} -> actB={actB}] "
                f"Count: MAE={mean_mae:.2f}, MAPE={mean_mape:.2f}% | "
                f"Pseudo: BCE={metrics['pseudo_BCE']:.4f}, corr={metrics['pseudo_corr']:.4f}, AUC(top/bot)={metrics['pseudo_AUC']:.4f}"
            )

            # =========================
            # VISUALIZATION (keep same style)
            # - for each pair, visualize only ONE subject (last in actB)
            # =========================
            if len(test_keys) > 0:
                sub_viz = test_keys[-1]
                one_trials = {sub_viz: test_trials[sub_viz]}
                one_ds = TrialWindowDataset(one_trials, cfg)
                pred_frame, pseudo_frame, X18_raw = infer_one_trial_framewise(model, one_ds, cfg, sub_viz)

                viz_tb_frac = cfg["tb_frac"]
                viz_smooth = cfg["count_smooth_win"]
                viz_alpha = 0.20

                plot_raw_with_top_bottom_shades(
                    act=actB,
                    sub=sub_viz,
                    fs=cfg["fs"],
                    X18_raw=X18_raw,
                    series=pseudo_frame,
                    tb_frac=viz_tb_frac,
                    title_suffix=f"(A={actA} train) raw + pseudo top/bottom spans",
                    smooth_win=viz_smooth,
                    alpha=viz_alpha,
                )

                plot_raw_with_top_bottom_shades(
                    act=actB,
                    sub=sub_viz,
                    fs=cfg["fs"],
                    X18_raw=X18_raw,
                    series=pred_frame,
                    tb_frac=viz_tb_frac,
                    title_suffix=f"(A={actA} train) raw + model pred top/bottom spans",
                    smooth_win=viz_smooth,
                    alpha=viz_alpha,
                )

                plot_latent_speed_boxplot(
                    speed=metrics["speed"],
                    pseudo_w=metrics["pseudo_w"],
                    tb_frac=viz_tb_frac,
                    title=f"actA={actA} -> actB={actB} | latent speed (steady vs transition)",
                )

                preds = [per_trial[k]["count_pred"] for k in test_keys if k in per_trial]
                gts = [per_trial[k]["count_gt"] for k in test_keys if k in per_trial]
                if len(preds) > 0:
                    plot_count_scatter(
                        preds=preds,
                        gts=gts,
                        title=f"actA={actA} -> actB={actB} | count pred vs GT (10 subjects)",
                    )

    # =========================
    # FINAL SUMMARY TABLE-LIKE PRINT
    # =========================
    print("\n" + "=" * 92)
    print("[SUMMARY] Activity-shift ordered pairs (A->B, A!=B)")
    for (a, b), s in pair_summary.items():
        print(
            f"  A={a} -> B={b} | "
            f"MAE={s['MAE']:.2f} | MAPE={s['MAPE']:.2f}% | "
            f"pseudo(BCE={s['pseudo_BCE']:.4f}, corr={s['pseudo_corr']:.4f}, AUC={s['pseudo_AUC']:.4f})"
        )


# =============================================================================
# 14) RUN
# =============================================================================
if __name__ == "__main__":
    run_all_activity_shift(CONFIG)
```
